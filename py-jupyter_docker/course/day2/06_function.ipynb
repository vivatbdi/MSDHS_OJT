## 1. Function คืออะไร

**Function** คือชุดคำสั่งที่ถูกรวมไว้ภายใต้ชื่อเดียว

เมื่อต้องการใช้ชุดคำสั่งนั้น เราสามารถเรียกชื่อ Function แทนการเขียนคำสั่งทั้งหมดซ้ำ

Python มี Function ที่เราเคยใช้แล้ว เช่น

| Function | การใช้งาน |
|---|---|
| `print()` | แสดงผล |
| `len()` | นับจำนวนสมาชิก |
| `sum()` | หาผลรวม |
| `min()` | หาค่าน้อยที่สุด |
| `max()` | หาค่ามากที่สุด |
| `type()` | ตรวจสอบชนิดข้อมูล |
| `int()` | แปลงเป็นจำนวนเต็ม |
| `float()` | แปลงเป็นจำนวนทศนิยม |

Function เหล่านี้เรียกว่า **built-in functions** เพราะ Python เตรียมไว้ให้ใช้งาน

`````python
respondent_counts = [
    120,
    85,
    96,
    74,
]

total_respondents = sum(respondent_counts)

total_respondents
`````

ตัวอย่างด้านบนใช้ `sum()` หาผลรวมของจำนวนผู้ตอบ

เมื่อเรียก Function เราไม่จำเป็นต้องเขียนขั้นตอนการบวกสมาชิกแต่ละค่าเอง เพียงส่งข้อมูลเข้าไปแล้วรับผลลัพธ์กลับมา

## 3. การสร้างและเรียกใช้ Function

Python ใช้คำสั่ง `def` เพื่อสร้าง Function

รูปแบบพื้นฐานคือ

```python
def function_name(parameter):
    statement
    return result
```

องค์ประกอบสำคัญ ได้แก่

- `def` ระบุว่ากำลังสร้าง Function
- `function_name` คือชื่อ Function
- `parameter` คือข้อมูลที่ Function รอรับ
- `:` ปิดบรรทัดประกาศ Function
- คำสั่งภายในต้องเยื้อง
- `return` ส่งผลลัพธ์ออกจาก Function

`````python
def calculate_percentage(part, total):
    percentage = (part / total) * 100
    return percentage
`````

การรัน Cell ที่ประกาศ Function เป็นเพียงการบอก Python ว่า Function นี้มีชื่อและขั้นตอนอย่างไร

Function จะยังไม่คำนวณจนกว่าจะถูกเรียกใช้

`````python
calculate_percentage(18, 60)
`````

`````python
calculate_percentage(25, 80)
`````

ตอนนี้สูตรร้อยละถูกเขียนไว้เพียงครั้งเดียว

เมื่ออยากคำนวณข้อมูลชุดอื่น เราเพียงส่งค่าที่ต้องการเข้าไปใน `calculate_percentage()`

### Function ที่ไม่มี Parameter

Function บางประเภทไม่จำเป็นต้องรับข้อมูลจากภายนอก

`````python
def show_ready_message():
    return "Ready for data work"
`````

`````python
show_ready_message()
`````

Function ที่ไม่มี parameter ยังต้องมีวงเล็บ `()` ทั้งตอนประกาศและเรียกใช้

## 4. Parameter และ Argument

คำว่า **parameter** และ **argument** มีความหมายต่างกัน

- **Parameter** คือชื่อตัวแปรที่กำหนดไว้ตอนสร้าง Function
- **Argument** คือค่าจริงที่ส่งเข้าไปตอนเรียกใช้ Function

`````python
def calculate_percentage(part, total):
    percentage = (part / total) * 100
    return percentage
`````

ในบรรทัด

```python
def calculate_percentage(part, total):
```

- `part` เป็น parameter
- `total` เป็น parameter

เมื่อเรียกใช้

```python
calculate_percentage(54, 180)
```

- `54` เป็น argument ที่ส่งให้ `part`
- `180` เป็น argument ที่ส่งให้ `total`

`````python
calculate_percentage(54, 180)
`````

### Argument ตามตำแหน่ง

โดยปกติ Python จับคู่ argument กับ parameter ตามลำดับ

`````python
def calculate_difference(current_value, previous_value):
    difference = current_value - previous_value
    return difference
`````

`````python
calculate_difference(150, 120)
`````

ในตัวอย่างนี้

- `150` ถูกส่งให้ `current_value`
- `120` ถูกส่งให้ `previous_value`

หากสลับตำแหน่ง ผลลัพธ์จะเปลี่ยน

`````python
calculate_difference(120, 150)
`````

## 5. การใช้ `return`

`return` ใช้ส่งผลลัพธ์ออกจาก Function

ผลลัพธ์ที่ถูกส่งกลับสามารถนำไป

- เก็บในตัวแปร
- คำนวณต่อ
- ใช้ใน Condition
- เพิ่มลงใน List
- ส่งเข้า Function อื่น
- นำไปสร้างข้อความสรุป

`````python
def calculate_average_score(
    income_score,
    housing_score,
    mental_health_score,
):
    average_score = (
        income_score
        + housing_score
        + mental_health_score
    ) / 3

    return average_score
`````

`````python
average_need_score = calculate_average_score(
    4.2,
    3.8,
    4.5,
)

average_need_score
`````

`````python
summary = (
    f"คะแนนเฉลี่ยความต้องการความช่วยเหลือ"
    f"เท่ากับ {average_need_score:.2f} คะแนน"
)

summary
`````

### ความแตกต่างระหว่าง `print()` กับ `return`

- `print()` ใช้แสดงผลให้ผู้ใช้อ่าน
- `return` ใช้ส่งค่ากลับเพื่อนำไปใช้ต่อ

Function ที่มีเพียง `print()` จะแสดงข้อความได้ แต่ไม่มีค่าผลลัพธ์ส่งกลับ

`````python
def show_percentage(part, total):
    percentage = (part / total) * 100
    print(f"Percentage: {percentage:.2f}%")
`````

`````python
result = show_percentage(54, 180)

result
`````

ค่าของ `result` คือ `None` เพราะ Function ไม่มี `return`

`````python
def calculate_percentage(part, total):
    percentage = (part / total) * 100
    return percentage
`````

`````python
result = calculate_percentage(54, 180)

result
`````

แนวทางสำหรับงานข้อมูลคือ

- ใช้ `return` เมื่อผลลัพธ์ต้องนำไปใช้ต่อ
- ใช้ `print()` เมื่อต้องการแสดงข้อความระหว่างการทำงาน
- Function เดียวสามารถมีทั้ง `print()` และ `return` ได้ แต่ควรใช้เท่าที่จำเป็น

## 6. Function ที่มี Parameter หลายตัว

Function สามารถรับข้อมูลได้หลายค่า

ตัวอย่างเช่น การตรวจสอบว่าข้อมูลพร้อมนำเข้ารายงานหรือไม่ อาจต้องใช้

- สถานะรายการ
- การให้ความยินยอม
- สถานะข้อมูลซ้ำ

`````python
def check_report_eligibility(
    status,
    consent_given,
    is_duplicate,
):
    is_eligible = (
        status == "Completed"
        and consent_given
        and not is_duplicate
    )

    return is_eligible
`````

`````python
check_report_eligibility(
    "Completed",
    True,
    False,
)
`````

`````python
check_report_eligibility(
    "Draft",
    True,
    False,
)
`````

### การคืนค่ามากกว่าหนึ่งค่า

Function สามารถคืนผลลัพธ์ได้มากกว่าหนึ่งค่า

`````python
def summarize_scores(scores):
    total_score = sum(scores)
    average_score = total_score / len(scores)

    return total_score, average_score
`````

`````python
scores = [
    4.2,
    3.8,
    4.5,
]

total_score, average_score = summarize_scores(
    scores
)

print(total_score)
print(average_score)
`````

Function `summarize_scores()` คืนผลลัพธ์สองค่า ได้แก่

1. ผลรวม
2. ค่าเฉลี่ย

เราจึงใช้ตัวแปรสองตัวรับผลลัพธ์ตามลำดับ

## 7. Default Argument และ Keyword Argument

เราสามารถกำหนดค่าเริ่มต้นให้ parameter ได้ เรียกว่า **default argument**

เหมาะเมื่อ Function มีค่าที่ใช้บ่อย และผู้เรียกไม่จำเป็นต้องส่งค่าทุกครั้ง

`````python
def calculate_percentage(
    part,
    total,
    decimal_places=2,
):
    percentage = (part / total) * 100
    return round(percentage, decimal_places)
`````

`````python
calculate_percentage(25, 80)
`````

`````python
calculate_percentage(25, 80, 1)
`````

หากไม่ส่ง `decimal_places` Function จะใช้ค่าเริ่มต้นเป็น `2`

หากส่งค่าใหม่ เช่น `1` Function จะใช้ค่าที่ส่งเข้าไปแทน

### Keyword Argument

เราสามารถระบุชื่อ parameter ตอนเรียกใช้ Function ได้

`````python
calculate_percentage(
    part=25,
    total=80,
    decimal_places=1,
)
`````

Keyword argument มีประโยชน์เมื่อ

- Function มี parameter หลายตัว
- ต้องการให้โค้ดอ่านง่าย
- ต้องการส่งค่าโดยไม่พึ่งตำแหน่งเพียงอย่างเดียว

### ลำดับของ Parameter

Parameter ที่มีค่าเริ่มต้นควรอยู่หลัง parameter ที่ไม่มีค่าเริ่มต้น

รูปแบบที่ถูกต้องคือ

```python
def function_name(required_value, optional_value=0):
    ...
```

## 8. Function สำหรับคำนวณตัวชี้วัด

ในงานวิเคราะห์ข้อมูล เรามักต้องคำนวณตัวชี้วัดเดิมซ้ำ เช่น

- ร้อยละ
- คะแนนเฉลี่ย
- อัตราการเปลี่ยนแปลง
- สัดส่วนข้อมูลครบถ้วน
- จำนวนหรือสัดส่วนผู้เข้าเกณฑ์

การสร้าง Function ช่วยให้สูตรมีชื่อ อ่านง่าย และแก้ไขได้จากจุดเดียว

### ตัวอย่าง: Function คำนวณร้อยละ

ก่อนหารควรตรวจสอบว่า `total` ไม่เท่ากับศูนย์

`````python
def calculate_percentage(part, total):
    if total == 0:
        return None

    percentage = (part / total) * 100
    return percentage
`````

`````python
calculate_percentage(54, 180)
`````

`````python
calculate_percentage(0, 0)
`````

เมื่อ `total` เป็น `0` Function คืนค่า `None` เพื่อหลีกเลี่ยง `ZeroDivisionError`

การเลือกว่าจะคืน `None`, `0` หรือข้อความแจ้งเตือนควรขึ้นอยู่กับกฎของงานและความหมายของข้อมูล

### ตัวอย่าง: Function คำนวณอัตราการเปลี่ยนแปลง

สูตรคือ

```text
ค่าปัจจุบัน - ค่าก่อนหน้า
--------------------------
       ค่าก่อนหน้า
```

`````python
def calculate_growth_rate(
    current_value,
    previous_value,
):
    if previous_value == 0:
        return None

    growth_rate = (
        current_value - previous_value
    ) / previous_value

    return growth_rate
`````

`````python
calculate_growth_rate(
    current_value=150,
    previous_value=120,
)
`````

### ตัวอย่าง: Function สร้างข้อความสรุป

`````python
def create_area_summary(
    area,
    respondent_count,
    urgent_count,
):
    urgent_percentage = (
        urgent_count / respondent_count
    ) * 100

    summary = (
        f"พื้นที่ {area} มีผู้ตอบ "
        f"{respondent_count} คน และมีผู้ที่"
        f"ต้องการความช่วยเหลือเร่งด่วน "
        f"{urgent_percentage:.2f}%"
    )

    return summary
`````

`````python
create_area_summary(
    area="Bangkok",
    respondent_count=60,
    urgent_count=18,
)
`````

Function สามารถคืนค่าที่เป็นตัวเลข ข้อความ Boolean, List, Dictionary หรือชนิดข้อมูลอื่นได้ ขึ้นอยู่กับผลลัพธ์ที่ต้องการ

## 9. Function ที่มี Condition

Function สามารถมี `if`, `elif` และ `else` อยู่ภายในได้

เหมาะกับกฎที่ต้องใช้ซ้ำ เช่น

- จัดกลุ่มคะแนน
- ตรวจสอบสถานะ
- ระบุระดับการติดตาม
- ตรวจสอบข้อมูลผิดปกติ
- กำหนดว่ารายการควรถูกนำไปใช้ต่อหรือไม่

### ตัวอย่าง: จัดกลุ่มระดับความเปราะบาง

`````python
def classify_vulnerability(
    vulnerability_score,
):
    if (
        vulnerability_score < 0
        or vulnerability_score > 5
    ):
        group = "Invalid Score"
    elif vulnerability_score >= 4.0:
        group = "High"
    elif vulnerability_score >= 2.5:
        group = "Medium"
    else:
        group = "Low"

    return group
`````

`````python
classify_vulnerability(4.3)
`````

`````python
classify_vulnerability(3.1)
`````

`````python
classify_vulnerability(6.0)
`````

Logic การจัดกลุ่มถูกเขียนไว้ใน Function เพียงจุดเดียว

หากเกณฑ์เปลี่ยน เราสามารถแก้ไข Function แล้วใช้กฎใหม่กับข้อมูลทุกส่วนที่เรียก Function นี้

### ตัวอย่าง: ตรวจสอบสถานะแบบสำรวจ

`````python
def check_survey_status(status):
    valid_statuses = [
        "Completed",
        "Verified",
    ]

    if status in valid_statuses:
        result = "Valid"
    else:
        result = "Need Review"

    return result
`````

### ตัวอย่าง: กำหนดระดับการติดตาม

`````python
def determine_follow_up_level(
    has_urgent_need,
    vulnerability_score,
):
    if (
        has_urgent_need
        and vulnerability_score >= 4.0
    ):
        level = "Immediate Follow-up"
    elif has_urgent_need:
        level = "Follow-up"
    else:
        level = "Routine"

    return level
`````

## 10. Function ร่วมกับ Loop

Function มีประโยชน์มากเมื่อนำไปใช้กับ Loop

แนวคิดคือ

1. Function ดูแลกฎหรือการคำนวณหนึ่งอย่าง
2. Loop ดูแลการเรียก Function ซ้ำกับข้อมูลหลายรายการ
3. ผลลัพธ์ถูกเก็บใน List หรือ Dictionary

### ตัวอย่าง: จัดกลุ่มคะแนนหลายรายการ

`````python
vulnerability_scores = [
    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]

vulnerability_groups = []

for score in vulnerability_scores:
    group = classify_vulnerability(score)
    vulnerability_groups.append(group)

vulnerability_groups
`````

แทนที่จะเขียนเงื่อนไขทั้งหมดภายใน Loop เรานำกฎไปเก็บใน `classify_vulnerability()`

ทำให้หน้าที่ของแต่ละส่วนชัดเจน

- Function จัดกลุ่มคะแนน
- Loop ประมวลผลคะแนนหลายค่า

### ตัวอย่าง: ตรวจสอบหลายสถานะ

`````python
survey_statuses = [
    "Completed",
    "Draft",
    "Verified",
    "Incomplete",
]

status_results = []

for status in survey_statuses:
    result = check_survey_status(status)
    status_results.append(result)

status_results
`````

### ตัวอย่าง: สร้าง Dictionary ผลลัพธ์

`````python
scores_by_respondent = {
    "R001": 4.3,
    "R002": 3.1,
    "R003": 2.2,
    "R004": 4.8,
}

groups_by_respondent = {}

for respondent_id, score in (
    scores_by_respondent.items()
):
    group = classify_vulnerability(score)

    groups_by_respondent[
        respondent_id
    ] = group

groups_by_respondent
`````

การใช้ Dictionary ช่วยรักษาความสัมพันธ์ระหว่างรหัสผู้ตอบกับผลการจัดกลุ่ม

`````python
def determine_follow_up_level(
    has_urgent_need,
    vulnerability_score,
):
    if (
        has_urgent_need
        and vulnerability_score >= 4.0
    ):
        return "Immediate Follow-up"

    if has_urgent_need:
        return "Follow-up"

    return "Routine"
`````

`````python
print(
    determine_follow_up_level(
        True,
        4.5,
    )
)

print(
    determine_follow_up_level(
        True,
        3.6,
    )
)

print(
    determine_follow_up_level(
        False,
        4.8,
    )
)
`````

## แบบฝึกหัดที่ 8: Function ทำความสะอาดชื่อคอลัมน์

สร้าง Function ชื่อ `clean_column_name`

Function ต้อง

1. ลบช่องว่างหัวท้าย
2. เปลี่ยนข้อความเป็นตัวพิมพ์เล็ก
3. เปลี่ยนช่องว่างระหว่างคำเป็น underscore

> ### เฉลยแบบฝึกหัดที่ 8
>
> ใช้ `.strip()`, `.lower()` และ `.replace()` ตามลำดับ

`````python
def clean_column_name(column_name):
    cleaned_name = column_name.strip()
    cleaned_name = cleaned_name.lower()
    cleaned_name = cleaned_name.replace(
        " ",
        "_",
    )

    return cleaned_name
`````

> ### ตัวอย่างเฉลยแบบฝึกหัดที่ 10
>
> **งาน:** ตรวจสอบว่าข้อมูลผู้ตอบพร้อมนำไปวิเคราะห์หรือไม่
>
> **Input**
>
> - `respondent_id`
> - `status`
> - `consent_given`
> - `is_duplicate`
>
> **Output:** ค่า Boolean
>
> **เงื่อนไข**
>
> - รหัสผู้ตอบต้องไม่ว่าง
> - สถานะต้องเป็น `"Completed"`
> - ต้องได้รับความยินยอม
> - ต้องไม่ใช่ข้อมูลซ้ำ
>
> **ชื่อ Function:** `is_record_ready`
>
> คำตอบของผู้เรียนอาจแตกต่างกันได้ตามลักษณะงาน หาก Function มีชื่อ หน้าที่ Input และ Output ชัดเจน

`````python
def is_record_ready(
    respondent_id,
    status,
    consent_given,
    is_duplicate,
):
    is_ready = (
        respondent_id != ""
        and status == "Completed"
        and consent_given
        and not is_duplicate
    )

    return is_ready
`````

`````python
print(
    is_record_ready(
        respondent_id="R001",
        status="Completed",
        consent_given=True,
        is_duplicate=False,
    )
)

print(
    is_record_ready(
        respondent_id="",
        status="Completed",
        consent_given=True,
        is_duplicate=False,
    )
)
`````
