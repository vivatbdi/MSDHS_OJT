## 9) ตัวอย่างงานจริง: เลือกไฟล์ .csv ทั้งหมดในโฟลเดอร์

Data Engineer มักต้อง:
1) เข้าไปดูไฟล์ทั้งหมดในโฟลเดอร์
2) เลือกเฉพาะไฟล์ที่นามสกุลตรง เช่น `.csv`
3) สร้าง full path เพื่อส่งต่อไปอ่านด้วย `csv` หรือ `pandas`

เราจะสร้างโฟลเดอร์ `data` และไฟล์จำลอง 2-3 ไฟล์ก่อน

````python
data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

# สร้างไฟล์จำลอง
sample_files = ["sales_2025_01.csv", "sales_2025_02.csv", "readme.txt"]
for fn in sample_files:
    fp = os.path.join(data_dir, fn)
    with open(fp, "w", encoding="utf-8") as f:
        f.write("dummy")

os.listdir(data_dir)
````

คัดเฉพาะไฟล์ที่ลงท้ายด้วย `.csv`

````python
all_files = os.listdir(data_dir)
csv_files = [fn for fn in all_files if fn.lower().endswith(".csv")]

csv_files
````

สร้าง full path ของไฟล์ CSV เพื่อเตรียมส่งต่อให้ขั้นตอนอ่านข้อมูล

````python
csv_paths = [os.path.join(data_dir, fn) for fn in csv_files]
csv_paths
````

## 10) แบบฝึกหัดสั้น ๆ

ให้ทำตามนี้:
1) สร้างโฟลเดอร์ชื่อ `raw_data`
2) สร้างไฟล์จำลองชื่อ `a.csv`, `b.csv`, `note.txt` ในโฟลเดอร์นั้น
3) ใช้ `os.listdir()` + เงื่อนไข เพื่อเลือกเฉพาะไฟล์ `.csv`
4) สร้าง full path ของไฟล์ `.csv` ทุกไฟล์

✅ ทำใน cell ด้านล่าง

````python
# 1) สร้างโฟลเดอร์
raw_dir = "raw_data"
os.makedirs(raw_dir, exist_ok=True)

# 2) สร้างไฟล์จำลอง
for fn in ["a.csv", "b.csv", "note.txt"]:
    fp = os.path.join(raw_dir, fn)
    with open(fp, "w", encoding="utf-8") as f:
        f.write("dummy")

# 3) เลือกเฉพาะ .csv
files = os.listdir(raw_dir)
only_csv = [fn for fn in files if fn.endswith(".csv")]

# 4) สร้าง full path
only_csv_paths = [os.path.join(raw_dir, fn) for fn in only_csv]

print("csv files:", only_csv)
print("csv paths:", only_csv_paths)
````
