## 1. Loop คืออะไร

**Loop** คือการสั่งให้ Python ทำงานซ้ำกับข้อมูลแต่ละรายการ หรือทำซ้ำตามจำนวนรอบที่กำหนด

กำหนดข้อมูลยอดขายดังนี้

```python
sales_list = [12000, 18000, 9500, 22000]
```

หากต้องการแสดงค่าทุกค่าโดยไม่ใช้ Loop เราอาจเขียนว่า

```python
print(sales_list[0])
print(sales_list[1])
print(sales_list[2])
print(sales_list[3])
```

วิธีนี้ไม่เหมาะเมื่อข้อมูลมีจำนวนมาก หรือจำนวนข้อมูลเปลี่ยนแปลงอยู่เสมอ

````python
sales_list = [
    12000,
    18000,
    9500,
    22000,
]

print(sales_list[0])
print(sales_list[1])
print(sales_list[2])
print(sales_list[3])
````

เมื่อใช้ Loop เราสามารถเขียนคำสั่งที่ต้องการทำซ้ำเพียงครั้งเดียว

````python
sales_list = [
    12000,
    18000,
    9500,
    22000,
]

for sales in sales_list:
    print(sales)
````

Python จะนำค่าจาก `sales_list` ออกมาทีละค่า แล้วเก็บไว้ในตัวแปรชั่วคราวชื่อ `sales`

ลำดับการทำงานคือ

1. `sales` มีค่าเป็น `12000`
2. รัน `print(sales)`
3. `sales` เปลี่ยนเป็น `18000`
4. รัน `print(sales)` อีกครั้ง
5. ทำซ้ำจนถึงสมาชิกตัวสุดท้าย

## 2. การใช้ `for` Loop กับ List

รูปแบบพื้นฐานของ `for` Loop คือ

```python
for item in list_name:
    statement
```

อ่านได้ว่า

> สำหรับแต่ละ `item` ที่อยู่ใน `list_name` ให้ทำคำสั่งที่เยื้องอยู่ภายใต้ Loop

องค์ประกอบสำคัญคือ

- ใช้คำว่า `for`
- ตามด้วยชื่อตัวแปรชั่วคราว
- ตามด้วยคำว่า `in`
- ตามด้วย List ที่ต้องการวนซ้ำ
- ปิดบรรทัดด้วย colon `:`
- คำสั่งที่ต้องทำซ้ำต้องเยื้องเข้าไป

````python
survey_areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
    "Songkhla",
]

for area in survey_areas:
    print(area)
````

ควรตั้งชื่อตัวแปรชั่วคราวให้สื่อความหมายของสมาชิกแต่ละรายการ

ตัวอย่างที่อ่านง่าย

```python
for area in survey_areas:
    print(area)
```

ตัวอย่างที่อ่านยากกว่า

```python
for x in survey_areas:
    print(x)
```

แม้ทั้งสองแบบทำงานได้เหมือนกัน แต่ชื่อ `area` ช่วยให้เข้าใจทันทีว่า Loop กำลังทำงานกับข้อมูลประเภทใด

### ชื่อตัวแปรแบบเอกพจน์และพหูพจน์

แนวทางที่ช่วยให้โค้ดอ่านง่ายคือ

- List ใช้ชื่อพหูพจน์ เช่น `areas`
- สมาชิกหนึ่งรายการใช้ชื่อเอกพจน์ เช่น `area`

ตัวอย่าง

```python
for area in areas:
    print(area)
```

ตัวอย่างอื่น ได้แก่

```python
for score in scores:
for file_name in file_names:
for status in statuses:
for respondent in respondents:
```

````python
support_needs = [
    "Income",
    "Housing",
    "Mental health",
]

for support_need in support_needs:
    print(f"Support need: {support_need}")
````

## 3. การทำงานภายใน Loop

ภายใน Loop สามารถมีคำสั่งได้มากกว่าหนึ่งบรรทัด

ทุกคำสั่งที่เยื้องอยู่ในระดับเดียวกันจะถูกทำซ้ำในแต่ละรอบ

````python
survey_areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

for area in survey_areas:
    print("กำลังประมวลผลพื้นที่")
    print(area)
    print("---")
````

ผลลัพธ์ในแต่ละรอบประกอบด้วย 3 คำสั่ง

1. แสดงข้อความ `"กำลังประมวลผลพื้นที่"`
2. แสดงชื่อพื้นที่
3. แสดงเส้นแบ่ง

เมื่อ Loop จบรอบหนึ่ง Python จะนำพื้นที่ถัดไปมาประมวลผล

### คำสั่งภายในและภายนอก Loop

คำสั่งที่เยื้องอยู่ภายใต้ `for` จะทำซ้ำทุกครั้ง

คำสั่งที่ไม่เยื้องจะทำงานหลังจาก Loop จบแล้ว

````python
survey_areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

for area in survey_areas:
    print("Inside loop:", area)

print("Loop completed")
````

ข้อความ `"Loop completed"` แสดงเพียงครั้งเดียว เพราะอยู่นอก Loop

### การคำนวณภายใน Loop

เราสามารถนำสมาชิกแต่ละค่าไปคำนวณได้

สมมติว่ามีจำนวนผู้ตอบแบบสำรวจในหลายพื้นที่ และต้องการเพิ่มค่าประมาณการอีก 10%

````python
respondent_counts = [
    120,
    85,
    96,
    74,
]

for respondent_count in respondent_counts:
    estimated_count = respondent_count * 1.10
    print(estimated_count)
````

ตัวอย่างด้านบนแสดงผลลัพธ์ออกมาทันที แต่ยังไม่ได้เก็บผลลัพธ์ไว้ใช้ต่อ

หากต้องการนำผลลัพธ์ไปวิเคราะห์หรือสร้างรายงานภายหลัง ควรเก็บผลลัพธ์ไว้ใน List ใหม่

## 4. การสะสมผลลัพธ์จาก Loop

การสะสมผลลัพธ์จาก Loop ที่ใช้บ่อยมี 2 รูปแบบ

1. เก็บผลลัพธ์แต่ละรายการใน List ใหม่
2. รวมค่าทั้งหมดไว้ในตัวแปรสะสม

ควรเลือกวิธีให้ตรงกับผลลัพธ์ที่ต้องการ

### 4.1 การสร้าง List ผลลัพธ์

ขั้นตอนพื้นฐานคือ

1. สร้าง List ว่าง
2. วนซ้ำข้อมูลต้นทาง
3. คำนวณหรือแปลงค่า
4. ใช้ `.append()` เพิ่มผลลัพธ์เข้า List

````python
respondent_counts = [
    120,
    85,
    96,
    74,
]

estimated_counts = []

for respondent_count in respondent_counts:
    estimated_count = respondent_count * 1.10
    estimated_counts.append(estimated_count)

estimated_counts
````

จากตัวอย่าง

- `estimated_counts = []` สร้าง List ว่าง
- `estimated_count` เก็บผลลัพธ์ของแต่ละรอบ
- `.append()` เพิ่มผลลัพธ์เข้า List
- จำนวนสมาชิกใน `estimated_counts` เท่ากับจำนวนสมาชิกใน `respondent_counts`

### ตัวอย่าง: คำนวณร้อยละหลายรายการ

สูตรร้อยละคือ

```text
จำนวนที่สนใจ ÷ จำนวนทั้งหมด × 100
```

````python
urgent_counts = [
    18,
    25,
    12,
    30,
]

total_counts = [
    60,
    80,
    50,
    100,
]

urgent_percentages = []

for urgent_count, total_count in zip(
    urgent_counts,
    total_counts,
):
    percentage = (
        urgent_count / total_count
    ) * 100

    urgent_percentages.append(percentage)

urgent_percentages
````

ตัวอย่างนี้ใช้ `zip()` เพื่อจับคู่จำนวนเร่งด่วนกับจำนวนทั้งหมดในตำแหน่งเดียวกัน

รายละเอียดของ `zip()` จะอธิบายในหัวข้อถัดไป

### 4.2 การรวมค่าด้วยตัวแปรสะสม

หากต้องการผลรวมเพียงค่าเดียว สามารถสร้างตัวแปรสะสม เช่น `total`

ขั้นตอนคือ

1. กำหนดค่าเริ่มต้นเป็น `0`
2. วนซ้ำข้อมูล
3. เพิ่มค่าแต่ละรายการเข้าในตัวแปรสะสม

````python
respondent_counts = [
    120,
    85,
    96,
    74,
]

total_respondents = 0

for respondent_count in respondent_counts:
    total_respondents = (
        total_respondents
        + respondent_count
    )

total_respondents
````

สามารถเขียนคำสั่งเพิ่มค่าแบบย่อด้วย `+=`

````python
respondent_counts = [
    120,
    85,
    96,
    74,
]

total_respondents = 0

for respondent_count in respondent_counts:
    total_respondents += respondent_count

total_respondents
````

คำสั่ง

```python
total_respondents += respondent_count
```

มีความหมายเหมือนกับ

```python
total_respondents = (
    total_respondents
    + respondent_count
)
```

ในงานจริง หากต้องการเพียงผลรวมของ List สามารถใช้ `sum()` ได้โดยตรง

```python
sum(respondent_counts)
```

ตัวอย่าง Loop นี้มีไว้เพื่ออธิบายแนวคิดของตัวแปรสะสม

````python
sum(respondent_counts)
````

### การนับจำนวนรายการที่เข้าเงื่อนไข

ตัวแปรสะสมสามารถใช้เป็นตัวนับได้

````python
vulnerability_scores = [
    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]

high_risk_count = 0

for score in vulnerability_scores:
    if score >= 4.0:
        high_risk_count += 1

high_risk_count
````

ทุกครั้งที่พบคะแนนตั้งแต่ `4.0` ขึ้นไป ตัวแปร `high_risk_count` จะเพิ่มขึ้น 1

## 5. การใช้ Loop ร่วมกับ Condition

Loop และ Condition มักใช้ร่วมกันเมื่อต้องตรวจสอบหรือจัดกลุ่มข้อมูลหลายรายการ

รูปแบบพื้นฐานคือ

```python
for item in items:
    if condition:
        statement
    else:
        another_statement
```

ในแต่ละรอบ Python จะ

1. นำข้อมูลหนึ่งรายการออกมา
2. ตรวจสอบเงื่อนไข
3. เลือกผลลัพธ์ตามเงื่อนไข
4. ทำซ้ำกับรายการถัดไป

### ตัวอย่าง: จัดกลุ่มคะแนนความเปราะบาง

````python
vulnerability_scores = [
    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]

vulnerability_groups = []

for score in vulnerability_scores:
    if score >= 4.0:
        group = "High"
    elif score >= 2.5:
        group = "Medium"
    else:
        group = "Low"

    vulnerability_groups.append(group)

vulnerability_groups
````

Logic การจัดกลุ่มถูกเขียนเพียงครั้งเดียว แต่ Python นำไปใช้กับคะแนนทุกค่าใน List

แนวคิดนี้คล้ายกับการเขียนสูตร `IFS` ใน Excel แล้วลากสูตรลงหลายแถว

### ตัวอย่าง: ตรวจสอบหลายสถานะ

````python
survey_statuses = [
    "Completed",
    "Draft",
    "Verified",
    "Incomplete",
    "Completed",
]

valid_statuses = [
    "Completed",
    "Verified",
]

status_results = []

for status in survey_statuses:
    if status in valid_statuses:
        result = "Valid"
    else:
        result = "Need Review"

    status_results.append(result)

status_results
````

ในแต่ละรอบ Python ตรวจสอบว่าสถานะอยู่ใน `valid_statuses` หรือไม่ แล้วเก็บผลลัพธ์ไว้ใน `status_results`

### ตัวอย่าง: เก็บเฉพาะข้อมูลที่เข้าเงื่อนไข

Loop ไม่จำเป็นต้องสร้างผลลัพธ์ให้ทุกค่าเสมอไป

เราสามารถเพิ่มเฉพาะค่าที่เข้าเงื่อนไขลงใน List ใหม่

````python
vulnerability_scores = [
    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]

high_risk_scores = []

for score in vulnerability_scores:
    if score >= 4.0:
        high_risk_scores.append(score)

high_risk_scores
````

List `high_risk_scores` มีเฉพาะคะแนนที่มากกว่าหรือเท่ากับ `4.0`

### ตรวจสอบค่าผิดปกติก่อนจัดกลุ่ม

หากคะแนนควรอยู่ระหว่าง 0 ถึง 5 ควรตรวจสอบช่วงค่าก่อนจัดกลุ่ม

````python
vulnerability_scores = [
    4.3,
    -1.0,
    3.1,
    6.0,
    2.2,
]

score_results = []

for score in vulnerability_scores:
    if score < 0 or score > 5:
        result = "Invalid Score"
    elif score >= 4.0:
        result = "High"
    elif score >= 2.5:
        result = "Medium"
    else:
        result = "Low"

    score_results.append(result)

score_results
````

การตรวจสอบค่าที่อยู่นอกช่วงก่อน ช่วยป้องกันไม่ให้ค่า `6.0` ถูกจัดเป็น `"High"` ทั้งที่เป็นข้อมูลผิดปกติ

## 6. การใช้ `range()`

`range()` ใช้สร้างลำดับจำนวนเต็มสำหรับการวนซ้ำ

เหมาะเมื่อ

- ต้องการทำซ้ำตามจำนวนรอบ
- ต้องการสร้างเลขลำดับ
- ต้องการสร้างหมายเลขเดือนหรือรอบ
- ต้องการทำงานตามตำแหน่งของข้อมูล

รูปแบบที่ใช้บ่อย ได้แก่

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

ค่า `stop` จะไม่รวมอยู่ในผลลัพธ์

````python
for number in range(5):
    print(number)
````

ผลลัพธ์คือ

```text
0
1
2
3
4
```

ไม่รวมเลข `5`

### `range(start, stop)`

กำหนดทั้งค่าเริ่มต้นและค่าสิ้นสุด

````python
for month in range(1, 13):
    print(month)
````

`range(1, 13)` ให้เลขตั้งแต่ 1 ถึง 12 เพราะหยุดก่อน 13

### `range(start, stop, step)`

ค่า `step` กำหนดระยะการเพิ่มหรือลดในแต่ละรอบ

````python
for number in range(0, 11, 2):
    print(number)
````

ผลลัพธ์คือเลขคู่ตั้งแต่ 0 ถึง 10

```text
0
2
4
6
8
10
```

## 7. การใช้ `enumerate()`

`enumerate()` ใช้เมื่อต้องการทั้ง

- ลำดับหรือ index
- ค่าของสมาชิก

รูปแบบคือ

```python
for index, item in enumerate(items):
    statement
```

````python
survey_areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

for index, area in enumerate(survey_areas):
    print(index, area)
````

โดยปกติ `enumerate()` เริ่มนับจาก `0`

หากต้องการแสดงลำดับสำหรับผู้อ่าน มักกำหนด `start=1`

````python
survey_areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

for row_number, area in enumerate(
    survey_areas,
    start=1,
):
    print(f"{row_number}. {area}")
````

### ตัวอย่าง: สร้างรหัสลำดับ

ควรใช้ `enumerate()` แทนการสร้างตัวแปรนับเอง เมื่อสิ่งที่ต้องการคือหมายเลขลำดับของสมาชิกใน List

## 8. การใช้ `zip()`

`zip()` ใช้จับคู่ข้อมูลจากหลาย List ตามตำแหน่งเดียวกัน

ตัวอย่างเช่น

```python
areas = ["Bangkok", "Chiang Mai"]
respondent_counts = [120, 85]
```

การใช้ `zip()` จะจับคู่เป็น

- `"Bangkok"` กับ `120`
- `"Chiang Mai"` กับ `85`

รูปแบบคือ

```python
for value_1, value_2 in zip(
    list_1,
    list_2,
):
    statement
```

````python
areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

respondent_counts = [
    120,
    85,
    96,
]

for area, respondent_count in zip(
    areas,
    respondent_counts,
):
    print(area, respondent_count)
````

### ตัวอย่าง: สร้างข้อความสรุป

````python
areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

respondent_counts = [
    120,
    85,
    96,
]

summary_messages = []

for area, respondent_count in zip(
    areas,
    respondent_counts,
):
    message = (
        f"{area}: respondents = "
        f"{respondent_count}"
    )

    summary_messages.append(message)

summary_messages
````

### การใช้ `zip()` กับมากกว่าสอง List

สามารถจับคู่ข้อมูลจาก 3 List หรือมากกว่าได้

````python
areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

urgent_counts = [
    18,
    25,
    12,
]

total_counts = [
    60,
    80,
    50,
]

for area, urgent_count, total_count in zip(
    areas,
    urgent_counts,
    total_counts,
):
    percentage = (
        urgent_count / total_count
    ) * 100

    print(
        f"{area}: "
        f"{percentage:.2f}%"
    )
````

### ข้อควรระวังของ `zip()`

`zip()` จะหยุดเมื่อ List ที่สั้นที่สุดหมดสมาชิก

ตัวอย่างต่อไปนี้มีชื่อพื้นที่ 3 ค่า แต่มีจำนวนผู้ตอบเพียง 2 ค่า

````python
areas = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

respondent_counts = [
    120,
    85,
]

for area, respondent_count in zip(
    areas,
    respondent_counts,
):
    print(area, respondent_count)
````

ผลลัพธ์มีเพียง 2 คู่ และ `"Khon Kaen"` จะไม่ถูกนำมาประมวลผล

ดังนั้น ก่อนใช้ `zip()` ควรตรวจสอบว่า List ที่ต้องจับคู่มีจำนวนสมาชิกสอดคล้องกัน

````python
print("จำนวนพื้นที่:", len(areas))

print(
    "จำนวนชุดข้อมูลผู้ตอบ:",
    len(respondent_counts),
)
````

## 9. การใช้ Loop กับ Dictionary

Dictionary เก็บข้อมูลเป็นคู่ `key: value`

รูปแบบคือ

```python
for key, value in dictionary.items():
    statement
```

````python
respondents_by_area = {
    "Bangkok": 120,
    "Chiang Mai": 85,
    "Khon Kaen": 96,
}

for area, respondent_count in (
    respondents_by_area.items()
):
    print(area, respondent_count)
````

จากตัวอย่าง

- `respondent_count` รับค่า value

### การวนซ้ำเฉพาะ key

### สร้าง Dictionary ใหม่จาก Loop

เราสามารถสร้าง Dictionary ว่าง แล้วเพิ่ม key-value ในแต่ละรอบ

````python
respondents_by_area = {
    "Bangkok": 120,
    "Chiang Mai": 85,
    "Khon Kaen": 96,
    "Songkhla": 74,
}

area_sample_status = {}

for area, respondent_count in (
    respondents_by_area.items()
):
    if respondent_count >= 100:
        status = "Sufficient Sample"
    else:
        status = "Need More Data"

    area_sample_status[area] = status

area_sample_status
````

3. กำหนดสถานะ
4. ใช้ชื่อพื้นที่เป็น key ของ Dictionary ใหม่
5. ใช้สถานะเป็น value

## 10. การใช้ `break` และ `continue`

ภายใน Loop เราสามารถควบคุมการทำงานเพิ่มเติมด้วย

- `break` — หยุด Loop ทันที
- `continue` — ข้ามรอบปัจจุบันและไปยังรอบถัดไป

ควรใช้เมื่อช่วยให้ Logic ชัดเจนขึ้น และไม่ควรใช้มากเกินไปจนทำให้ลำดับการทำงานเข้าใจยาก

### 10.1 การใช้ `break`

ตัวอย่าง: หยุดค้นหาเมื่อพบข้อมูลที่ต้องการ

````python
respondent_ids = [
    "R001",
    "R002",
    "R003",
    "R004",
]

target_id = "R003"

for respondent_id in respondent_ids:
    print(f"Checking {respondent_id}")

    if respondent_id == target_id:
        print("Target found")
        break
````

เมื่อพบ `"R003"` คำสั่ง `break` จะหยุด Loop จึงไม่ตรวจสอบ `"R004"`

### 10.2 การใช้ `continue`

ตัวอย่าง: ข้ามข้อมูลที่ไม่พร้อมใช้งาน

````python
scores = [
    4.2,
    None,
    3.8,
    None,
    4.5,
]

valid_scores = []

for score in scores:
    if score is None:
        continue

    valid_scores.append(score)

valid_scores
````

เมื่อ `score` เป็น `None` คำสั่ง `continue` จะข้ามคำสั่งที่เหลือในรอบนั้น และไปยังค่าถัดไป

### ทางเลือกที่ไม่ใช้ `continue`

Logic เดียวกันสามารถเขียนด้วย Condition ปกติได้

````python
scores = [
    4.2,
    None,
    3.8,
    None,
    4.5,
]

valid_scores = []

for score in scores:
    if score is not None:
        valid_scores.append(score)

valid_scores
````

ทั้งสองรูปแบบให้ผลลัพธ์เหมือนกัน

ควรเลือกรูปแบบที่อ่านง่ายและสื่อ Logic ได้ชัดเจนที่สุด

## แบบฝึกหัดที่ 3: คำนวณร้อยละหลายพื้นที่ด้วย `zip()`

กำหนดข้อมูลดังนี้

```python
urgent_counts = [18, 25, 12, 30]
total_counts = [60, 80, 50, 100]
```

ให้ทำงานต่อไปนี้

1. สร้าง List ว่างชื่อ `urgent_percentages`
2. ใช้ `zip()` จับคู่ข้อมูลจากทั้งสอง List
3. คำนวณร้อยละด้วยสูตร

```text
จำนวนเร่งด่วน ÷ จำนวนทั้งหมด × 100
```

4. เก็บผลลัพธ์แต่ละค่าใน `urgent_percentages`

````python
urgent_counts = [
    18,
    25,
    12,
    30,
]

total_counts = [
    60,
    80,
    50,
    100,
]
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 3
>
> `zip()` จับคู่สมาชิกที่อยู่ในตำแหน่งเดียวกันของทั้งสอง List
>
> - `18` จับคู่กับ `60`
> - `25` จับคู่กับ `80`
> - `12` จับคู่กับ `50`
> - `30` จับคู่กับ `100`

````python
urgent_percentages = []

for urgent_count, total_count in zip(
    urgent_counts,
    total_counts,
):
    urgent_percentage = (
        urgent_count / total_count
    ) * 100

    urgent_percentages.append(
        urgent_percentage
    )

urgent_percentages
````

## แบบฝึกหัดที่ 4: จัดกลุ่มระดับความเปราะบาง

```python

    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]
```

กำหนดเกณฑ์ดังนี้

| คะแนน | กลุ่ม |
|---:|---|
| ตั้งแต่ 4.0 ขึ้นไป | High |
| ตั้งแต่ 2.5 แต่น้อยกว่า 4.0 | Medium |
| น้อยกว่า 2.5 | Low |

ให้ใช้ Loop ร่วมกับ Condition สร้าง List ชื่อ `vulnerability_groups`

````python
vulnerability_scores = [
    4.3,
    3.1,
    2.2,
    4.8,
    3.7,
    1.9,
]
````

## แบบฝึกหัดที่ 5: ตรวจสอบคะแนนผิดปกติและนับกลุ่มความเสี่ยงสูง

กำหนดคะแนนดังนี้

```python
vulnerability_scores = [
    4.3,
    -1.0,
    3.1,
    6.0,
    4.8,
    2.2,
]
```

ให้ใช้ Loop ทำงานต่อไปนี้

1. สร้าง List ชื่อ `score_results`
2. คะแนนต่ำกว่า 0 หรือสูงกว่า 5 ให้เป็น `"Invalid Score"`
3. คะแนนตั้งแต่ 4.0 ขึ้นไปให้เป็น `"High"`
4. คะแนนตั้งแต่ 2.5 แต่น้อยกว่า 4.0 ให้เป็น `"Medium"`
5. คะแนนที่เหลือให้เป็น `"Low"`
6. นับจำนวนคะแนนที่ถูกจัดเป็น `"High"` และเก็บใน `high_risk_count`

คะแนนที่ไม่ถูกต้องต้องไม่นำมานับเป็นกลุ่มความเสี่ยงสูง

````python
vulnerability_scores = [
    4.3,
    -1.0,
    3.1,
    6.0,
    4.8,
    2.2,
]
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 5
>
> ต้องตรวจสอบคะแนนที่อยู่นอกช่วงก่อนจัดระดับ
>
> ตัวนับ `high_risk_count` เริ่มต้นที่ `0` และเพิ่มขึ้นครั้งละ 1 เฉพาะเมื่อคะแนนอยู่ในกลุ่ม `"High"`

````python
score_results = []
high_risk_count = 0

for score in vulnerability_scores:
    if score < 0 or score > 5:
        result = "Invalid Score"
    elif score >= 4.0:
        result = "High"
        high_risk_count += 1
    elif score >= 2.5:
        result = "Medium"
    else:
        result = "Low"

    score_results.append(result)

print(score_results)
print("High-risk count:", high_risk_count)
````

## แบบฝึกหัดที่ 9: ข้ามค่าว่างและหยุดเมื่อพบเป้าหมาย

กำหนดข้อมูลดังนี้

```python
respondent_ids = [
    "R001",
    None,
    "R002",
    "R003",
    "R004",
]

target_id = "R003"
```

ให้ใช้ Loop ทำงานตามลำดับต่อไปนี้

1. หากค่าเป็น `None` ให้ใช้ `continue` ข้ามรายการนั้น
2. แสดงข้อความ `Checking: ...` สำหรับรหัสที่ไม่ใช่ `None`
3. หากพบ `target_id` ให้แสดงข้อความ `"R003 found"`
4. ใช้ `break` หยุด Loop หลังพบเป้าหมาย

````python
respondent_ids = [
    "R001",
    None,
    "R002",
    "R003",
    "R004",
]

target_id = "R003"
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 9
>
> - `continue` ข้ามรอบที่พบ `None`
> - `break` หยุด Loop ทันทีหลังพบรหัสเป้าหมาย
> - `"R004"` จะไม่ถูกตรวจสอบ เพราะ Loop หยุดหลังพบ `"R003"`

````python
for respondent_id in respondent_ids:
    if respondent_id is None:
        continue

    print(f"Checking: {respondent_id}")

    if respondent_id == target_id:
        print(f"{respondent_id} found")
        break
````
