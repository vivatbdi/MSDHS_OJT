## 1. Condition คืออะไร

ตัวอย่างคำถามที่เป็นเงื่อนไข เช่น

## 2. Comparison Operators

Comparison operators คือตัวดำเนินการที่ใช้เปรียบเทียบค่า

| Operator | ความหมาย | ตัวอย่าง |
|---|---|---|
| `==` | เท่ากับ | `sales == target` |
| `!=` | ไม่เท่ากับ | `region != "Bangkok"` |
| `>` | มากกว่า | `sales > target` |
| `<` | น้อยกว่า | `sales < target` |
| `>=` | มากกว่าหรือเท่ากับ | `sales >= target` |
| `<=` | น้อยกว่าหรือเท่ากับ | `sales <= target` |

ข้อควรระวัง:

- `=` ใช้กำหนดค่าให้ตัวแปร
- `==` ใช้เปรียบเทียบว่าเท่ากันหรือไม่

````python
sales = 85000
target = 100000

sales == target
````

````python
sales != target
````

````python
sales < target
````

````python
sales >= target
````

## 3. การใช้ `if`

คำสั่ง `if` ใช้เมื่อต้องการให้ Python ทำบางอย่าง “เมื่อเงื่อนไขเป็นจริง”

รูปแบบพื้นฐานคือ

```python
if condition:
    statement
```

จุดสำคัญคือ

- หลังเงื่อนไขต้องมีเครื่องหมาย colon `:`
- บรรทัดที่อยู่ภายใต้ `if` ต้องเยื้องเข้าไป

````python
sales = 120000
target = 100000

if sales >= target:
    print("Achieved target")
````

ในตัวอย่างด้านบน Python จะตรวจสอบว่า `sales >= target` เป็นจริงหรือไม่

ถ้าเป็นจริง จึงแสดงข้อความ `"Achieved target"`

ถ้าไม่เป็นจริง จะไม่มีอะไรเกิดขึ้น เพราะเรายังไม่ได้กำหนดทางเลือกอื่น

````python
sales = 85000
target = 100000

if sales >= target:
    print("Achieved target")
````

## 4. การใช้ `if-else`

ถ้าต้องการให้ Python ทำอย่างหนึ่งเมื่อเงื่อนไขเป็นจริง และทำอีกอย่างเมื่อเงื่อนไขเป็นเท็จ เราใช้ `if-else`

รูปแบบคือ

```python
if condition:
    statement_when_true
else:
    statement_when_false
```

````python
sales = 85000
target = 100000

if sales >= target:
    result = "Achieved"
else:
    result = "Not Achieved"

result
````

ตัวอย่างนี้คล้ายกับสูตร Excel

```excel
=IF(Sales>=Target,"Achieved","Not Achieved")
```

ใน Python เราเขียนแยกเป็นหลายบรรทัด เพื่อให้อ่านและตรวจสอบขั้นตอนได้ชัดเจน

````python
customer_spending = 25000
vip_threshold = 20000

if customer_spending >= vip_threshold:
    customer_group = "VIP"
else:
    customer_group = "Regular"

customer_group
````

## 5. การใช้ `if-elif-else`

ถ้ามีมากกว่า 2 เงื่อนไข เราสามารถใช้ `elif`

`elif` ย่อมาจาก “else if”

รูปแบบคือ

```python
if condition_1:
    statement_1
elif condition_2:
    statement_2
elif condition_3:
    statement_3
else:
    statement_when_no_condition_is_true
```

### ตัวอย่าง: จัดกลุ่มยอดขาย

สมมติเราต้องการจัดกลุ่มยอดขายตามเกณฑ์ดังนี้

| ยอดขาย | กลุ่ม |
|---:|---|
| ตั้งแต่ 100,000 ขึ้นไป | High |
| ตั้งแต่ 50,000 ถึงน้อยกว่า 100,000 | Medium |
| น้อยกว่า 50,000 | Low |

````python
sales = 75000

if sales >= 100000:
    sales_group = "High"
elif sales >= 50000:
    sales_group = "Medium"
else:
    sales_group = "Low"

sales_group
````

Python จะตรวจสอบเงื่อนไขจากบนลงล่าง

ในตัวอย่างนี้

1. ตรวจสอบก่อนว่า `sales >= 100000` หรือไม่
2. ถ้าไม่ใช่ จึงตรวจสอบว่า `sales >= 50000` หรือไม่
3. ถ้าไม่เข้าเงื่อนไขใดเลย จึงใช้ `else`

ลำดับของเงื่อนไขจึงสำคัญมาก

````python
sales = 125000

if sales >= 100000:
    sales_group = "High"
elif sales >= 50000:
    sales_group = "Medium"
else:
    sales_group = "Low"

sales_group
````

### ตัวอย่าง: จัดระดับความเสี่ยงของรายการ

สมมติเราตรวจสอบจำนวนวันที่ชำระเงินล่าช้า

| จำนวนวันที่ล่าช้า | ระดับความเสี่ยง |
|---:|---|
| 0 วัน | No Delay |
| 1-30 วัน | Low Risk |
| 31-60 วัน | Medium Risk |
| มากกว่า 60 วัน | High Risk |

````python
delay_days = 45

if delay_days == 0:
    risk_level = "No Delay"
elif delay_days <= 30:
    risk_level = "Low Risk"
elif delay_days <= 60:
    risk_level = "Medium Risk"
else:
    risk_level = "High Risk"

risk_level
````

## 6. Logical Operators: `and`, `or`, `not`

ในงานข้อมูลจริง เงื่อนไขมักไม่ได้มีเงื่อนไขเดียว

เช่น

- ยอดขายมากกว่า 100,000 และ อยู่ในกรุงเทพฯ
- สถานะเป็น Completed หรือ Approved
- ไม่ใช่รายการที่ถูกยกเลิก

Python มี logical operators สำหรับรวมหลายเงื่อนไข

## `and`

ใช้เมื่อทุกเงื่อนไขต้องเป็นจริงพร้อมกัน

| เงื่อนไขที่ 1 | เงื่อนไขที่ 2 | ผลลัพธ์ของ `and` |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

````python
sales = 150000
region = "Bangkok"

if sales >= 100000 and region == "Bangkok":
    result = "High sales in Bangkok"
else:
    result = "Other"

result
````

ตัวอย่างนี้คล้ายกับ Excel

```excel
=IF(AND(Sales>=100000,Region="Bangkok"),"High sales in Bangkok","Other")
```

## `or`

ใช้เมื่อมีอย่างน้อยหนึ่งเงื่อนไขเป็นจริง

````python
status = "Approved"

if status == "Completed" or status == "Approved":
    result = "Valid transaction"
else:
    result = "Need review"

result
````

ตัวอย่างนี้คล้ายกับ Excel

```excel
=IF(OR(Status="Completed",Status="Approved"),"Valid transaction","Need review")
```

## `not`

ใช้กลับค่าของเงื่อนไข

ถ้าเงื่อนไขเดิมเป็น `True` จะกลายเป็น `False`  
ถ้าเงื่อนไขเดิมเป็น `False` จะกลายเป็น `True`

````python
is_cancelled = False

if not is_cancelled:
    result = "Include in report"
else:
    result = "Exclude from report"

result
````

## 7. Membership Operators: `in` และ `not in`

ในงานข้อมูล เรามักต้องตรวจสอบว่าค่าหนึ่งอยู่ในกลุ่มค่าที่สนใจหรือไม่

เช่น

- จังหวัดนี้อยู่ในพื้นที่เป้าหมายหรือไม่
- สถานะนี้อยู่ในสถานะที่อนุญาตหรือไม่
- สินค้านี้อยู่ในกลุ่มสินค้าที่ต้องติดตามหรือไม่

Python ใช้ `in` และ `not in` สำหรับงานลักษณะนี้

````python
target_regions = ["Bangkok", "Chiang Mai", "Khon Kaen"]

region = "Bangkok"

region in target_regions
````

````python
region = "Phuket"

region in target_regions
````

````python
status = "Cancelled"
valid_statuses = ["Completed", "Approved", "Shipped"]

if status not in valid_statuses:
    result = "Need review"
else:
    result = "Valid transaction"

result
````

แนวคิดนี้คล้ายกับการตรวจสอบใน Excel ว่าค่าหนึ่งอยู่ใน list หรือ lookup range ที่กำหนดไว้หรือไม่

ใน Python เราสามารถเก็บกลุ่มค่าที่ต้องการตรวจสอบไว้ใน list แล้วใช้ `in`

## 9. เงื่อนไขแบบซ้อนกัน

บางครั้งเงื่อนไขหนึ่งอาจอยู่ภายในอีกเงื่อนไขหนึ่ง เรียกว่า nested condition

อย่างไรก็ตาม สำหรับผู้เริ่มต้นควรใช้เท่าที่จำเป็น เพราะถ้าซ้อนหลายชั้นเกินไป โค้ดจะอ่านยาก

````python
is_member = True
total_spending = 18000

if is_member:
    if total_spending >= 10000:
        discount_rate = 0.10
    else:
        discount_rate = 0.05
else:
    discount_rate = 0.00

discount_rate
````

ตัวอย่างนี้มี logic ว่า

- ถ้าเป็นสมาชิก และยอดซื้ออย่างน้อย 10,000 ได้ส่วนลด 10%
- ถ้าเป็นสมาชิก แต่ยอดซื้อต่ำกว่า 10,000 ได้ส่วนลด 5%
- ถ้าไม่เป็นสมาชิก ไม่ได้ส่วนลด

แม้ nested condition จะใช้ได้ แต่ในงานจริงควรจัดโครงสร้างให้ไม่ซับซ้อนเกินไป

ควรเขียนเรียงใหม่แบบนี้

````python
sales = 150000

if sales >= 100000:
    group = "High"
elif sales >= 50000:
    group = "Medium"
else:
    group = "Low"

group
````
