# 4) Lists (ลิสต์) สำหรับงานวิเคราะห์ข้อมูลเบื้องต้น

ในการทำงานกับข้อมูล เรามักมี “ข้อมูลหลายค่า” เช่น
- คะแนนของนักเรียนหลายคน
- ยอดขายรายวันหลายวัน
- รายชื่อจังหวัดหลายจังหวัด

ใน Python เราเก็บข้อมูลหลายค่าไว้ด้วย **list**  
ให้คิดว่า list คล้าย “ข้อมูล 1 คอลัมน์” ใน Excel (หลายแถว)

---

## สิ่งที่คุณจะได้จากบทนี้
- เข้าใจว่า list คืออะไร และใช้แทนข้อมูลหลายค่าได้
- สร้าง list และเรียกดูข้อมูลใน list
- เพิ่ม/แก้ไขค่าใน list
- ใช้ฟังก์ชันที่ใช้บ่อยในการสรุปข้อมูล: `len()`, `sum()`, `min()`, `max()`
- ทำแบบฝึกหัดเพื่อสรุปข้อมูลเบื้องต้นจาก list

---

## ภาพรวมเนื้อหา (ลำดับการเรียน)
1. Lists เทียบกับ Excel (คอลัมน์/ช่วงข้อมูล)
2. สร้าง list และดูค่า
3. เลือกข้อมูลด้วย index (ตำแหน่ง)
4. เพิ่ม/แก้ไขข้อมูลใน list
5. สรุปข้อมูลจาก list (len/sum/min/max/average)
6. แบบฝึกหัดสั้น ๆ ในสไตล์งาน data

## A. Lists เทียบกับ Excel

ใน Excel:
- คอลัมน์หนึ่ง ๆ เช่น B2:B10 คือข้อมูลหลายแถว
- เรามักใช้สูตรกับช่วงข้อมูล เช่น `=SUM(B2:B10)`

ใน Python:
- เราเก็บค่าหลายค่าไว้ใน **list**
- แล้วใช้คำสั่งสรุป เช่น `sum(list)` หรือ `len(list)`

✅ ลองดูตัวอย่าง list แรกของคุณ

```python
scores = [80, 75, 90, 60]
scores
```

## B. ดูจำนวนข้อมูล (เหมือนนับจำนวนแถว)

`len()` ใช้ดูจำนวนสมาชิกใน list  
คล้ายกับการนับจำนวนแถวในช่วงข้อมูลของ Excel

✅ ลองรัน

```python
len(scores)
```

## C. เลือกข้อมูลด้วย index (ตำแหน่ง)

ใน list แต่ละค่ามี “ตำแหน่ง” เริ่มจาก 0

- ตัวแรก = index 0
- ตัวที่สอง = index 1
- ตัวสุดท้าย = index -1

✅ ลองรัน

```python
scores[0]
```

```python
scores[1]
```

```python
scores[-1]
```

## D. เพิ่มข้อมูลใน list (append)

งานจริงมักมีข้อมูลเพิ่มเข้ามา เช่น เพิ่มคะแนนอีกคน หรือเพิ่มยอดขายอีกวัน

ใช้ `append()` เพื่อเพิ่มข้อมูล “ต่อท้าย” list

✅ ลองรัน

```python
scores.append(85)
scores
```

## E. แก้ไขข้อมูลใน list

ถ้าพบว่าค่าบางตัวผิด (data cleaning เบื้องต้น) เราแก้ได้ด้วย index

ตัวอย่าง: ค่า index 1 เดิมเป็น 75 แต่ควรเป็น 78

✅ ลองรัน

```python
scores[1] = 78
scores
```

## F. สรุปข้อมูลจาก list (เหมือนสูตรใน Excel)

ฟังก์ชันที่ใช้บ่อยมากในงาน data:
- `sum()` ผลรวม (เหมือน SUM)
- `min()` ค่าน้อยสุด (เหมือน MIN)
- `max()` ค่ามากสุด (เหมือน MAX)
- `len()` จำนวนข้อมูล (เหมือนนับจำนวนแถว)
- ค่าเฉลี่ย (เหมือน AVERAGE) = `sum(...) / len(...)`

✅ ลองรันทีละ cell

```python
sum(scores)
```

```python
min(scores)
```

```python
max(scores)
```

```python
len(scores)
```

### ค่าเฉลี่ย (Average)

ใน Excel: `=AVERAGE(B2:B10)`  
ใน Python: `sum(scores) / len(scores)`

✅ ลองรัน

```python
avg_score = sum(scores) / len(scores)
avg_score
```

## G. เลือกบางส่วนของ list (slicing) แบบง่าย

บางครั้งเราอยากดู “บางช่วง” ของข้อมูล  
เช่น 3 ตัวแรก หรือ 2 ตัวสุดท้าย

- `scores[:3]` = 3 ตัวแรก
- `scores[-2:]` = 2 ตัวสุดท้าย

✅ ลองรัน

```python
scores[:3]
```

```python
scores[-2:]
```

## H. แบบฝึกหัด 1: ยอดขายรายวัน (Daily Sales)

สถานการณ์:
คุณมียอดขาย 7 วัน (หนึ่งสัปดาห์) เก็บใน list  
ต้องการหา:
- ผลรวมยอดขาย
- ค่าเฉลี่ยยอดขายต่อวัน
- ยอดขายมากสุด/น้อยสุด

✅ รัน cell แล้วลองเปลี่ยนตัวเลขให้เป็นข้อมูลของคุณ

```python
daily_sales = [1200, 1500, 900, 1800, 2000, 1700, 1600]

total = sum(daily_sales)
avg = total / len(daily_sales)
mn = min(daily_sales)
mx = max(daily_sales)

print("Total sales =", total)
print("Average per day =", avg)
print("Min =", mn)
print("Max =", mx)
```

สมมติ list ของคะแนนมีค่าหนึ่ง “ผิดปกติ” เช่น -10 (คะแนนติดลบ)  
เราจะลอง “ดูข้อมูล” ก่อน แล้วค่อยจัดการในบทถัด ๆ ไป

✅ รัน cell นี้ แล้วสังเกตว่าอะไรแปลก

✅ ลองใช้ `min()` และ `max()` กับ scores2
