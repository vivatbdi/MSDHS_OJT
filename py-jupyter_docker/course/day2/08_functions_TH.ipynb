## A. Functions เทียบกับ Excel

ใน Excel:
- เราใช้สูตร เช่น `SUM()`, `AVERAGE()`, `IF()`
- สูตรหนึ่งสามารถใช้ซ้ำได้หลายเซลล์

ใน Python:
- Function ก็ทำหน้าที่เหมือน “สูตรที่ตั้งชื่อไว้”
- เรียกใช้ซ้ำได้กับข้อมูลหลายชุด

ตัวอย่าง function ที่คุณใช้มาแล้ว:
`sum()`, `len()`, `min()`, `max()`

## B. ใช้ function ที่มีอยู่แล้ว (Built‑in Functions)

ลองดูตัวอย่างจาก list ของคะแนน

````python
scores = [80, 75, 90, 60]

print("Count =", len(scores))
print("Total =", sum(scores))
print("Min =", min(scores))
print("Max =", max(scores))
````

## C. ทำไมเราต้องเขียน function เอง?

สมมติว่าเราต้องตรวจ “ผ่าน/ไม่ผ่าน” หลายครั้ง  
ถ้าเขียน if–else ซ้ำ ๆ จะยาวและอ่านยาก

เราจึงเขียน function เพื่อ:
- ซ่อนรายละเอียด
- เรียกใช้สั้น ๆ

## D. เขียน function แรกของคุณ

โครงสร้างพื้นฐาน:
```python
def function_name(input):
    ทำอะไรบางอย่าง
    return ผลลัพธ์
```

````python
def check_pass(score):
    if score >= 50:
        return "Pass"
    else:
        return "Fail"
````

````python
check_pass(75)
````

ลองใช้ function กับค่าต่าง ๆ

````python
check_pass(45)
````

````python
check_pass(90)
````

## E. ใช้ function กับข้อมูลหลายค่า (list)

เมื่อรวม function + loop  
เราจะได้ workflow ที่เหมือนงาน data จริงมาก

✅ ลองรัน

````python
scores = [80, 45, 90, 30, 60]

for s in scores:
    print(s, "->", check_pass(s))
````

## F. Function ที่คืนค่าตัวเลข (เช่น ค่าเฉลี่ย)

Function ไม่จำเป็นต้องคืนข้อความ  
สามารถคืนค่าตัวเลขเพื่อเอาไปวิเคราะห์ต่อได้

✅ ตัวอย่าง: function คำนวณค่าเฉลี่ย

````python
def average(values):
    total = 0
    for v in values:
        total = total + v
    return total / len(values)

average(scores)
````

## G. ใช้ function กับ dictionary

สมมติว่า 1 record (dictionary) มีข้อมูลคะแนน  
เราเขียน function เพื่อวิเคราะห์ record นั้นได้

✅ ลองรัน

````python
student = {"name": "Alice", "score": 72}

def add_pass_status(record):
    record["passed"] = record["score"] >= 50
    return record

add_pass_status(student)
````

## H. ใช้ function กับตารางข้อมูล (list of dictionaries)

นี่คือรูปแบบที่ใกล้กับ Pandas มาก

✅ ลองรัน

````python
students = [
    {"name": "Alice", "score": 82},
    {"name": "Bob", "score": 45},
    {"name": "Charlie", "score": 91}
]

for s in students:
    add_pass_status(s)

students
````

## I. แบบฝึกหัด 1: เขียน function ตรวจรายได้

ให้เขียน function ชื่อ `is_high_income`  
- รับค่า salary
- คืนค่า True ถ้า salary >= 50000
- คืนค่า False ถ้าน้อยกว่า

แล้วลองใช้กับตัวเลขหลายค่า

````python
def is_high_income(salary):
    return salary >= 50000
````

````python
print(is_high_income(48000))
````

````python
print(is_high_income(60000))
````

## J. แบบฝึกหัด 2: ใช้ function กับข้อมูลหลายแถว

มี list ของพนักงาน (list of dictionaries)

ให้:
- เขียน function เพิ่ม key `high_income`
- ใช้กับพนักงานทุกคน

````python
employees = [
    {"name": "Ann", "salary": 52000},
    {"name": "Ben", "salary": 45000},
    {"name": "Cat", "salary": 60000}
]

def add_high_income(record):
    record["high_income"] = record["salary"] >= 50000
    return record

for e in employees:
    add_high_income(e)

employees
````
