## 1. Function คืออะไร

**Function** คือชุดคำสั่งที่ถูกรวมไว้ภายใต้ชื่อเดียว

เมื่อต้องการใช้งานชุดคำสั่งนั้น เราสามารถเรียกชื่อ function ได้ แทนที่จะเขียนคำสั่งทั้งหมดซ้ำอีกครั้ง

Python มี function ที่เราเคยใช้มาแล้ว เช่น

| Function | ใช้ทำอะไร |
|---|---|
| `print()` | แสดงผล |
| `len()` | นับจำนวนสมาชิก |
| `sum()` | หาผลรวม |
| `min()` | หาค่าน้อยที่สุด |
| `max()` | หาค่ามากที่สุด |
| `type()` | ตรวจสอบชนิดข้อมูล |

Function เหล่านี้เป็น function ที่ Python เตรียมไว้ให้  
ในบทนี้ เราจะเรียนรู้วิธีสร้าง function ของเราเอง

````python
sales_list = [12000, 18000, 9500, 22000]

total_sales = sum(sales_list)
total_sales
````

ตัวอย่างด้านบนใช้ function `sum()` เพื่อหาผลรวมจากรายการยอดขาย

## 3. การสร้าง Function ด้วย `def`

เราสร้าง function ด้วยคำสั่ง `def`

รูปแบบพื้นฐานคือ

```python
def function_name(parameter):
    statement
    return result
```

ส่วนประกอบหลักคือ

- `def` ใช้บอก Python ว่าเรากำลังสร้าง function
- `function_name` คือชื่อ function
- `parameter` คือข้อมูลที่ function ต้องรับเข้าไป
- `return` คือผลลัพธ์ที่ function ส่งกลับออกมา

````python
def add_vat(price):
    vat_rate = 0.07
    price_with_vat = price * (1 + vat_rate)
    return price_with_vat
````

เมื่อสร้าง function แล้ว เรายังไม่ได้รันการคำนวณทันที

function จะทำงานเมื่อเรา “เรียกใช้” function นั้น

````python
add_vat(1000)
````

````python
add_vat(2500)
````

ตอนนี้เราเขียนสูตรคำนวณ VAT ไว้เพียงครั้งเดียวใน function `add_vat()`

เมื่ออยากคำนวณราคาหลังรวม VAT ของราคาใด ก็ส่งราคานั้นเข้าไปใน function

## 4. Parameter และ Argument

คำว่า **parameter** และ **argument** มักทำให้ผู้เริ่มต้นสับสน

ให้เข้าใจง่าย ๆ ดังนี้

- **Parameter** คือชื่อตัวแปรที่ function รอรับ
- **Argument** คือค่าจริงที่เราส่งเข้าไปตอนเรียกใช้ function

````python
def calculate_discount(price, discount_rate):
    discounted_price = price * (1 - discount_rate)
    return discounted_price
````

ใน function นี้

```python
def calculate_discount(price, discount_rate):
```

`price` คือ parameter
`discount_rate` คือ parameter

เมื่่อเรียกใช้ function:

````python
calculate_discount(3500, 0.15)
````

ค่าที่ส่งเข้าไปคือ argument

- `3500` คือ argument ที่ส่งให้ `price`
- `0.15` คือ argument ที่ส่งให้ `discount_rate`

## 5. การใช้ `return`

`return` ใช้ส่งผลลัพธ์ออกจาก function

ถ้า function ไม่มี `return` Python จะทำงานภายใน function ได้ แต่จะไม่มีค่าผลลัพธ์ส่งกลับมาให้เราเก็บไว้ใช้ต่อ

````python
def calculate_profit_margin(revenue, cost):
    profit = revenue - cost
    profit_margin = profit / revenue
    return profit_margin
````

````python
margin = calculate_profit_margin(250000, 175000)

margin
````

เพราะ function มี `return` เราจึงสามารถนำผลลัพธ์ไปเก็บในตัวแปร `margin` และนำไปใช้ต่อได้

### ตัวอย่าง function ที่ใช้ `print()` แต่ไม่มี `return`

````python
def show_profit_margin(revenue, cost):
    profit = revenue - cost
    profit_margin = profit / revenue
    print(f"Profit margin: {profit_margin:.2%}")
````

````python
result = show_profit_margin(250000, 175000)

result
````

จะเห็นว่า function แสดงข้อความได้ แต่ค่า `result` เป็น `None`

เพราะ function นี้ไม่มี `return`

แนวทางสำหรับงานข้อมูลคือ

- ใช้ `return` เมื่อต้องการนำผลลัพธ์ไปใช้ต่อ
- ใช้ `print()` เมื่อต้องการแสดงผลให้คนอ่านเห็น

## 6. Function สำหรับคำนวณ Metric

ในงานวิเคราะห์ข้อมูล เรามักต้องคำนวณ metric ซ้ำหลายครั้ง เช่น

- profit
- profit margin
- growth rate
- average order value
- conversion rate
- percentage difference

การสร้าง function สำหรับ metric เหล่านี้ช่วยให้ code อ่านง่าย และลดการคำนวณผิด

### ตัวอย่าง: คำนวณ Profit Margin

````python
def profit_margin(revenue, cost):
    profit = revenue - cost
    margin = profit / revenue
    return margin
````

````python
profit_margin(120000, 75000)
````

### ตัวอย่าง: คำนวณ Growth Rate

สูตรที่ใช้คือ

```text
(current_value - previous_value) / previous_value
```

````python
def growth_rate(current_value, previous_value):
    growth = (current_value - previous_value) / previous_value
    return growth
````

````python
growth_rate(150000, 120000)
````

```text
total_sales / number_of_orders
```

````python
def average_order_value(total_sales, number_of_orders):
    aov = total_sales / number_of_orders
    return aov
````

````python
average_order_value(250000, 125)
````

## การสร้างข้อความสรุปจาก Function

Function สามารถคืนค่าที่เป็นข้อความได้เช่นกัน

ตัวอย่างเช่น สร้างข้อความสรุปอัตรากำไร

## 7. Function ที่มี Condition ภายใน

Function สามารถมีเงื่อนไขอยู่ภายในได้

เหมาะกับงานที่ต้องใช้ business rule ซ้ำ เช่น

- จัดกลุ่มยอดขาย
- จัดกลุ่มลูกค้า
- ตรวจสอบสถานะรายการ
- คำนวณส่วนลด
- ระบุว่ารายการควรนำเข้า report หรือไม่

### ตัวอย่าง: Function สำหรับจัดกลุ่มยอดขาย

````python
def classify_sales(sales):
    if sales >= 100000:
        group = "High"
    elif sales >= 50000:
        group = "Medium"
    else:
        group = "Low"
    
    return group
````

````python
classify_sales(125000)
````

````python
classify_sales(72000)
````

````python
classify_sales(35000)
````

เมื่อเราเขียน logic การจัดกลุ่มไว้ใน function แล้ว เราสามารถใช้ซ้ำได้ทุกครั้งที่ต้องจัดกลุ่มยอดขาย

ข้อดีคือถ้าเกณฑ์เปลี่ยน เราแก้ไขที่ function เพียงจุดเดียว

### ตัวอย่าง: Function สำหรับตรวจสอบสถานะ transaction

````python
def check_transaction_status(status):
    valid_statuses = ["Completed", "Approved"]
    
    if status in valid_statuses:
        result = "Valid"
    else:
        result = "Need Review"
    
    return result
````

````python
check_transaction_status("Completed")
````

````python
check_transaction_status("Pending")
````

### ตัวอย่าง: Function สำหรับคำนวณส่วนลดจากเงื่อนไข

````python
def calculate_discount_rate(is_member, total_spending):
    if is_member and total_spending >= 10000:
        discount_rate = 0.10
    elif is_member:
        discount_rate = 0.05
    else:
        discount_rate = 0.00
    
    return discount_rate
````

````python
calculate_discount_rate(True, 15000)
````

````python
calculate_discount_rate(True, 8000)
````

````python
calculate_discount_rate(False, 15000)
````

## 8. Function กับ Loop

Function จะมีประโยชน์มากขึ้นเมื่อใช้ร่วมกับ loop

แนวคิดคือ

1. สร้าง function สำหรับ logic หนึ่งอย่าง
2. ใช้ loop เพื่อเรียก function นั้นกับข้อมูลหลายรายการ
3. เก็บผลลัพธ์ไว้ใน list หรือ dictionary

### ตัวอย่าง: จัดกลุ่มยอดขายหลายรายการด้วย Function

````python
sales_list = [120000, 75000, 45000, 180000, 52000]

sales_groups = []

for sales in sales_list:
    group = classify_sales(sales)
    sales_groups.append(group)

sales_groups
````

เมื่อใช้ function ร่วมกับ loop โค้ดจะอ่านง่ายขึ้น

แทนที่จะเขียนเงื่อนไขทั้งหมดไว้ใน loop เราสามารถย้าย logic ไปไว้ใน function ชื่อ `classify_sales()`

### ตัวอย่าง: ตรวจสอบ transaction หลายรายการด้วย Function

````python
statuses = ["Completed", "Pending", "Approved", "Cancelled", "Completed"]

status_results = []

for status in statuses:
    result = check_transaction_status(status)
    status_results.append(result)

status_results
````

รูปแบบนี้ช่วยให้ code เป็นระบบมากขึ้น

- function ดูแล logic
- loop ดูแลการทำซ้ำกับข้อมูลหลายรายการ

## 9. Function สำหรับทำความสะอาดข้อมูลเบื้องต้น

ในงานข้อมูล เรามักต้องทำความสะอาดข้อมูลก่อนวิเคราะห์ เช่น

- ลบช่องว่างหัวท้าย
- ทำข้อความให้เป็นตัวพิมพ์เล็ก
- เปลี่ยนช่องว่างเป็น underscore
- แปลงชื่อคอลัมน์ให้เป็นมาตรฐาน
- ลบ comma ออกจากตัวเลขที่เป็นข้อความ

เราสามารถสร้าง function เพื่อทำงานเหล่านี้ซ้ำได้

### ตัวอย่าง: ทำความสะอาดชื่อคอลัมน์

สมมติว่าได้ชื่อคอลัมน์จาก Excel แบบนี้

```python
" Sales Amount "
```

แต่เราต้องการแปลงเป็น

```python
" sales_amount "
```

````python
def clean_column_name(column_name):
    cleaned_name = column_name.strip()
    cleaned_name = cleaned_name.lower()
    cleaned_name = cleaned_name.replace(" ", "_")
    return cleaned_name
````

````python
clean_column_name(" Sales Amount ")
````

````python
clean_column_name(" Customer Name ")
````

Function นี้มีประโยชน์มากเมื่อได้รับไฟล์ Excel ที่ชื่อคอลัมน์ไม่เป็นมาตรฐาน เช่น

- มีช่องว่างเกิน
- ใช้ตัวพิมพ์ใหญ่และเล็กปนกัน
- มีช่องว่างระหว่างคำ

การแปลงชื่อคอลัมน์ให้เป็นรูปแบบเดียวกันช่วยให้ทำงานต่อใน pandas ได้ง่ายขึ้น

### ตัวอย่าง: แปลงตัวเลขที่มี comma

````python
def convert_text_to_number(text_value):
    cleaned_text = text_value.replace(",", "")
    number = float(cleaned_text)
    return number
````

````python
convert_text_to_number("12,500")
````

````python
convert_text_to_number("1,250,000")
````

Function นี้ช่วยแปลงข้อมูลตัวเลขที่ถูก copy มาจาก Excel หรืออ่านมาเป็นข้อความให้กลายเป็นตัวเลขที่คำนวณต่อได้

## 10. การตั้งชื่อและออกแบบ Function ที่ดี

Function ที่ดีควรอ่านแล้วเข้าใจว่าใช้ทำอะไร

แนวทางเบื้องต้น:

1. ชื่อ function ควรเป็นคำกริยาหรือสื่อถึงการกระทำ
2. ใช้ตัวพิมพ์เล็กและ underscore
3. หนึ่ง function ควรทำงานหลักอย่างเดียว
4. ถ้าต้องใช้ผลลัพธ์ต่อ ควรมี `return`
5. หลีกเลี่ยงชื่อกว้างเกินไป เช่น `do_stuff()` หรือ `process()`

## ตัวอย่างชื่อ Function

| ชื่อ Function | ความเห็น |
|---|---|
| `calculate_profit()` | ดี เพราะบอกว่าคำนวณกำไร |
| `classify_sales()` | ดี เพราะบอกว่าจัดกลุ่มยอดขาย |
| `clean_column_name()` | ดี เพราะบอกว่าทำความสะอาดชื่อคอลัมน์ |
| `check_transaction_status()` | ดี เพราะบอกว่าตรวจสอบสถานะ transaction |
| `data()` | ไม่ดี เพราะกว้างเกินไป |
| `abc()` | ไม่ดี เพราะไม่สื่อความหมาย |
| `process()` | ใช้ได้บางกรณี แต่สำหรับผู้เริ่มต้นควรตั้งชื่อให้เฉพาะเจาะจงกว่า |

## การใส่คำอธิบายสั้น ๆ ใน Function

เราสามารถใส่ข้อความอธิบาย function ได้ เรียกว่า docstring

Docstring ช่วยบอกว่า function นี้ทำอะไร รับค่าอะไร และคืนค่าอะไร

````python
def calculate_profit_margin(revenue, cost):
    """
    Calculate profit margin from revenue and cost.
    """
    profit = revenue - cost
    margin = profit / revenue
    return margin
````

สำหรับหลักสูตรพื้นฐาน ยังไม่จำเป็นต้องเขียน docstring ยาว  
แต่ควรเริ่มรู้จักแนวคิดว่า function ที่ดีควรอธิบายตัวเองได้
