# 6) Conditions (if-else) สำหรับงานวิเคราะห์ข้อมูลเบื้องต้น

ในการวิเคราะห์ข้อมูล เราไม่ได้แค่คำนวณค่า  
แต่เราต้อง **ตัดสินใจจากข้อมูล** เช่น
- คะแนนนี้ “ผ่านหรือไม่ผ่าน”
- ยอดขายวันนี้ “ถึงเป้าหรือยัง”
- ค่านี้ “ผิดปกติหรือไม่”

ใน Python เราใช้ **เงื่อนไข (condition)** ด้วย `if`, `elif`, `else`  
เพื่อให้โค้ด “ตัดสินใจ” ได้เหมือนที่เราคิดตอนดูข้อมูล

---

## สิ่งที่คุณจะได้จากบทนี้
- เข้าใจแนวคิด if–else สำหรับการตัดสินใจจากข้อมูล
- ใช้เครื่องหมายเปรียบเทียบ (`>`, `<`, `>=`, `<=`, `==`, `!=`)
- ใช้ if ร่วมกับ loop เพื่อวิเคราะห์ข้อมูลหลายค่า
- สร้างเงื่อนไขแบบหลายกรณี (if–elif–else)
- ทำแบบฝึกหัดสไตล์งาน data (ผ่าน/ไม่ผ่าน, ถึงเป้า, ค่าผิดปกติ)

---

## A. Conditions เทียบกับ Excel

ใน Excel เรามักใช้สูตร:
`=IF(A1 >= 50, "Pass", "Fail")`

ใน Python เราเขียนเงื่อนไขด้วย `if-else`  
แนวคิดเหมือนกัน:
- ถ้าเงื่อนไขเป็นจริง → ทำอย่างหนึ่ง
- ถ้าไม่จริง → ทำอีกอย่างหนึ่ง

✅ เริ่มจากตัวอย่างง่าย ๆ

```python
score = 75

if score >= 50:
    print("Pass")
else:
    print("Fail")
```

## B. เครื่องหมายเปรียบเทียบที่ใช้บ่อย

ในงาน data เราใช้เครื่องหมายเหล่านี้บ่อยมาก:
- `>` มากกว่า
- `<` น้อยกว่า
- `>=` มากกว่าหรือเท่ากับ
- `<=` น้อยกว่าหรือเท่ากับ
- `==` เท่ากับ
- `!=` ไม่เท่ากับ

✅ ลองรันและดูผลลัพธ์ (True / False)

```python
x = 10

print(x > 5)
```

```python
print(x < 5)
```

```python
print(x == 10)
```

```python
print(x != 10)
```

## C. if + loop = วิเคราะห์ข้อมูลหลายค่า

จุดพลังของ Python คือการใช้ `if` ร่วมกับ `for-loop`  
เพื่อวิเคราะห์ข้อมูลทั้งชุด (เหมือนใช้ IF กับหลายแถวใน Excel)

✅ ตัวอย่าง: ตรวจผ่าน/ไม่ผ่านของคะแนนหลายคน

```python
scores = [80, 45, 90, 30, 60]

for s in scores:
    if s >= 50:
        print(s, "-> Pass")
    else:
        print(s, "-> Fail")
```

## D. นับจำนวนตามเงื่อนไข (Count)

สถานการณ์งานจริง:
- อยากรู้ว่ามีกี่คน “ผ่าน”
- หรือมีกี่วันที่ยอดขาย “ถึงเป้า”

เราจะใช้ตัวแปรสะสมจำนวน

✅ ลองรัน

```python
count_pass = 0

for s in scores:
    if s >= 50:
        count_pass = count_pass + 1

count_pass
```

## E. หลายเงื่อนไขด้วย if–elif–else

บางครั้งข้อมูลมีหลายระดับ เช่น เกรด
- >= 80 → A
- >= 70 → B
- >= 60 → C
- อื่น ๆ → F

ใน Python ใช้ `elif` เพื่อเขียนให้อ่านง่าย

✅ ลองรัน

```python
score = 72

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

grade
```

## F. ใช้ if–elif–else กับข้อมูลหลายค่า

นำตัวอย่างเกรดไปใช้กับ list ของคะแนน

✅ ลองรัน

```python
scores = [85, 72, 64, 40, 90]

for s in scores:
    if s >= 80:
        grade = "A"
    elif s >= 70:
        grade = "B"
    elif s >= 60:
        grade = "C"
    else:
        grade = "F"
    
    print(s, "->", grade)
```

## G. แบบฝึกหัด 1: ตรวจสอบยอดขายถึงเป้าหรือไม่

สถานการณ์:
- มียอดขายรายวันใน list
- เป้ายอดขายต่อวัน = 1600

ให้แสดงผลว่าแต่ละวัน “ถึงเป้า” หรือ “ไม่ถึงเป้า”

```python
daily_sales = [1200, 1500, 900, 1800, 2000, 1700, 1600]
target = 1600

for x in daily_sales:
    if x >= target:
        print(x, "-> Meet target")
    else:
        print(x, "-> Not meet target")
```

## H. แบบฝึกหัด 2: แยกข้อมูลผิดปกติ

สมมติว่า:
- คะแนน < 0 หรือ > 100 ถือว่าผิดปกติ

ให้ตรวจสอบคะแนนแต่ละค่า และพิมพ์ว่า
- Normal
- Abnormal

```python
scores2 = [80, 75, -10, 90, 105, 60]

for s in scores2:
    if s < 0 or s > 100:
        print(s, "-> Abnormal")
    else:
        print(s, "-> Normal")
```
