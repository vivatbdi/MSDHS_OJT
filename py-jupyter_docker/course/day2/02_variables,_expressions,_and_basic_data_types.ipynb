## 1. ตัวแปรคืออะไร

**ตัวแปร** หรือ **variable** คือชื่อที่เราใช้เก็บค่าไว้ใน Python

เมื่อต้องใช้ค่านั้นซ้ำ เราสามารถเรียกใช้ผ่านชื่อตัวแปรได้ แทนที่จะเขียนค่าซ้ำหลายครั้ง

ตัวอย่างเช่น หากเราต้องการเก็บยอดขายของสินค้า เราอาจสร้างตัวแปรชื่อ `sales`

````python
sales = 15000
sales
````

จากตัวอย่างด้านบน

```python

```

หมายความว่า เรากำหนดค่า 15000 ให้กับตัวแปรชื่อ sales

เครื่องหมาย = ใน Python หมายถึง “กำหนดค่าให้ตัวแปร” ไม่ใช่เครื่องหมายเท่ากับแบบสมการทางคณิตศาสตร์

## ตัวแปรช่วยให้งานอ่านง่ายขึ้น

## 3. นิพจน์และตัวดำเนินการพื้นฐาน

**นิพจน์** หรือ **expression** คือส่วนของโค้ดที่ Python สามารถประมวลผลแล้วได้ค่าออกมา

ตัวอย่างเช่น

```python
100 + 50
sales * vat_rate
(total_sales / customer_count)
```

ตัวดำเนินการพื้นฐานที่ใช้บ่อย ได้แก่

| ตัวดำเนินการ | ความหมาย | ตัวอย่าง |
|---|---|---|
|+ | บวก | 10 + 5 |
|- | ลบ | 10 - 5 |
|* | คูณ |	10 * 5 |
|/ | หาร |	10 / 5 |
|** | ยกกำลัง |	10 ** 2 |
|% | เศษจากการหาร |	10 % 3 |
|// | หารแบบปัดเศษลง |	10 // 3 |

````python
revenue = 120000
cost = 75000

profit = revenue - cost
profit
````

````python
profit_margin = profit / revenue
profit_margin
````

ใน Excel เราอาจเขียนสูตรหา profit margin ว่า

```excel
=(Revenue-Cost)/Revenue
```

ใน Python เราสามารถแยกเป็นขั้นตอนโดยใช้ตัวแปร ทำให้โค้ดอ่านง่ายขึ้น

````python
profit = revenue - cost
profit_margin = profit / revenue
profit_margin
````

## 7. การแปลงชนิดข้อมูล

บางครั้งเราจำเป็นต้องแปลงชนิดข้อมูลก่อนนำไปคำนวณ

ฟังก์ชันที่ใช้บ่อย ได้แก่

| ฟังก์ชัน | ใช้ทำอะไร |
|---|---|
| `int()` | แปลงเป็นจำนวนเต็ม |
| `float()` | แปลงเป็นจำนวนทศนิยม |
| `str()` | แปลงเป็นข้อความ |
| `bool()` | แปลงเป็นค่าจริง/เท็จ |

````python
amount_text = "12500"

amount_number = int(amount_text)

amount_number
````

````python
type(amount_number)
````

````python
price_text = "125.50"

price_number = float(price_text)

price_number
````

ถ้าข้อความไม่ใช่รูปแบบตัวเลขที่ถูกต้อง Python จะแปลงไม่ได้

````python
# ลองรันเพื่อดู error ที่เกิดขึ้น

int("Bangkok")
````

ในงานข้อมูลจริง ปัญหาชนิดข้อมูลเกิดขึ้นบ่อย เช่น

- ตัวเลขถูกอ่านมาเป็นข้อความ
- วันที่ถูกอ่านมาเป็นข้อความ
- รหัสที่ขึ้นต้นด้วยเลขศูนย์ถูกมองเป็นตัวเลข
- ช่องว่างหรือ comma ทำให้แปลงเป็นตัวเลขไม่ได้

ตัวอย่างเช่น `"12,500"` ไม่สามารถแปลงด้วย `int()` ตรง ๆ ได้ เพราะมี comma อยู่ในข้อความ

````python
sales_text = "12,500"

clean_sales_text = sales_text.replace(",", "")
sales_number = int(clean_sales_text)

sales_number
````

โค้ดด้านบนใช้แนวคิดง่าย ๆ คือ

1. เริ่มจากข้อความ `"12,500"`
2. ลบ comma ออกด้วย `.replace(",", "")`
3. แปลงข้อความที่เหลือเป็นตัวเลขด้วย `int()`

รายละเอียดของ string method จะเรียนลึกขึ้นในบทถัดไป  
ตอนนี้ให้เห็นก่อนว่าชนิดข้อมูลมีผลต่อการคำนวณ

## 8. การสร้างข้อความจากตัวแปรด้วย f-string

ในการวิเคราะห์ข้อมูล เรามักต้องการสร้างข้อความสรุปผลลัพธ์ เช่น

- ยอดขายรวมเท่ากับ 125,000 บาท
- อัตรากำไรเท่ากับ 37.5%
- จังหวัดที่เลือกคือ Bangkok

Python มีวิธีสร้างข้อความจากตัวแปรที่อ่านง่าย เรียกว่า **f-string**

````python
total_sales = 125000
summary = f"ยอดขายรวมเท่ากับ {total_sales} บาท"

summary
````

การเขียน f-string ทำได้โดยใส่ตัวอักษร `f` หน้า quote แล้วใส่ชื่อตัวแปรไว้ใน `{}`

```python
f"ยอดขายรวมเท่ากับ {total_sales} บาท"
```

เราสามารถใส่นิพจน์ไว้ใน `{}` ได้ด้วย

````python
revenue = 120000
cost = 75000

profit = revenue - cost
profit_margin = profit / revenue

summary = f"กำไรเท่ากับ {profit} บาท และอัตรากำไรเท่ากับ {profit_margin:.2%}"

summary
````

ในตัวอย่างด้านบน

```python
{profit_margin:.2%}
```

หมายถึง แสดงค่า `profit_margin` เป็นเปอร์เซ็นต์ โดยมีทศนิยม 2 ตำแหน่ง

การจัดรูปแบบลักษณะนี้มีประโยชน์มากเวลาสร้างข้อความสรุปผลการวิเคราะห์
