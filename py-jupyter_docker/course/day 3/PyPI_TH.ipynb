## 1) PyPI คืออะไร?

**PyPI (Python Package Index)** คือ “คลังแพ็กเกจของ Python”  
ให้คิดเหมือน:
- App Store / Play Store แต่เป็นของ Python
- หรือเหมือนที่เราไปโหลด “เครื่องมือเสริม” มาใช้งาน

บน PyPI มีแพ็กเกจนับแสน เช่น:
- `pandas` สำหรับตารางข้อมูล
- `numpy` สำหรับคำนวณเชิงตัวเลข
- `requests` สำหรับเรียก API
- `openpyxl` สำหรับไฟล์ Excel

✅ งาน Data Engineer จะใช้แพ็กเกจเยอะมาก  
เพราะเราต้องเชื่อมต่อข้อมูลจากหลายแหล่งและจัดการหลายรูปแบบ

## 2) `pip` คืออะไร?

`pip` คือเครื่องมือสำหรับ “ติดตั้งแพ็กเกจจาก PyPI” ลงในสภาพแวดล้อม Python ของเรา

ตัวอย่างคำสั่ง (รันใน Terminal/Command Prompt):
```bash
pip install pandas
```

### เรื่องสำคัญที่มือใหม่สับสน
- ✅ `pip install ...` = ติดตั้งแพ็กเกจ (ทำใน Terminal หรือใน Notebook ผ่าน `!pip`)
- ✅ `import pandas` = เรียกใช้แพ็กเกจในโค้ด (ทำใน Python)

คนละขั้นตอนกัน แต่ต้องใช้คู่กัน

## 5) ตรวจสอบแพ็กเกจและเวอร์ชัน

Data Engineer ต้องระวังเรื่อง “เวอร์ชัน” เพราะ:
- เครื่อง A กับเครื่อง B เวอร์ชันไม่เท่ากัน → ผลลัพธ์/พฤติกรรมอาจต่างกัน
- งานจริงต้องทำให้รันซ้ำได้ (reproducible)

คำสั่งที่ใช้บ่อย:
- ดูเวอร์ชันของแพ็กเกจ: `package.__version__`
- ดูรายการแพ็กเกจ: `pip list`
- ดูข้อมูลแพ็กเกจ: `pip show package_name`

✅ ลองรันดูรายการแพ็กเกจบางส่วน

````python
!pip list --format=columns
````

## 6) Virtual Environment (แนวคิดสำคัญสำหรับ Data Engineer)

เวลาทำงานจริง เรามักมีหลายโปรเจกต์ เช่น
- โปรเจกต์ A ใช้ pandas เวอร์ชันหนึ่ง
- โปรเจกต์ B ใช้อีกเวอร์ชันหนึ่ง

ถ้าใช้ Python environment เดียวกันทั้งหมด อาจชนกันได้

**Virtual environment** คือ “แยกสภาพแวดล้อม” ต่อโปรเจกต์  
ทำให้แต่ละงานมีชุดแพ็กเกจของตัวเอง

คำสั่งพื้นฐาน (แนวทางหนึ่ง):
```bash
python -m venv .venv
# Windows
.venv\Scripts\activate
# macOS/Linux
source .venv/bin/activate

pip install pandas requests
```

📌 ในคลาส อาจยังไม่ต้องทำทุกขั้นตอน  
แต่ต้อง “เข้าใจแนวคิด” เพราะจำเป็นมากกับสาย Data Engineer

## 7) ปัญหาที่เจอบ่อย (Troubleshooting)

### กรณี 1: ติดตั้งแล้ว แต่ import ไม่ได้
สาเหตุยอดฮิต:
- ติดตั้งคนละ environment กับที่ Notebook ใช้อยู่

แนวทางแก้:
- ตรวจสอบว่า Notebook ใช้ kernel ไหน (Kernel → Change Kernel)
- ลองติดตั้งด้วย `!pip` ภายใน Notebook (เพื่อให้ตรง environment)

### กรณี 2: Permission denied / Access denied
- อาจต้องใช้ `pip install --user package`
- หรือขอสิทธิ์จากผู้ดูแลระบบ (เครื่ององค์กร)

### กรณี 3: ไม่มีอินเทอร์เน็ต / ติดตั้งไม่ได้
- งานองค์กรบางแห่งต้องตั้ง proxy หรือใช้ mirror ภายใน
- บางกรณีต้องดาวน์โหลดไฟล์ wheel มาก่อนแล้วติดตั้งแบบออฟไลน์
