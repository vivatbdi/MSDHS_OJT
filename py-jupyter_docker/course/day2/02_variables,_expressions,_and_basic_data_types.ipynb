## 1. ตัวแปรและการกำหนดค่า

**ตัวแปร** หรือ **variable** คือชื่อที่ใช้เก็บค่าไว้ใน Python

เมื่อเราต้องใช้ค่านั้นอีกครั้ง เราสามารถเรียกผ่านชื่อตัวแปรแทนการเขียนค่าซ้ำ

ตัวอย่างต่อไปนี้สร้างตัวแปรชื่อ `sales` เพื่อเก็บยอดขาย

````python
sales = 15000

sales
````

คำสั่ง

```python
sales = 15000
```

หมายถึง กำหนดค่า `15000` ให้กับตัวแปรชื่อ `sales`

เครื่องหมาย `=` ใน Python เรียกว่า **assignment operator** ใช้สำหรับกำหนดค่าให้ตัวแปร ไม่ได้ใช้ตรวจสอบว่าค่าสองค่าเท่ากันหรือไม่

### การนำตัวแปรกลับมาใช้

หลังจากสร้างตัวแปรแล้ว เราสามารถนำค่าที่เก็บไว้ไปคำนวณต่อได้

````python
sales = 15000
vat_rate = 0.07

vat_amount = sales * vat_rate

vat_amount
````

จากตัวอย่าง

- `sales` เก็บยอดขาย
- `vat_rate` เก็บอัตราภาษี
- `vat_amount` เก็บผลลัพธ์จากการคำนวณภาษี

การใช้ชื่อตัวแปรที่สื่อความหมายทำให้โค้ดอ่านและตรวจสอบได้ง่ายกว่าการเขียนตัวเลขโดยตรง

### การเปลี่ยนค่าของตัวแปร

ตัวแปรสามารถได้รับค่าใหม่ได้ เมื่อกำหนดค่าใหม่ ค่าก่อนหน้าจะถูกแทนที่

## 4. ลำดับการคำนวณ

Python ใช้ลำดับการคำนวณคล้ายคณิตศาสตร์ทั่วไป

1. วงเล็บ `()`
2. ยกกำลัง `**`
3. คูณ หาร หารปัดเศษ และหาเศษ `*`, `/`, `//`, `%`
4. บวกและลบ `+`, `-`

เมื่อสูตรมีหลายขั้นตอน ควรใช้วงเล็บเพื่อให้ความหมายชัดเจน

````python
result_without_parentheses = 10 + 5 * 2

result_without_parentheses
````

````python
result_with_parentheses = (10 + 5) * 2

result_with_parentheses
````

ผลลัพธ์ต่างกันเพราะ Python คำนวณการคูณก่อนการบวก

- `10 + 5 * 2` ได้ `20`
- `(10 + 5) * 2` ได้ `30`

ตัวอย่างสูตรในงานข้อมูลที่ควรใช้วงเล็บคือ

````python
revenue = 120000
cost = 75000

gross_profit_margin = (revenue - cost) / revenue

gross_profit_margin
````

### การจัดการตัวเลขที่มี comma

ข้อความ `"12,500"` ยังไม่สามารถแปลงเป็นจำนวนเต็มโดยตรง เพราะมี comma อยู่ภายใน

เราต้องลบ comma ก่อน แล้วจึงแปลงเป็นตัวเลข

````python
sales_text = "12,500"

clean_sales_text = sales_text.replace(",", "")
sales_number = int(clean_sales_text)

print(clean_sales_text)
print(sales_number)
print(type(sales_number))
````

ขั้นตอนการทำงานคือ

1. เริ่มจากข้อความ `"12,500"`
2. ใช้ `.replace(",", "")` ลบ comma
3. ได้ข้อความ `"12500"`
4. ใช้ `int()` แปลงเป็นจำนวนเต็ม `12500`

รายละเอียดเกี่ยวกับ String Method จะเรียนเพิ่มเติมในบทถัดไป

### รหัสไม่ควรถูกแปลงเป็นตัวเลขเสมอไป

ไม่ใช่ข้อมูลที่ประกอบด้วยตัวเลขทุกค่าจะต้องแปลงเป็น `int`

ตัวอย่างข้อมูลที่ควรเก็บเป็นข้อความ ได้แก่

- รหัสบุคคล
- รหัสครัวเรือน
- รหัสจังหวัด
- รหัสไปรษณีย์

เพราะข้อมูลเหล่านี้ใช้ระบุตัวตนหรือหมวดหมู่ ไม่ได้ใช้คำนวณทางคณิตศาสตร์

## 8. การสร้างข้อความด้วย f-string

ในการวิเคราะห์ข้อมูล เรามักต้องนำค่าจากตัวแปรมาสร้างข้อความสรุป เช่น

- จำนวนผู้ตอบทั้งหมดเท่ากับ 180 คน
- คะแนนเฉลี่ยเท่ากับ 4.25 คะแนน
- พื้นที่ที่เลือกคือ Bangkok

Python มีวิธีสร้างข้อความจากตัวแปรที่เรียกว่า **f-string**

````python
total_respondents = 180

summary = f"จำนวนผู้ตอบแบบสำรวจทั้งหมดเท่ากับ {total_respondents} คน"

summary
````

การเขียน f-string ประกอบด้วย

1. ใส่ตัวอักษร `f` หน้า quote
2. เขียนข้อความตามปกติ
3. ใส่ชื่อตัวแปรไว้ภายใน `{}`

รูปแบบพื้นฐานคือ

```python
f"ข้อความ {variable}"
```

### การแสดงจำนวนทศนิยม

````python
average_score = 4.1267

summary = f"คะแนนเฉลี่ยเท่ากับ {average_score:.2f} คะแนน"

summary
````

`:.2f` หมายถึง แสดงตัวเลขเป็นทศนิยม 2 ตำแหน่ง

### การแสดงค่าเป็นเปอร์เซ็นต์

หากค่าเก็บอยู่ในรูปอัตราส่วน เช่น `0.30` สามารถใช้ `:.2%` เพื่อแสดงเป็นเปอร์เซ็นต์ได้

````python
urgent_need_ratio = 0.30

summary = (
    f"สัดส่วนผู้ที่ต้องการความช่วยเหลือเร่งด่วน"
    f"เท่ากับ {urgent_need_ratio:.2%}"
)

summary
````

`:.2%` จะ

1. คูณค่าด้วย 100
2. แสดงเครื่องหมาย `%`
3. แสดงทศนิยม 2 ตำแหน่ง

ดังนั้นค่า `0.30` จะแสดงเป็น `30.00%`

### การจัดรูปแบบจำนวนเงินด้วย comma

````python
support_budget = 1250000

summary = f"งบประมาณช่วยเหลือทั้งหมดเท่ากับ {support_budget:,} บาท"

summary
````

การใช้ `:,` ช่วยเพิ่ม comma คั่นหลักพัน ทำให้ข้อความจำนวนเงินอ่านง่ายขึ้น

ผลลัพธ์คือ


## แบบฝึกหัดที่ 10: ประยุกต์ใช้ความรู้ทั้งหมด

ให้ทำงานต่อไปนี้

2. แปลง `households_with_children` เป็นจำนวนเต็ม
3. คำนวณสัดส่วนครัวเรือนที่มีเด็ก
4. คำนวณร้อยละครัวเรือนที่มีเด็ก
5. ตรวจสอบว่าคะแนนเฉลี่ยถึงเกณฑ์ความเสี่ยงสูงหรือไม่

```text
พื้นที่ Chiang Mai มีข้อมูลทั้งหมด 240 ครัวเรือน โดยมีครัวเรือนที่มีเด็ก 96 ครัวเรือน คิดเป็น 40.00% และสถานะความเสี่ยงสูงคือ False
```

````python
area_name = "Chiang Mai"
total_households = "240"
households_with_children = "96"
average_vulnerability_score = 3.85
high_risk_threshold = 4.0
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 10
>
> ขั้นตอนสำคัญคือ
>
> 1. แปลงจำนวนครัวเรือนจาก `str` เป็น `int`
> 2. คำนวณอัตราส่วนด้วยการหาร
> 3. คูณด้วย 100 เพื่อหาร้อยละ
> 4. เปรียบเทียบคะแนนกับเกณฑ์เพื่อสร้างค่า Boolean
> 5. นำผลลัพธ์ทั้งหมดมาสร้างข้อความด้วย f-string

````python
total_households_number = int(total_households)

households_with_children_number = int(
    households_with_children
)

households_with_children_ratio = (
    households_with_children_number
    / total_households_number
)

households_with_children_percent = (
    households_with_children_ratio * 100
)

is_high_risk = (
    average_vulnerability_score
    >= high_risk_threshold
)

summary_text = (
    f"พื้นที่ {area_name} มีข้อมูลทั้งหมด "
    f"{total_households_number} ครัวเรือน "
    f"โดยมีครัวเรือนที่มีเด็ก "
    f"{households_with_children_number} ครัวเรือน "
    f"คิดเป็น {households_with_children_percent:.2f}% "
    f"และสถานะความเสี่ยงสูงคือ {is_high_risk}"
)

summary_text
````
