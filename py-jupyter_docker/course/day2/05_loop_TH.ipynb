# 5) Looping (for-loop) สำหรับงานวิเคราะห์ข้อมูลเบื้องต้น

ในงานวิเคราะห์ข้อมูล เรามักต้อง “ทำสิ่งเดิมซ้ำ ๆ” กับข้อมูลหลายค่า เช่น
- คำนวณคะแนนทีละคน
- ตรวจว่าแต่ละค่า “ผ่าน/ไม่ผ่าน”
- รวมยอดขายทีละวัน
- นับจำนวนรายการที่เข้าเงื่อนไข

ใน Python เราใช้ **for-loop** เพื่อวนทำงานกับข้อมูลใน list ทีละค่า

---

## สิ่งที่คุณจะได้จากบทนี้
- เข้าใจแนวคิดการวนลูป (loop) และโครงสร้าง `for ... in ...`
- วนผ่าน list เพื่อแสดงผล/คำนวณ/สรุปข้อมูล
- ใช้ตัวแปรสะสม (accumulator) เช่น ผลรวม, จำนวนครั้ง
- ใช้ `range()` และ `enumerate()` แบบพื้นฐาน
- ทำแบบฝึกหัดสไตล์งาน data (รวมค่า, นับ, คัดกรอง)

---

## ภาพรวมเนื้อหา (ลำดับการเรียน)
1. Loop เทียบกับ Excel (ลากสูตรลงทั้งคอลัมน์ / Fill Down)
2. for-loop พื้นฐาน: วนทีละค่าใน list
3. สร้างผลรวม/ค่าเฉลี่ยด้วยลูป (accumulator)
4. นับจำนวนตามเงื่อนไข (count)
5. `range()` และ `enumerate()` แบบง่าย
6. แบบฝึกหัด: ผ่าน/ไม่ผ่าน, ยอดขาย, ค่าผิดปกติ

## A. Loop เทียบกับ Excel

ใน Excel เวลาเรามีสูตรในแถวแรก แล้วต้องคำนวณทุกแถว เรามัก:
- ลากสูตรลง (Fill Down)
- หรือใช้สูตรกับช่วงข้อมูลทั้งคอลัมน์

ใน Python แนวคิดคล้ายกัน:
- เรา “วน” ผ่านข้อมูลทีละค่า
- แล้วทำงานเดิมซ้ำ ๆ ให้ครบทุกค่า

✅ เริ่มจาก list ของคะแนน

````python
scores = [80, 75, 90, 60, 85]
scores
````

## B. for-loop พื้นฐาน: วนทีละค่าใน list

โครงสร้างพื้นฐาน:
```python
for ค่า in list:
    ทำอะไรบางอย่างกับค่า
```

✅ ลองรัน: พิมพ์คะแนนทีละตัว

````python
for s in scores:
    print(s)
````

สังเกตว่าเราไม่ต้องเขียน `print(scores[0])`, `print(scores[1])` ...  
ลูปช่วยให้ทำซ้ำได้อัตโนมัติ

## C. สร้างผลรวมด้วยลูป (Accumulator)

แม้ `sum(scores)` จะทำได้เลย  
แต่การทำด้วยลูปช่วยให้คุณเข้าใจ “วิธีคิด” และต่อยอดไปสรุปแบบอื่นได้

แนวคิด:
- เริ่มจาก `total = 0`
- วนทีละค่า แล้วบวกสะสมเข้าไป

✅ ลองรัน

````python
total = 0
for s in scores:
    total = total + s

total
````

### ค่าเฉลี่ยด้วยลูป

ค่าเฉลี่ย = ผลรวม / จำนวนข้อมูล

✅ ลองรัน

````python
avg = total / len(scores)
avg
````

## D. นับจำนวนตามเงื่อนไข (Count)

งาน data ที่พบบ่อย:
- นับจำนวนคนที่ “ผ่าน” (>= 50)
- นับจำนวนวันยอดขายเกินเป้า
- นับจำนวนค่าที่ผิดปกติ

เราจะใช้ตัวแปร `count` เพื่อสะสมจำนวน

✅ ลองรัน: นับจำนวนคะแนนที่มากกว่าหรือเท่ากับ 50

````python
count_pass = 0

for s in scores:
    if s >= 50:
        count_pass = count_pass + 1

count_pass
````

## E. สร้าง list ใหม่จากข้อมูลเดิม (คัดเฉพาะค่าที่สนใจ)

บ่อยครั้งเราต้อง “กรองข้อมูล” เช่น เอาเฉพาะคะแนนที่ผ่าน

✅ ลองรัน: สร้าง list ใหม่ชื่อ `passed_scores`

````python
passed_scores = []

for s in scores:
    if s >= 50:
        passed_scores.append(s)

passed_scores
````

## F. ใช้ range() เพื่อวนตามจำนวนครั้ง

`range(n)` ให้ตัวเลข 0 ถึง n-1  
ใช้เมื่อต้องการวน “จำนวนครั้ง” หรือวนตาม “ตำแหน่ง (index)”

✅ ลองรัน: วน 5 ครั้ง

````python
for i in range(5):
    print(i)
````

ตัวอย่างงาน data: ถ้าอยากพิมพ์ “ลำดับ” ของคะแนนด้วย (1, 2, 3, ...)

✅ ลองรัน

````python
for i in range(len(scores)):
    print("ลำดับที่", i+1, "คะแนน =", scores[i])
````

## G. ใช้ enumerate() เพื่อได้ทั้งลำดับและค่า (สะดวกกว่า)

`enumerate(list)` จะให้ทั้ง
- index (ลำดับเริ่มจาก 0)
- value (ค่าจริง)

✅ ลองรัน

````python
for i, s in enumerate(scores):
    print("index", i, "คะแนน =", s)
````

ถ้าอยากให้ลำดับเริ่มจาก 1 (เหมือน Excel แถวเริ่มที่ 1) สามารถทำแบบนี้:

✅ ลองรัน

````python
for i, s in enumerate(scores, start=1):
    print("ลำดับที่", i, "คะแนน =", s)
````

## H. แบบฝึกหัด 1: ยอดขายรายวัน (Daily Sales)

คุณมี list ของยอดขาย 7 วัน  
ให้คำนวณ:
1) ผลรวมยอดขาย (`total`)
2) ค่าเฉลี่ยต่อวัน (`avg`)

✅ ทำโดยใช้ for-loop (ไม่ใช้ `sum()` ในข้อนี้)

````python
daily_sales = [1200, 1500, 900, 1800, 2000, 1700, 1600]

# TODO: เขียนลูปหาผลรวม
total = 0
for x in daily_sales:
    total = total + x

avg = total / len(daily_sales)

print("Total =", total)
print("Average =", avg)
````

## I. แบบฝึกหัด 2: นับจำนวนวันที่ยอดขายเกินเป้า

ให้กำหนด “เป้ายอดขายต่อวัน” = 1600  
แล้วนับว่ามีกี่วันที่ยอดขาย **มากกว่าหรือเท่ากับ** เป้า

✅ ทำโดยใช้ for-loop + if


````python
target = 1600
count_meet_target = 0

for x in daily_sales:
    if x >= target:
        count_meet_target = count_meet_target + 1

print("Days meeting target =", count_meet_target)
````

## J. แบบฝึกหัด 3: คัดกรองค่าที่ผิดปกติ (Outliers แบบง่าย)

สมมติว่า “คะแนนติดลบ” เป็นค่าที่ผิดปกติ  
ให้สร้าง list ใหม่ชื่อ `bad_scores` เพื่อเก็บค่าที่ < 0

✅ ทำโดยใช้ for-loop + append

````python
scores2 = [80, 75, -10, 90, 60, -3]

bad_scores = []

for s in scores2:
    if s < 0:
        bad_scores.append(s)

print("Bad scores =", bad_scores)
````
