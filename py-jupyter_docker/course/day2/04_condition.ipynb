## 1. เงื่อนไขและค่า Boolean

**เงื่อนไข** หรือ **condition** คือข้อความที่ Python สามารถตรวจสอบได้ว่าเป็นจริงหรือเท็จ

ตัวอย่างคำถามที่เขียนเป็นเงื่อนไขได้ เช่น

- คะแนนมากกว่าหรือเท่ากับ 4.0 หรือไม่
- สถานะเท่ากับ `"Completed"` หรือไม่
- ข้อมูลได้รับความยินยอมหรือไม่
- พื้นที่อยู่ในรายการพื้นที่เป้าหมายหรือไม่

ผลลัพธ์ของเงื่อนไขเป็นค่า Boolean ได้แก่

- `True` — เงื่อนไขเป็นจริง
- `False` — เงื่อนไขเป็นเท็จ

````python
vulnerability_score = 4.2
threshold = 4.0

vulnerability_score >= threshold
````

เงื่อนไข

```python
vulnerability_score >= threshold
```

หมายถึง

> คะแนนความเปราะบางมากกว่าหรือเท่ากับเกณฑ์หรือไม่

เนื่องจาก `4.2` มากกว่า `4.0` ผลลัพธ์จึงเป็น `True`

````python
status = "Draft"

status == "Completed"
````

ผลลัพธ์เป็น `False` เพราะข้อความ `"Draft"` ไม่เท่ากับ `"Completed"`

ค่า Boolean สามารถเก็บไว้ในตัวแปรเพื่อนำไปใช้ต่อได้

````python
status = "Completed"

is_completed = status == "Completed"

is_completed
````

การตั้งชื่อตัวแปร Boolean มักใช้คำขึ้นต้นที่สื่อถึงคำถาม เช่น

- `is_completed`
- `is_high_risk`
- `has_urgent_need`
- `has_consent`
- `can_be_included`
- `should_follow_up`

ชื่อเหล่านี้ช่วยให้โค้ดอ่านคล้ายประโยคคำถามและเข้าใจได้ง่าย

## 2. ตัวดำเนินการเปรียบเทียบ

ตัวดำเนินการเปรียบเทียบ หรือ **comparison operators** ใช้เปรียบเทียบค่าตั้งแต่สองค่าขึ้นไป

| Operator | ความหมาย | ตัวอย่าง |
|---|---|---|
| `==` | เท่ากับ | `status == "Completed"` |
| `!=` | ไม่เท่ากับ | `status != "Cancelled"` |
| `>` | มากกว่า | `score > 4.0` |
| `<` | น้อยกว่า | `age < 18` |
| `>=` | มากกว่าหรือเท่ากับ | `score >= 4.0` |
| `<=` | น้อยกว่าหรือเท่ากับ | `delay_days <= 30` |

ข้อควรระวัง:

- `=` ใช้กำหนดค่าให้ตัวแปร
- `==` ใช้เปรียบเทียบว่าค่าสองค่าเท่ากันหรือไม่

````python
vulnerability_score = 3.8
threshold = 4.0

vulnerability_score == threshold
````

````python
vulnerability_score != threshold
````

````python
vulnerability_score < threshold
````

````python
vulnerability_score >= threshold
````

## 3. การใช้ `if`

คำสั่ง `if` ใช้เมื่อต้องการให้ Python ทำงานบางอย่างเฉพาะเมื่อเงื่อนไขเป็นจริง

รูปแบบพื้นฐานคือ

```python
if condition:
    statement
```

จุดสำคัญ ได้แก่

- หลังเงื่อนไขต้องมี colon `:`
- คำสั่งภายใต้ `if` ต้องเยื้องเข้าไป
- หากเงื่อนไขเป็น `False` คำสั่งภายในจะไม่ทำงาน

````python
vulnerability_score = 4.2
threshold = 4.0

if vulnerability_score >= threshold:
    print("Needs follow-up")
````

เนื่องจากเงื่อนไขเป็นจริง Python จึงแสดงข้อความ `"Needs follow-up"`

````python
vulnerability_score = 3.5
threshold = 4.0

if vulnerability_score >= threshold:
    print("Needs follow-up")
````

ตัวอย่างนี้ไม่แสดงผลลัพธ์ เพราะเงื่อนไขเป็น `False` และยังไม่มีคำสั่งสำหรับกรณีที่ไม่เข้าเงื่อนไข

### การเก็บผลลัพธ์จาก `if`

นอกจากแสดงข้อความแล้ว เรายังสามารถกำหนดค่าให้ตัวแปรภายในเงื่อนไข

````python
vulnerability_score = 4.2
threshold = 4.0

follow_up_status = "Normal"

if vulnerability_score >= threshold:
    follow_up_status = "Needs Follow-up"

follow_up_status
````

ในตัวอย่างนี้

1. กำหนดค่าเริ่มต้นของ `follow_up_status` เป็น `"Normal"`
2. ตรวจสอบคะแนน
3. หากคะแนนถึงเกณฑ์ จึงเปลี่ยนค่าเป็น `"Needs Follow-up"`

อย่างไรก็ตาม หากต้องการกำหนดผลลัพธ์ทั้งกรณีจริงและเท็จอย่างชัดเจน ควรใช้ `if-else`

## 4. การใช้ `if-else`

`if-else` ใช้เมื่อต้องการกำหนดทางเลือกสองทาง

- ทำอย่างหนึ่งเมื่อเงื่อนไขเป็นจริง
- ทำอีกอย่างเมื่อเงื่อนไขเป็นเท็จ

รูปแบบคือ

```python
if condition:
    statement_when_true
else:
    statement_when_false
```

````python
vulnerability_score = 3.5
threshold = 4.0

if vulnerability_score >= threshold:
    follow_up_status = "Needs Follow-up"
else:
    follow_up_status = "Normal"

follow_up_status
````

ตัวอย่างนี้คล้ายกับสูตร Excel

```excel
=IF(
    VulnerabilityScore>=Threshold,
    "Needs Follow-up",
    "Normal"
)
```

### ตัวอย่าง: ตรวจสอบความครบถ้วนของข้อมูล

````python
missing_field_count = 0

if missing_field_count == 0:
    data_status = "Complete"
else:
    data_status = "Incomplete"

data_status
````

### ตัวอย่าง: ตรวจสอบการให้ความยินยอม

````python
consent_given = False

if consent_given:
    consent_status = "Consent Given"
else:
    consent_status = "No Consent"

consent_status
````

## 5. การใช้ `if-elif-else`

หากมีผลลัพธ์มากกว่าสองระดับ สามารถใช้ `elif`

`elif` ย่อมาจาก **else if**

รูปแบบคือ

```python
if condition_1:
    statement_1
elif condition_2:
    statement_2
elif condition_3:
    statement_3
else:
    statement_when_no_condition_is_true
```

Python จะตรวจสอบเงื่อนไขจากบนลงล่าง และใช้เฉพาะเงื่อนไขแรกที่เป็นจริง

### ตัวอย่าง: จัดกลุ่มคะแนนความเปราะบาง

กำหนดเกณฑ์ดังนี้

| คะแนน | กลุ่ม |
|---:|---|
| ตั้งแต่ 4.0 ขึ้นไป | High |
| ตั้งแต่ 2.5 แต่น้อยกว่า 4.0 | Medium |
| น้อยกว่า 2.5 | Low |

````python
vulnerability_score = 3.6

if vulnerability_score >= 4.0:
    vulnerability_group = "High"
elif vulnerability_score >= 2.5:
    vulnerability_group = "Medium"
else:
    vulnerability_group = "Low"

vulnerability_group
````

Python ตรวจสอบตามลำดับดังนี้

1. `3.6 >= 4.0` เป็น `False`
2. `3.6 >= 2.5` เป็น `True`
3. กำหนดกลุ่มเป็น `"Medium"`
4. ไม่ตรวจสอบเงื่อนไขถัดไปอีก

````python
sales = 125000

if sales >= 100000:
    sales_group = "High"
elif sales >= 50000:
    sales_group = "Medium"
else:
    sales_group = "Low"

sales_group
````

### ตัวอย่าง: จัดระดับการติดตาม

````python
risk_score = 4.6

if risk_score >= 4.5:
    follow_up_level = "Immediate"
elif risk_score >= 3.5:
    follow_up_level = "Urgent"
elif risk_score >= 2.0:
    follow_up_level = "Standard"
else:
    follow_up_level = "Routine"

follow_up_level
````

## 6. การจัดลำดับเงื่อนไข

ลำดับของ `if-elif-else` มีความสำคัญ เพราะ Python ใช้เงื่อนไขแรกที่เป็นจริง

เมื่อใช้เกณฑ์แบบระดับ ควรเรียงจาก

> เกณฑ์สูงที่สุดไปยังเกณฑ์ต่ำที่สุด

หรือเรียงจากเงื่อนไขที่เฉพาะเจาะจงกว่าไปยังเงื่อนไขที่กว้างกว่า

### ตัวอย่างลำดับที่ไม่เหมาะสม

โค้ดต่อไปนี้ต้องการให้คะแนนตั้งแต่ 4.0 ขึ้นไปเป็น `"High"` แต่กำหนดเงื่อนไข `"Medium"` ไว้ก่อน

````python
vulnerability_score = 4.5

if vulnerability_score >= 2.5:
    vulnerability_group = "Medium"
elif vulnerability_score >= 4.0:
    vulnerability_group = "High"
else:
    vulnerability_group = "Low"

vulnerability_group
````

ผลลัพธ์คือ `"Medium"` เพราะเงื่อนไข

```python
vulnerability_score >= 2.5
```

เป็นจริงก่อน Python จึงไม่ตรวจสอบ `elif` ที่อยู่ถัดไป

### ลำดับที่ถูกต้อง

````python
vulnerability_score = 4.5

if vulnerability_score >= 4.0:
    vulnerability_group = "High"
elif vulnerability_score >= 2.5:
    vulnerability_group = "Medium"
else:
    vulnerability_group = "Low"

vulnerability_group
````

### การกำหนดขอบเขตของแต่ละช่วง

ในตัวอย่างต่อไปนี้

```python
if score >= 4.0:
    group = "High"
elif score >= 2.5:
    group = "Medium"
else:
    group = "Low"
```

เงื่อนไข `"Medium"` หมายถึงคะแนนตั้งแต่ `2.5` แต่น้อยกว่า `4.0` โดยอัตโนมัติ เพราะคะแนนตั้งแต่ `4.0` ถูกตรวจสอบในเงื่อนไขแรกแล้ว

จึงไม่จำเป็นต้องเขียนว่า

```python
elif score >= 2.5 and score < 4.0:
```

แม้รูปแบบดังกล่าวจะทำงานได้ก็ตาม

## 7. Logical operators

ในงานข้อมูลจริง เรามักต้องตรวจสอบมากกว่าหนึ่งเงื่อนไขพร้อมกัน

Python มี logical operators หลัก 3 ตัว ได้แก่

- `and`
- `or`
- `not`

### 7.1 `and`

ใช้ `and` เมื่อต้องการให้ทุกเงื่อนไขเป็นจริงพร้อมกัน

| เงื่อนไขที่ 1 | เงื่อนไขที่ 2 | ผลลัพธ์ |
|---|---|---|
| `True` | `True` | `True` |
| `True` | `False` | `False` |
| `False` | `True` | `False` |
| `False` | `False` | `False` |

````python
vulnerability_score = 4.3
has_urgent_need = True

is_priority = (
    vulnerability_score >= 4.0
    and has_urgent_need
)

is_priority
````

ผลลัพธ์เป็น `True` เพราะ

- คะแนนมากกว่าหรือเท่ากับ 4.0
- มีความต้องการเร่งด่วน

ทั้งสองเงื่อนไขเป็นจริงพร้อมกัน

````python
vulnerability_score = 4.3
has_urgent_need = False

is_priority = (
    vulnerability_score >= 4.0
    and has_urgent_need
)

is_priority
````

ผลลัพธ์เป็น `False` เพราะแม้คะแนนจะถึงเกณฑ์ แต่ `has_urgent_need` เป็น `False`

### 7.2 `or`

ใช้ `or` เมื่อต้องการให้อย่างน้อยหนึ่งเงื่อนไขเป็นจริง

````python
status = "Verified"

if status == "Completed" or status == "Verified":
    survey_status = "Valid"
else:
    survey_status = "Need Review"

survey_status
````

ผลลัพธ์เป็น `"Valid"` เพราะเงื่อนไขที่สองเป็นจริง แม้เงื่อนไขแรกจะเป็นเท็จก็ตาม

### 7.3 `not`

ใช้ `not` เพื่อกลับค่า Boolean

- `not True` ได้ `False`
- `not False` ได้ `True`

````python
is_duplicate = False

not is_duplicate
````

````python
is_duplicate = False

if not is_duplicate:
    duplicate_status = "Unique Record"
else:
    duplicate_status = "Duplicate Record"

duplicate_status
````

เงื่อนไข

```python
if not is_duplicate:
```

อ่านได้ว่า

> ถ้ารายการนี้ไม่ใช่ข้อมูลซ้ำ

### การรวม logical operators หลายตัว

เมื่อเงื่อนไขยาว ควรใช้วงเล็บและแยกบรรทัดเพื่อให้อ่านง่าย

````python
status = "Completed"
consent_given = True
is_duplicate = False

include_in_report = (
    status == "Completed"
    and consent_given
    and not is_duplicate
)

include_in_report
````

ตัวแปร `include_in_report` จะเป็น `True` เมื่อ

1. สถานะเป็น `"Completed"`
2. ได้รับความยินยอม
3. ไม่ใช่รายการซ้ำ

ทุกเงื่อนไขต้องเป็นจริงพร้อมกัน

## 8. Membership operators: `in` และ `not in`

Membership operators ใช้ตรวจสอบว่าค่าหนึ่งอยู่หรือไม่อยู่ในกลุ่มข้อมูล

- `in` — อยู่ในกลุ่ม
- `not in` — ไม่อยู่ในกลุ่ม

มักใช้ร่วมกับ List เช่น

- รายชื่อพื้นที่เป้าหมาย
- รายชื่อสถานะที่ยอมรับ
- รายชื่อประเภทผู้รับบริการ
- รายชื่อค่าที่ต้องติดตาม

## 9. เงื่อนไขแบบซ้อน

เงื่อนไขแบบซ้อน หรือ **nested condition** คือการเขียน `if` ภายใน `if` อีกชั้นหนึ่ง

เหมาะเมื่อการตรวจสอบเงื่อนไขลำดับถัดไปควรเกิดขึ้นเฉพาะเมื่อเงื่อนไขแรกเป็นจริง

````python
has_urgent_need = True
vulnerability_score = 3.7

if has_urgent_need:
    if vulnerability_score >= 4.0:
        follow_up_level = "Immediate Follow-up"
    else:
        follow_up_level = "Follow-up"
else:
    follow_up_level = "Routine"

follow_up_level
````

Logic ของตัวอย่างคือ

- หากมีความต้องการเร่งด่วน จึงตรวจสอบคะแนนเพิ่มเติม
- หากคะแนนตั้งแต่ 4.0 ขึ้นไป ให้ติดตามทันที
- หากคะแนนต่ำกว่า 4.0 ให้ติดตามตามปกติ
- หากไม่มีความต้องการเร่งด่วน ให้ติดตามตามรอบ

Nested condition สามารถใช้ได้ แต่ไม่ควรซ้อนหลายชั้นเกินไป เพราะจะทำให้โค้ดอ่านยาก

### ทางเลือกที่ไม่ใช้ nested condition

Logic เดียวกันสามารถเขียนด้วย `if-elif-else` และ `and`

````python
has_urgent_need = True
vulnerability_score = 3.7

if has_urgent_need and vulnerability_score >= 4.0:
    follow_up_level = "Immediate Follow-up"
elif has_urgent_need:
    follow_up_level = "Follow-up"
else:
    follow_up_level = "Routine"

follow_up_level
````

- อ่านง่าย

- ไม่ซับซ้อนเกินความจำเป็น
