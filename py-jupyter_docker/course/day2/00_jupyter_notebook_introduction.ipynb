# การใช้งาน Jupyter Notebook เบื้องต้น

ก่อนเริ่มเรียน Python เราจะทำความเข้าใจเครื่องมือที่ใช้ในการเรียน คือ **Jupyter Notebook**

Notebook นี้จะเป็นพื้นที่สำหรับ:
- เขียนโค้ด
- ดูผลลัพธ์
- จดบันทึกและอธิบายการวิเคราะห์

##ผลการเรียนรู้ที่คาดหวัง

- ผู้เรียนสามารถอธิบายองค์ประกอบพื้นฐานของ Jupyter Notebook ได้แก่ Cell, Code Cell และ Markdown Cell ได้
- ผู้เรียนสามารถรันคำสั่งใน Code Cell และตรวจสอบผลลัพธ์ (Output) ได้อย่างถูกต้อง
- ผู้เรียนสามารถแก้ไขและทดลองปรับเปลี่ยนคำสั่งใน Notebook ได้อย่างเหมาะสม
- ผู้เรียนมีความเข้าใจรูปแบบการทำงานแบบ เขียน → รัน → ตรวจสอบผลลัพธ์ ในบริบทของ Jupyter Notebook

## Cell คืออะไร

Jupyter Notebook แบ่งเนื้อหาออกเป็น “Cell”

มี 2 ประเภทหลัก:
- **Code Cell**: ใช้เขียนและรัน Python  
- **Markdown Cell**: ใช้เขียนข้อความอธิบาย

เราจะใช้ cell ทั้งสองแบบร่วมกันตลอดบทเรียนนี้

```python
print("นี่คือ Code Cell")
```
