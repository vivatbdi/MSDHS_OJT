Python กับงานวิเคราะห์ข้อมูล

Python เป็นภาษาโปรแกรมที่ได้รับความนิยมในงานวิเคราะห์ข้อมูล เนื่องจาก:

- เขียนคำสั่งเพื่อทำงานซ้ำได้โดยอัตโนมัติ

***เปรียบเทียบการทำงาน: Excel vs Python***

| การทำงาน | Excel | Python |
|----------|-------|--------|
| การเก็บข้อมูล | Cell | Variable |
| การคำนวณ | Formula | Expression |
| การทำซ้ำ | Copy สูตร | Loop |
| การจัดการข้อมูล | Table | List / Structure |

Python ไม่ได้แทนที่ Excel แต่เป็น “เครื่องมือเสริม”  
ที่ช่วยให้เราทำงานได้ **ยืดหยุ่นและเป็นระบบมากขึ้น**

```python
# ตัวอย่างข้อมูลยอดขาย (หน่วย: บาท)
sales = [1200, 1500, 1800, 1100]

# คำนวณยอดรวม
total_sales = sum(sales)

print(total_sales)
```

ในตัวอย่างนี้:

- `sales` เปรียบเสมือน “คอลัมน์ข้อมูล” ใน Excel  
- `sum(sales)` ทำหน้าที่เหมือนสูตร `=SUM()`  
- ผลลัพธ์ที่แสดงออกมา คือค่าที่ Python คำนวณให้  

นี่คือรูปแบบพื้นฐานของการใช้ Python ในงานข้อมูล

ถ้าเราต้องการเปลี่ยนแปลงข้อมูล เราจะต้องทำอย่างไร?

ใน Excel:
- มักแก้ค่าที่ cell โดยตรง

ใน Python:
- เราจะแก้ “คำสั่ง” แล้วรันใหม่

```python
# แก้ไขข้อมูลยอดขาย
sales = [1200, 1500, 1800, 2000]

# คำนวณใหม่
total = sum(sales)

print(total)
```

จากตัวอย่างนี้ เราจะเห็นรูปแบบการทำงานดังนี้:

1. เขียนคำสั่ง (Code)  
2. รัน (Run)  
3. ดูผลลัพธ์ (Output)  
4. ปรับปรุง (Edit)  

แล้วทำซ้ำอีกครั้ง

นี่เป็นข้อแตกต่างของการทำงานของ python และ Excel

- **Excel**: แก้ไขข้อมูลใน cell แล้วผลลัพธ์เปลี่ยนทันที  
- **Python**: แก้ไข “คำสั่ง” แล้วรันใหม่เพื่อดูผลลัพธ์ 

### การทำงานกับข้อมูลเบื้องต้นด้วย Python

เราจะเริ่มต้นใช้ Python กับ “ข้อมูลจริง” โดยเชื่อมโยงกับสิ่งที่ผู้เรียนคุ้นเคยจาก **Microsoft Excel**

เปรียบเทียบการวิเคราะห์ข้อมูล: Excel vs Python

สมมติว่าใน Excel เรามีข้อมูลยอดขาย:

| | A | B |
|---|---|---|
|1| เดือน | ยอดขาย |
|2| ม.ค. | 1200 |
|3| ก.พ. | 1500 |
|4| มี.ค. | 1800 |

และเราสามารถใช้สูตรเพื่อหายอดรวม:
> =SUM(B2:B4)

ใน Python เราสามารถเขียนได้ดังนี้:

```python
# ข้อมูลยอดขาย (หน่วย: บาท)
sales = [1200, 1500, 1800, 2000]
```

```python
# คำนวณยอดรวม
total = sum(sales)

print(total)
```

- `sales` คือ “ข้อมูลหลายค่า” → คล้ายคอลัมน์ใน Excel  
- `[1200, 1500, 1800]` คือ “รายการของตัวเลข” (List)  
- `sum()` ทำหน้าที่เหมือน `SUM()` ใน Excel

เราไม่จำเป็นต้องเข้าใจทุกอย่างในตอนนี้ ตัวอย่างนี้ต้องการแสดงให้เห็นว่า Python สามารถทำสิ่งเดียวกับ Excel ได้

```python
# หาจำนวนข้อมูล
count = len(sales)

print(count)
```

เทียบการวิเคราะห์ข้อมูลโดยการคำนวน

| EXCEL | PYTHON |
|------|--------|
| =SUM() | sum() |
| =COUNT() | len() |
