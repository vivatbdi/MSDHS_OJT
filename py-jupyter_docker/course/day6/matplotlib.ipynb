```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
```

```python
df = pd.read_csv(url)
```

```python
bill_np = df["total_bill"].to_numpy()
```

```python
plt.plot(bill_np)
plt.show()
```

```python
plt.plot(bill_np)
plt.xlabel("Observation index")
plt.ylabel("Total bill ($)")
plt.title("Total Bill Values")
plt.show()
```

```python
tip_np = df["tip"].to_numpy()
```

```python
plt.scatter(bill_np, tip_np)
plt.xlabel("Total bill ($)")
plt.ylabel("Tip ($)")
plt.title("Tip vs Total Bill")
plt.show()
```

```python
tip_pct_mean = np.mean(tip_np/bill_np)
```

```python
plt.scatter(bill_np, tip_np)
plt.axline((0, 0), slope=tip_pct_mean, linestyle="--")
plt.xlabel("Total bill ($)")
plt.ylabel("Tip ($)")
plt.title("Tip vs Total Bill (mean tip %)")
plt.show()
```

```python
plt.hist(bill_np, bins=20)
plt.xlabel("Total bill ($)")
plt.ylabel("Frequency")
plt.title("Distribution of Total Bill")
plt.show()
```

```python
smoker_np = df["smoker"].to_numpy()
```

```python
plt.hist(bill_np[smoker_np == "Yes"], bins=20, alpha=0.7, label="Smoker")
plt.hist(bill_np[smoker_np == "No"], bins=20, alpha=0.7, label="Non-smoker")

plt.xlabel("Total bill ($)")
plt.ylabel("Frequency")
plt.title("Total Bill Distribution by Smoker Status")
plt.legend()
plt.show()
```
