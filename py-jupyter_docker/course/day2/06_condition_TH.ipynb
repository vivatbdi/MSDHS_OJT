# 6) Conditions (if-else) สำหรับงานวิเคราะห์ข้อมูลเบื้องต้น

ในการวิเคราะห์ข้อมูล เราไม่ได้แค่คำนวณค่า  
แต่เราต้อง **ตัดสินใจจากข้อมูล** เช่น
- คะแนนนี้ “ผ่านหรือไม่ผ่าน”
- ยอดขายวันนี้ “ถึงเป้าหรือยัง”
- ค่านี้ “ผิดปกติหรือไม่”

ใน Python เราใช้ **เงื่อนไข (condition)** ด้วย `if`, `elif`, `else`  
เพื่อให้โค้ด “ตัดสินใจ” ได้เหมือนที่เราคิดตอนดูข้อมูล

---

## สิ่งที่คุณจะได้จากบทนี้

- สร้างเงื่อนไขแบบหลายกรณี (if–elif–else)
- ทำแบบฝึกหัดสไตล์งาน data (ผ่าน/ไม่ผ่าน, ถึงเป้า, ค่าผิดปกติ)

## A. Conditions เทียบกับ Excel

ใน Excel เรามักใช้สูตร:
`=IF(A1 >= 50, "Pass", "Fail")`

ใน Python เราเขียนเงื่อนไขด้วย `if-else`  
แนวคิดเหมือนกัน:
- ถ้าเงื่อนไขเป็นจริง → ทำอย่างหนึ่ง
- ถ้าไม่จริง → ทำอีกอย่างหนึ่ง

✅ เริ่มจากตัวอย่างง่าย ๆ

```python
score = 75

if score >= 50:
    print("Pass")
else:
    print("Fail")
```

## B. เครื่องหมายเปรียบเทียบที่ใช้บ่อย

ในงาน data เราใช้เครื่องหมายเหล่านี้บ่อยมาก:
- `>` มากกว่า
- `<` น้อยกว่า
- `>=` มากกว่าหรือเท่ากับ
- `<=` น้อยกว่าหรือเท่ากับ
- `==` เท่ากับ
- `!=` ไม่เท่ากับ

✅ ลองรันและดูผลลัพธ์ (True / False)

```python
x = 10

print(x > 5)
```

```python
print(x < 5)
```

```python
print(x == 10)
```

```python
print(x != 10)
```

## C. if + loop = วิเคราะห์ข้อมูลหลายค่า

จุดพลังของ Python คือการใช้ `if` ร่วมกับ `for-loop`  
เพื่อวิเคราะห์ข้อมูลทั้งชุด (เหมือนใช้ IF กับหลายแถวใน Excel)

✅ ตัวอย่าง: ตรวจผ่าน/ไม่ผ่านของคะแนนหลายคน

```python
scores = [80, 45, 90, 30, 60]

for s in scores:
    if s >= 50:
        print(s, "-> Pass")
    else:
        print(s, "-> Fail")
```

## D. นับจำนวนตามเงื่อนไข (Count)

สถานการณ์งานจริง:
- อยากรู้ว่ามีกี่คน “ผ่าน”
- หรือมีกี่วันที่ยอดขาย “ถึงเป้า”

เราจะใช้ตัวแปรสะสมจำนวน

✅ ลองรัน

```python
count_pass = 0

for s in scores:
    if s >= 50:
        count_pass = count_pass + 1

count_pass
```

## E. หลายเงื่อนไขด้วย if–elif–else

บางครั้งข้อมูลมีหลายระดับ เช่น เกรด
- >= 80 → A
- >= 70 → B
- >= 60 → C
- อื่น ๆ → F

ใน Python ใช้ `elif` เพื่อเขียนให้อ่านง่าย

✅ ลองรัน

```python
score = 72

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

grade
```

## F. ใช้ if–elif–else กับข้อมูลหลายค่า

นำตัวอย่างเกรดไปใช้กับ list ของคะแนน

✅ ลองรัน

```python
scores = [85, 72, 64, 40, 90]

for s in scores:
    if s >= 80:
        grade = "A"
    elif s >= 70:
        grade = "B"
    elif s >= 60:
        grade = "C"
    else:
        grade = "F"
    
    print(s, "->", grade)
```

## G. การใช้งาน Boolean logic operator เพื่อเพิ่มเงื่อนไข

ในงานวิเคราะห์ข้อมูลจริง เงื่อนไขมักไม่ได้มีแค่ข้อเดียว  
แต่มีหลายเงื่อนไขประกอบกัน เช่น

- อายุ > 18 **และ** รายได้ > 30000  
- คะแนน < 0 **หรือ** > 100  
- ลูกค้าเป็นสมาชิกหรือไม่  

เราจึงต้องใช้ **Boolean logic operator** เพื่อเชื่อมหลายเงื่อนไขเข้าด้วยกัน

---

### Operator ที่ใช้บ่อย

#### 1) `and` → ต้องเป็นจริงทั้งสองเงื่อนไข

ใช้เมื่อเราต้องการให้ "ทุกเงื่อนไข" เป็นจริงพร้อมกัน

ตัวอย่าง: ต้องอายุเกิน 18 และมีรายได้เกิน 30000

### ตัวอย่างการใช้ `and`

ทั้งสองเงื่อนไขต้องเป็น True ถึงจะได้ผลลัพธ์เป็น True

```python
age = 25
income = 40000

age > 18 and income > 30000
```

### การใช้ `or` → อย่างน้อยหนึ่งเงื่อนไขเป็นจริง

ใช้เมื่อเข้าเงื่อนไขใดเงื่อนไขหนึ่งก็เพียงพอ

ตัวอย่าง: คะแนนที่ถือว่าผิดปกติ  
คือ น้อยกว่า 0 หรือ มากกว่า 100

```python
score = -5

score < 0 or score > 100
```

### การใช้ `not` → กลับค่าความจริง

`not` ใช้สำหรับ "กลับค่า"  
- ถ้าเดิมเป็น True → จะกลายเป็น False  
- ถ้าเดิมเป็น False → จะกลายเป็น True  

มีประโยชน์มากเมื่อเราต้องการตรวจว่า  
“ไม่เข้าเงื่อนไขผิดปกติ” เช่น ข้อมูลปกติ

```python
is_member = True

not is_member
```

### ใช้ Boolean Logic กับข้อมูลหลายแถว

ตัวอย่างสถานการณ์จริง:
เราต้องการตรวจสอบคะแนนหลายค่า ว่าค่าไหนผิดปกติ

เงื่อนไข:
- คะแนน < 0 หรือ > 100 → Abnormal
- นอกนั้น → Normal

สังเกตว่าเราใช้ `or` เพื่อรวมเงื่อนไขผิดปกติ 2 แบบเข้าด้วยกัน

```python
scores3 = [80, -5, 95, 120, 60]

for s in scores3:
    if s < 0 or s > 100:
        print(s, "-> Abnormal")
    else:
        print(s, "-> Normal")
```

## H. แบบฝึกหัด 1: ตรวจสอบยอดขายแบบ “ถึงเป้า + ไม่ผิดปกติ”

สถานการณ์:
- มียอดขายรายวันใน list
- เป้ายอดขายต่อวัน = 1600
- ยอดขายที่ “ผิดปกติ” คือ น้อยกว่า 0 หรือ มากกว่า 5000

ให้เขียนเงื่อนไขโดยใช้ **and / or / not** เพื่อแสดงผลว่าแต่ละวันเป็นอะไร:

- **Abnormal** → ถ้ายอดขาย < 0 **หรือ** > 5000  
- **Meet target** → ถ้า **ไม่ผิดปกติ** และ ยอดขาย >= เป้า  
- **Not meet target** → ถ้า **ไม่ผิดปกติ** และ ยอดขาย < เป้า

```python
daily_sales = [1200, 1500, 900, 1800, 2000, 1700, 1600, -50, 6000]
target = 1600

for x in daily_sales:
    is_abnormal = (x < 0) or (x > 5000)

    if is_abnormal:
        print(x, "-> Abnormal")
    elif (not is_abnormal) and (x >= target):
        print(x, "-> Meet target")
    else:
        print(x, "-> Not meet target")
```

## I. แบบฝึกหัด 2: สรุปผลคะแนนแบบ “ปกติ/ผิดปกติ + ผ่าน/ไม่ผ่าน”

สถานการณ์:
- คะแนนที่ “ผิดปกติ” คือ < 0 หรือ > 100
- ถ้าคะแนน “ปกติ” (ไม่ผิดปกติ) ให้ตัดสินว่า
  - Pass ถ้า >= 50
  - Fail ถ้าน้อยกว่า 50

ให้ใช้ **or** เพื่อหา “ผิดปกติ” และใช้ **not** เพื่อหา “ปกติ”

```python
scores2 = [80, 75, -10, 90, 105, 60, 49]

for s in scores2:
    is_abnormal = (s < 0) or (s > 100)
    is_normal = not is_abnormal

    if is_abnormal:
        print(s, "-> Abnormal")
    else:
        if s >= 50:
            print(s, "-> Normal, Pass")
        else:
            print(s, "-> Normal, Fail")
```
