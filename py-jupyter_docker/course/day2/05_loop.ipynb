## 1. Loop คืออะไร

**Loop** คือการสั่งให้ Python ทำงานซ้ำตามเงื่อนไขหรือรายการข้อมูลที่กำหนด

ตัวอย่างเช่น ถ้าเรามีรายการยอดขายหลายรายการ

```python
sales_list = [12000, 18000, 9500, 22000]
```

แทนที่จะเขียนคำสั่งคำนวณทีละรายการ เราสามารถใช้ loop เพื่อให้ Python ทำงานซ้ำกับทุกค่าใน list ได้

## ตัวอย่างแนวคิด

ถ้าต้องการแสดงยอดขายแต่ละรายการ

ใน Excel เราอาจดูข้อมูลทีละแถว หรือ copy สูตรลงหลายแถว

ใน Python เราเขียนว่า

```python
for sales in sales_list:
    print(sales)
```

ความหมายคือ

สำหรับยอดขายแต่ละค่าใน `sales_list` ให้ทำคำสั่งที่อยู่ภายใต้ loop

````python
sales_list = [12000, 18000, 9500, 22000]

for sales in sales_list:
    print(sales)
````

จากตัวอย่างด้านบน

- `sales_list` คือข้อมูลหลายค่าที่ต้องการวนซ้ำ
- `sales` คือชื่อตัวแปรชั่วคราวที่ใช้แทนค่าทีละตัว
- บรรทัดที่เยื้องเข้าไปคือคำสั่งที่ต้องการให้ทำซ้ำ

Python จะหยิบค่าจาก `sales_list` มาทีละตัว แล้วนำมาเก็บในตัวแปร `sales`

## 2. การใช้ `for` loop กับ list

รูปแบบพื้นฐานของ `for` loop คือ

```python
for item in list_name:
    do_something
```

ความหมาย คือ

สำหรับแต่ละ `item` ที่อยู่ใน `list_name` ให้ทำ `do_something`

````python
products = ["Keyboard", "Mouse", "Monitor", "Printer"]

for product in products:
    print(product)
````

ตัวแปรหลังคำว่า `for` สามารถตั้งชื่อได้ตามความหมายของข้อมูล

ตัวอย่างเช่น

```python
for product in products:
```

อ่านง่ายกว่า

```python
for x in products:
```

เพราะรู้ทันทีว่าแต่ละค่าที่วนซ้ำคือสินค้า

````python
regions = ["Bangkok", "Chiang Mai", "Khon Kaen", "Phuket"]

for region in regions:
    print(f"Region: {region}")
````

## 3. การใช้ Loop เพื่อคำนวณข้อมูลหลายรายการ

สมมติว่ามียอดขายหลายรายการ และต้องการคำนวณภาษีมูลค่าเพิ่ม 7% ให้ทุกรายการ

ใน Excel เราอาจเขียนสูตรในแถวแรก แล้ว drag formula ลงมาหลายแถว

ใน Python เราสามารถใช้ loop เพื่อทำงานแบบเดียวกัน

````python
sales_list = [12000, 18000, 9500, 22000]
vat_rate = 0.07

for sales in sales_list:
    vat_amount = sales * vat_rate
    print(vat_amount)
````

ตัวอย่างด้านบนแสดงผลภาษีของแต่ละยอดขาย

แต่ถ้าเราต้องการเก็บผลลัพธ์ไว้ใช้ต่อ ควรสร้าง list ใหม่เพื่อเก็บผลลัพธ์

## 4. การสร้าง List ใหม่จาก Loop

รูปแบบที่ใช้บ่อยในงานข้อมูลคือ

1. สร้าง list ว่างสำหรับเก็บผลลัพธ์
2. วนซ้ำข้อมูลต้นทาง
3. คำนวณหรือแปลงค่า
4. เพิ่มผลลัพธ์เข้า list ใหม่ด้วย `.append()`

````python
sales_list = [12000, 18000, 9500, 22000]
vat_rate = 0.07

vat_amounts = []

for sales in sales_list:
    vat_amount = sales * vat_rate
    vat_amounts.append(vat_amount)

vat_amounts
````

จากตัวอย่างนี้

- `vat_amounts = []` คือ list ว่างสำหรับเก็บผลลัพธ์
- loop จะคำนวณภาษีทีละรายการ
- `.append()` ใช้เพิ่มผลลัพธ์เข้าไปใน list

แนวคิดนี้คล้ายกับการสร้างคอลัมน์ใหม่ใน Excel ที่คำนวณจากคอลัมน์เดิม

### ตัวอย่าง: คำนวณราคาหลังรวม VAT

````python
prices = [1000, 2500, 3200, 4500]
vat_rate = 0.07

prices_with_vat = []

for price in prices:
    final_price = price * (1 + vat_rate)
    prices_with_vat.append(final_price)

prices_with_vat
````

### ตัวอย่าง: สร้างข้อความสรุปหลายรายการ

````python
products = ["Keyboard", "Mouse", "Monitor"]
sales_list = [12000, 8500, 22000]

summary_messages = []

for product in products:
    message = f"Product: {product}"
    summary_messages.append(message)

summary_messages
````

ตัวอย่างด้านบนยังใช้เฉพาะรายชื่อสินค้า

ถ้าต้องการใช้ข้อมูลจากหลาย list พร้อมกัน เช่น สินค้าและยอดขาย จะใช้ `zip()`

````python
products = ["Keyboard", "Mouse", "Monitor"]
sales_list = [12000, 8500, 22000]

summary_messages = []

for product, sales in zip(products, sales_list):
    message = f"{product}: sales = {sales}"
    summary_messages.append(message)

summary_messages
````

`zip()` ใช้จับคู่ข้อมูลจากหลาย list ตามตำแหน่ง

ตัวอย่างนี้จับคู่

| product | sales |
|---|---:|
| Keyboard | 12000 |
| Mouse | 8500 |
| Monitor | 22000 |

แนวคิดนี้คล้ายกับการทำงานกับข้อมูลหลายคอลัมน์ใน Excel แถวเดียวกัน

## 5. การใช้ Loop ร่วมกับ Condition

ในงานข้อมูล เรามักต้องตรวจสอบเงื่อนไขกับข้อมูลหลายรายการ

เช่น

- ยอดขายแต่ละรายการถึงเป้าหมายหรือไม่
- รายการใดควรนำเข้า report
- ลูกค้าแต่ละคนอยู่ในกลุ่มใด
- สถานะรายการใดต้องตรวจสอบเพิ่มเติม

เราสามารถใช้ loop ร่วมกับ `if`, `elif`, `else` ได้

````python
sales_list = [120000, 85000, 45000, 150000]

sales_groups = []

for sales in sales_list:
    if sales >= 100000:
        group = "High"
    elif sales >= 50000:
        group = "Medium"
    else:
        group = "Low"
    
    sales_groups.append(group)

sales_groups
````

ตัวอย่างนี้คล้ายกับการใช้สูตร `IFS` ใน Excel แล้ว drag formula ลงหลายแถว

แต่ใน Python เราเขียน logic เพียงครั้งเดียว แล้วใช้ loop ทำซ้ำกับทุกค่าใน list

### ตัวอย่าง: ตรวจสอบสถานะ transaction หลายรายการ

````python
statuses = ["Completed", "Approved", "Cancelled", "Pending", "Completed"]
valid_statuses = ["Completed", "Approved"]

transaction_results = []

for status in statuses:
    if status in valid_statuses:
        result = "Valid"
    else:
        result = "Need Review"
    
    transaction_results.append(result)

transaction_results
````

ในงานจริง logic แบบนี้อาจใช้เพื่อตรวจสอบข้อมูลจำนวนมาก เช่น transaction หลายพันรายการ

สำหรับบทนี้ เราใช้ list ขนาดเล็กเพื่อให้เห็นแนวคิดก่อน

## 6. การใช้ `range()`

`range()` ใช้สร้างลำดับตัวเลขสำหรับการวนซ้ำ

เหมาะเมื่อเราต้องการทำงานซ้ำตามจำนวนรอบที่กำหนด เช่น

- ทำซ้ำ 5 ครั้ง
- สร้างเลขลำดับ
- ประมวลผลตาม index

````python
for number in range(5):
    print(number)
````

ผลลัพธ์คือเลข 0 ถึง 4

เพราะ `range(5)` เริ่มจาก 0 และหยุดก่อน 5

````python
for month in range(1, 13):
    print(f"Month {month}")
````

`range(1, 13)` หมายถึงเริ่มจาก 1 และหยุดก่อน 13

จึงได้เลข 1 ถึง 12 ซึ่งเหมาะกับการแทนเดือน 1 ถึงเดือน 12

### ตัวอย่าง: สร้างชื่อไฟล์รายเดือน

````python
monthly_files = []

for month in range(1, 13):
    file_name = f"sales_{month}.xlsx"
    monthly_files.append(file_name)

monthly_files
````

ตัวอย่างนี้แสดงให้เห็นว่า loop สามารถช่วยสร้างรายการซ้ำ ๆ ได้

ถ้าทำใน Excel เราอาจต้องพิมพ์หรือ drag pattern แต่ใน Python เราสามารถสร้างด้วย code

## 7. การใช้ `enumerate()`

บางครั้งเราต้องการทั้ง

- ลำดับของรายการ
- ค่าของรายการ

Python มี `enumerate()` เพื่อช่วยงานนี้

````python
products = ["Keyboard", "Mouse", "Monitor"]

for index, product in enumerate(products):
    print(index, product)
````

โดยปกติ `enumerate()` จะเริ่มนับจาก 0

ถ้าต้องการให้เริ่มนับจาก 1 สามารถกำหนด `start=1`

````python
products = ["Keyboard", "Mouse", "Monitor"]

for row_number, product in enumerate(products, start=1):
    print(f"{row_number}. {product}")
````

แนวคิดนี้มีประโยชน์เมื่อเราต้องการสร้างลำดับรายการ คล้ายเลขแถวหรือลำดับในรายงาน

## 8. การใช้ Loop กับ Dictionary

Dictionary เก็บข้อมูลเป็นคู่ของ key และ value

เมื่อต้องการวนซ้ำ dictionary เราสามารถใช้ `.items()` เพื่อได้ทั้ง key และ value

````python
sales_by_region = {
    "Bangkok": 150000,
    "Chiang Mai": 85000,
    "Khon Kaen": 62000
}

for region, sales in sales_by_region.items():
    print(region, sales)
````

ตัวอย่างนี้เหมาะกับข้อมูลที่มีลักษณะเป็น “ชื่อกลุ่ม → ค่า”

เช่น

- ภูมิภาค → ยอดขาย
- สินค้า → จำนวนขาย
- สถานะ → จำนวนรายการ

### ตัวอย่าง: สร้างข้อความสรุปจาก dictionary

````python
sales_by_region = {
    "Bangkok": 150000,
    "Chiang Mai": 85000,
    "Khon Kaen": 62000
}

summary_messages = []

for region, sales in sales_by_region.items():
    message = f"{region}: total sales = {sales}"
    summary_messages.append(message)

summary_messages
````

### ตัวอย่าง: ตรวจสอบยอดขายแต่ละภูมิภาคเทียบเป้าหมาย

````python
sales_by_region = {
    "Bangkok": 150000,
    "Chiang Mai": 85000,
    "Khon Kaen": 62000
}

target = 100000

region_status = {}

for region, sales in sales_by_region.items():
    if sales >= target:
        status = "Achieved"
    else:
        status = "Below Target"
    
    region_status[region] = status

region_status
````

ตัวอย่างนี้สร้าง dictionary ใหม่ชื่อ `region_status`

โดยใช้

- key คือชื่อภูมิภาค
- value คือสถานะเทียบเป้าหมาย

แนวคิดนี้คล้ายกับการสร้างคอลัมน์สถานะใหม่จากข้อมูลเดิมใน Excel

## 9. Loop กับงานไฟล์หลายไฟล์

เช่น

- ไฟล์ยอดขายรายเดือน
- ไฟล์ข้อมูลจากแต่ละสาขา
- ไฟล์ข้อมูลจากแต่ละหน่วยงาน
- ไฟล์รายงานที่ต้องตั้งชื่อซ้ำตามช่วงเวลา

ในบทนี้เราจะยังไม่อ่านไฟล์จริง แต่จะจำลองแนวคิดการวนซ้ำกับรายชื่อไฟล์

````python
monthly_files = [
    "sales_jan.xlsx",
    "sales_feb.xlsx",
    "sales_mar.xlsx"
]

for file_name in monthly_files:
    print(f"Processing file: {file_name}")
````

ในบทต่อไป แนวคิดนี้จะถูกต่อยอดเป็น

```python
for file_name in monthly_files:
    read file
    clean data
    append result
```

ซึ่งช่วยลดงาน manual จากการเปิดไฟล์ทีละไฟล์ใน Excel

````python
monthly_files = [
    "sales_jan.xlsx",
    "sales_feb.xlsx",
    "sales_mar.xlsx"
]

processed_files = []

for file_name in monthly_files:
    result_message = f"{file_name} processed"
    processed_files.append(result_message)

processed_files
````

## 10. เมื่อใดควรใช้ Loop และเมื่อใดควรใช้วิธีอื่น

Loop มีประโยชน์มากเมื่อต้องทำงานซ้ำกับรายการข้อมูล

ควรใช้ loop เมื่อ

- ต้องทำงานซ้ำกับ list ของค่า
- ต้องประมวลผลไฟล์หลายไฟล์
- ต้องสร้างผลลัพธ์ทีละรายการ
- ต้องใช้ logic ที่แตกต่างกันตามเงื่อนไข
- ต้องเตรียม workflow ที่จะนำไปใช้ต่อใน function หรือ pipeline

แต่เมื่อเข้าสู่การใช้ pandas แล้ว งานบางอย่างไม่ควรเขียน loop ทีละแถว เช่น

- คำนวณคอลัมน์ใหม่จากทั้ง column
- filter ข้อมูลจำนวนมาก
- groupby หรือ summarize ข้อมูล

เพราะ pandas มีวิธีที่เร็วและอ่านง่ายกว่า

ในหลักสูตรนี้ ให้เข้าใจก่อนว่า loop คือเครื่องมือพื้นฐานสำหรับการทำงานซ้ำ  
จากนั้นเราจะเรียนว่าในงานข้อมูลตาราง เมื่อใดควรใช้ pandas แทน loop

## 11. ข้อผิดพลาดที่พบบ่อย

### 1. ลืมใส่ colon `:`

หลังบรรทัด `for` ต้องมี colon

````python
products = ["Keyboard", "Mouse"]
````

### 2. เยื้องบรรทัดไม่ถูกต้อง

คำสั่งที่ต้องการให้ทำซ้ำต้องเยื้องเข้าไปใต้ loop

````python
products = ["Keyboard", "Mouse"]

for product in products:
    print(product)
````

ถ้าเขียนคำสั่งนอก indentation คำสั่งนั้นจะไม่อยู่ใน loop

````python
products = ["Keyboard", "Mouse"]

for product in products:
    print("Inside loop:", product)

print("Outside loop")
````

### 3. ลืมสร้าง list ว่างก่อน `.append()`

ถ้าต้องการเก็บผลลัพธ์จาก loop ต้องสร้าง list ว่างก่อน

````python
sales_list = [12000, 18000]
vat_rate = 0.07

vat_amounts = []

for sales in sales_list:
    vat_amounts.append(sales * vat_rate)

vat_amounts
````

ตัวอย่างที่อ่านง่ายกว่า

    print(sales)

### 5. แก้ไข list ระหว่างวนซ้ำโดยไม่จำเป็น
