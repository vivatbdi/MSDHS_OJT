```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
```

```python
df = pd.read_csv(url)
```

```python
df.head()
```

```python
numeric_df = df[["total_bill", "tip", "size"]]
```

```python
corr = numeric_df.corr()
```

```python
corr
```

```python
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap (matplotlib)")
plt.show()
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap (seaborn)")
plt.show()
```

```python
grouped = df.groupby("sex")["tip"].mean()
```

```python
plt.bar(grouped.index, grouped.values)
plt.ylabel("Average Tip")
plt.title("Average Tip by Sex (matplotlib)")
plt.show()
```

```python
sns.barplot(data=df, x="sex", y="tip")
plt.title("Average Tip by Sex (seaborn)")
plt.show()
```

```python
plt.hist(df["total_bill"], bins=20)
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(data=df, x="total_bill", bins=20, kde=True)
plt.title("Total Bill Distribution")
plt.show()
```

```python
plt.boxplot(df["total_bill"])
plt.ylabel("Total Bill")
plt.title("Total Bill Boxplot (matplotlib)")
plt.show()
```

```python
sns.boxplot(data=df, x="day", y="total_bill")
plt.title("Total Bill by Day")
plt.show()
```

```python
sns.violinplot(data=df, x="day", y="total_bill", inner="quartile")
plt.title("Bill Distribution by Day")
plt.show()
```

```python
for sex in df["sex"].unique():
    subset = df[df["sex"] == sex]
    plt.scatter(subset["total_bill"], subset["tip"], label=sex)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.legend()
plt.show()
```

```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
plt.title("Tip vs Bill by Sex")
plt.show()
```

```python
sns.displot(
    data=df,
    x="total_bill",
    col="time",
    row="sex",
    bins=20
)
plt.show()
```
