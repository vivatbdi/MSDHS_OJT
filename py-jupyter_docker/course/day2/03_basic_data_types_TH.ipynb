## A. Data Types เทียบกับ Excel

ใน Excel เซลล์หนึ่ง ๆ อาจเป็น:
- ตัวเลข (คำนวณได้)
- ข้อความ (ต่อข้อความได้ แต่คำนวณไม่ได้)
- วันที่/เวลา (มักซ่อนรูปแบบอยู่)
- ค่าว่าง

ใน Python ก็มีแนวคิดคล้ายกัน แต่เรามองผ่าน “ชนิดข้อมูล (data type)”  
✅ ถ้าชนิดข้อมูลไม่ตรง งานคำนวณ/สรุปข้อมูลอาจผิดพลาด

ถัดไปเราจะเริ่มจาก “ตัวเลข”

## B. ตัวเลข: int และ float

- `int` = จำนวนเต็ม เช่น 10, 0, -5
- `float` = ทศนิยม เช่น 3.14, 0.5, 76.25

✅ ลองรัน

```python
a = 10

print(a, type(a))
```

```python
b = 76.25
print(b, type(b))
```

ในงาน data:
- จำนวนคน/จำนวนแถว มักเป็น `int`
- ค่าเฉลี่ย/อัตรา/เปอร์เซ็นต์ มักเป็น `float`

## C. ข้อความ: str (string)

ข้อความ เช่น ชื่อคน จังหวัด หมวดหมู่ จะเป็น `str`

✅ ลองรัน

```python
name = "สมชาย"

print(name, type(name))
```

### ระวัง: ตัวเลขที่ถูกเก็บเป็นข้อความ

ใน Excel บางครั้งตัวเลขถูกเก็บเป็นข้อความ (มุมเขียว ๆ)  
ใน Python ก็เหมือนกัน

✅ ดูความต่างระหว่าง 10 กับ "10"

```python
x = 10

print(x, type(x))
```

```python
y = "10"

print(y, type(y))
```

ลองคำนวณ/ต่อข้อความ แล้วสังเกตผลลัพธ์

```python
print(x + 5)      # คำนวณได้
```

```python
print(y + "5")      # ต่อข้อความได้
```

แต่ถ้าพยายามเอาข้อความไปบวกกับตัวเลข จะเกิด error (เหมือน Excel ที่คำนวณไม่ได้)

✅ ลองรัน (มี error)

```python
# จะเกิด error เพราะชนิดข้อมูลไม่ตรง
y + 5
```

## D. จริง/เท็จ: bool

ในงาน data เรามักมีค่าแบบ “ใช่/ไม่ใช่” เช่น
- ผ่าน/ไม่ผ่าน
- สมาชิก/ไม่เป็นสมาชิก
- พบ/ไม่พบ

ชนิดข้อมูลนี้คือ `bool` และมี 2 ค่าเท่านั้น:
- `True`
- `False`

✅ ลองรัน

```python
passed = True
print(passed, type(passed))
```

```python
is_member = False
print(is_member, type(is_member))
```

ตัวอย่างจากงานข้อมูล: การเปรียบเทียบจะให้ผลเป็น True/False

```python
score = 45

score >= 50
```

## E. ดูชนิดข้อมูลด้วย type()

`type()` ช่วยตรวจสอบว่า “ตอนนี้ข้อมูลเป็นชนิดอะไร”  
เหมือนการเช็คว่าใน Excel เซลล์นี้เป็นตัวเลขหรือข้อความ

✅ ลองรัน

```python
print(10, type(10))
```

```python
print(3.14, type(3.14))
```

```python
print("10", type("10"))
```

```python
print(True, type(True))
```

## F. การแปลงชนิดข้อมูล (Casting) ที่จำเป็นกับงาน data

งานจริงมักอ่านข้อมูลจากไฟล์ (CSV/Excel) แล้วบางคอลัมน์อาจเป็นข้อความ  
แต่เราต้องการคำนวณ → ต้อง “แปลง” ให้เป็นตัวเลขก่อน

ตัวอย่าง: แปลง "10" (ข้อความ) ให้เป็น 10 (ตัวเลข)

✅ ลองรัน

```python
y = "10"

print(y + 5)
```

```python
y_num = int(y)

print(y_num, type(y_num))
```

```python
print(y_num + 5)
```

### แปลงเป็น float

เช่น "76.25" → 76.25

```python
avg_text = "76.25"

print(avg_num + 1)
```

```python
avg_num = int(avg_text)
```

```python
avg_num = float(avg_text)

print(avg_num, type(avg_num))
```

```python
print(avg_num + 1)
```

### แปลงเป็นข้อความ (เพื่อจัดรูปแบบผลลัพธ์)

บางครั้งเราอยากต่อข้อความ เช่น ทำรายงานสั้น ๆ

```python
age = 20

print("Age = " + age)
```

```python
print("Age = " + str(age))
```

## G. ค่าว่าง/ข้อมูลหาย (Missing) เบื้องต้น

ในงานข้อมูล เราจะเจอ “ค่าว่าง” หรือ “ข้อมูลหาย” เสมอ  
ใน Python มักแทนด้วย `None`

✅ ลองรัน

```python
missing_value = None
print(missing_value, type(missing_value))
```

## I. แบบฝึกหัด 2: ตรวจสอบชนิดข้อมูลก่อนคำนวณ

ให้สร้างตัวแปร 3 ตัวดังนี้:
- `sales` เป็นตัวเลข
- `category` เป็นข้อความ
- `is_active` เป็น True/False

แล้วใช้ `type()` แสดงชนิดข้อมูลของแต่ละตัวแปร

```python
sales = 125000
category = "A"
is_active = True

print("sales:", type(sales))
print("category:", type(category))
print("is_active:", type(is_active))
```
