## 3. การสร้าง Series

**Series** คือโครงสร้างข้อมูลหนึ่งมิติของ pandas

สามารถมองว่า Series คล้าย

- คอลัมน์หนึ่งคอลัมน์ในตาราง
- List ที่มี index กำกับ
- ชุดค่าที่สามารถคำนวณพร้อมกันได้

เราสามารถสร้าง Series จาก List ด้วย `pd.Series()`

````python
need_scores_list = [
    4.2,
    3.8,
    4.7,
    2.9,
    3.5,
]

need_scores_series = pd.Series(
    need_scores_list
)

need_scores_series
````

Series ประกอบด้วยสองส่วนหลัก

- **index** อยู่ด้านซ้าย
- **value** อยู่ด้านขวา

หากไม่ได้กำหนด index เอง pandas จะสร้าง index เริ่มจาก `0` ให้อัตโนมัติ

````python
type(need_scores_series)
````

ผลลัพธ์คือ

```text
pandas.core.series.Series
```

แสดงว่าตัวแปรนี้เป็น pandas Series ไม่ใช่ List

### การกำหนดชื่อให้ Series

Series สามารถมีชื่อกำกับได้

````python
need_scores_series = pd.Series(
    need_scores_list,
    name="need_score",
)

need_scores_series
````

ชื่อ `"need_score"` ช่วยอธิบายความหมายของข้อมูล และจะมีประโยชน์เมื่อนำ Series ไปใช้ร่วมกับ DataFrame

### การเข้าถึงข้อมูลใน Series

````python
need_scores_series[0]
````

````python
need_scores_series[0:3]
````

Series สามารถเข้าถึงข้อมูลด้วย index และ slicing คล้าย List

อย่างไรก็ตาม ใน pandas ควรใช้ `.loc` หรือ `.iloc` เมื่อการเลือกข้อมูลมีความซับซ้อน เพื่อให้ระบุได้ชัดเจนว่ากำลังเลือกด้วย label หรือตำแหน่ง

## 4. ความแตกต่างระหว่าง List และ Series

List และ Series เป็นข้อมูลหนึ่งมิติเหมือนกัน แต่ Series มีความสามารถที่เหมาะกับงานข้อมูลมากกว่า

| ประเด็น | List | Series |
|---|---|---|
| มีลำดับข้อมูล | มี | มี |
| มี index | ใช้ตำแหน่ง | มี index กำกับ |
| คำนวณทั้งชุดโดยตรง | ไม่ได้ตามปกติ | ได้ |
| เปรียบเทียบทั้งชุด | ต้องใช้ Loop | ได้ |
| มีชื่อกำกับ | ไม่มีโดยตรง | มีได้ |
| ใช้เป็นคอลัมน์ DataFrame | ต้องแปลงก่อน | ใช้ได้โดยตรง |

### การคำนวณกับ Series ทั้งชุด

หากต้องการเพิ่มคะแนนทุกค่าอีก `0.5` สามารถเขียนกับ Series ได้โดยตรง

````python
need_scores_series + 0.5
````

หากใช้ List ต้องใช้ Loop หรือวิธีอื่นเพื่อคำนวณทีละค่า

````python
adjusted_scores = []

for score in need_scores_list:
    adjusted_scores.append(score + 0.5)

adjusted_scores
````

### การเปรียบเทียบค่าทั้ง Series

````python
need_scores_series >= 4.0
````

ผลลัพธ์คือ Series ของค่า Boolean

- `True` หมายถึงคะแนนถึงเกณฑ์
- `False` หมายถึงคะแนนไม่ถึงเกณฑ์

Boolean Series นี้สามารถนำไปกรองข้อมูลใน DataFrame ได้

## 5. การสร้าง DataFrame

**DataFrame** คือโครงสร้างข้อมูลสองมิติของ pandas

สามารถมองว่า DataFrame คล้ายกับ

- ตารางใน Excel
- Worksheet ที่มีหัวคอลัมน์
- ตารางฐานข้อมูล
- ชุดของ Series หลายชุดที่มี index ร่วมกัน

วิธีหนึ่งที่ใช้สร้าง DataFrame คือสร้างจาก Dictionary ที่แต่ละ value เป็น List

````python
survey_data = {
    "respondent_id": [1, 2, 3, 4, 5],
    "area": [
        "Bangkok",
        "Chiang Mai",
        "Bangkok",
        "Khon Kaen",
        "Songkhla",
    ],
    "age": [34, 52, 41, 29, 63],
    "need_score": [
        4.2,
        3.8,
        4.7,
        2.9,
        3.5,
    ],
    "status": [
        "Completed",
        "Completed",
        "Verified",
        "Draft",
        "Completed",
    ],
}
````

````python
df = pd.DataFrame(survey_data)

df
````

เมื่อสร้าง DataFrame จาก Dictionary

- key กลายเป็นชื่อคอลัมน์
- List ของแต่ละ key กลายเป็นข้อมูลในคอลัมน์
- สมาชิกตำแหน่งเดียวกันกลายเป็นข้อมูลในแถวเดียวกัน
- pandas สร้าง index ของแถวให้อัตโนมัติ

ข้อสำคัญคือ List ทุกชุดควรมีจำนวนสมาชิกเท่ากัน

## 6. องค์ประกอบของ DataFrame

DataFrame ประกอบด้วย

- **row** — แถวของข้อมูล
- **column** — ประเภทหรือตัวแปรของข้อมูล
- **index** — label ที่ใช้ระบุแถว
- **column name** — ชื่อที่ใช้ระบุคอลัมน์
- **value** — ค่าในแต่ละตำแหน่ง

ใน DataFrame ตัวอย่าง

- หนึ่งแถวแทนผู้ตอบหนึ่งคน
- หนึ่งคอลัมน์แทนข้อมูลหนึ่งประเภท

### เปรียบเทียบ DataFrame กับ Excel

| แนวคิด | Excel | pandas |
|---|---|---|
| ตารางข้อมูล | Worksheet หรือ Table | DataFrame |
| คอลัมน์หนึ่งคอลัมน์ | Column | Series |
| แถว | Row | Row |
| หัวตาราง | Header | Column name |
| เลขแถว | Row number | Index |
| ช่องข้อมูล | Cell | Value |
| เลือกคอลัมน์ | คลิกหัวคอลัมน์ | `df["column"]` |
| เลือกแถว | คลิกเลขแถว | `.loc` หรือ `.iloc` |
| กรองข้อมูล | Filter | Boolean selection |
| สร้างคอลัมน์ | ใส่สูตร | กำหนด Series ให้คอลัมน์ใหม่ |

DataFrame ไม่ได้แทนที่ Excel ในทุกงาน

Excel ยังสะดวกสำหรับ

- เปิดดูข้อมูลอย่างรวดเร็ว
- แก้ไขค่าด้วยมือ
- จัดรูปแบบรายงาน
- ตรวจสอบตารางด้วยสายตา

DataFrame เหมาะเมื่อ

- ต้องทำขั้นตอนซ้ำ
- ต้องประมวลผลข้อมูลจำนวนมาก
- ต้องกรองหรือคำนวณหลายเงื่อนไข
- ต้องการบันทึกขั้นตอนเป็นโค้ด

## 9. การเลือกข้อมูลด้วย `.loc`

`.loc` ใช้เลือกข้อมูลด้วย **label**

มักใช้กับ

- ชื่อ index
- ชื่อคอลัมน์
- Boolean Series

รูปแบบทั่วไปคือ

```python
df.loc[row_selection, column_selection]
```

### เลือกแถวตาม index

````python
df.loc[0]
````

### เลือกค่าจากแถวและชื่อคอลัมน์

````python
df.loc[0, "area"]
````

### เลือกหลายแถวและหลายคอลัมน์

````python
df.loc[
    0:2,
    ["respondent_id", "area", "need_score"],
]
````

ข้อควรระวังคือ slicing ของ `.loc` จะรวม label ปลายทางด้วย

ดังนั้น

```python
df.loc[0:2]
```

จะได้ index `0`, `1` และ `2`

### เลือกทุกแถวและบางคอลัมน์

````python
df.loc[
    :,
    ["area", "need_score"],
]
````

เครื่องหมาย `:` หมายถึงเลือกทุกแถว

## 10. การเลือกข้อมูลด้วย `.iloc`

`.iloc` ใช้เลือกข้อมูลด้วย **ตำแหน่งจำนวนเต็ม** หรือ integer location

รูปแบบทั่วไปคือ

```python
df.iloc[row_position, column_position]
```

ตำแหน่งเริ่มจาก `0` เช่นเดียวกับ index ของ Python

### เลือกแถวแรก

## 14. การใช้ Function กับ DataFrame

เราสามารถสร้าง Function เพื่อกำหนดกฎ แล้วใช้ `.apply()` กับ Series ได้

ตัวอย่างเกณฑ์

| คะแนน | ระดับ |
|---:|---|
| ตั้งแต่ 4.0 ขึ้นไป | High |
| ตั้งแต่ 3.0 แต่น้อยกว่า 4.0 | Medium |
| น้อยกว่า 3.0 | Low |

````python
def classify_need(score):
    if score >= 4.0:
        return "High"
    elif score >= 3.0:
        return "Medium"
    else:
        return "Low"
````

ควรทดสอบ Function กับค่าเดียวก่อน

````python
classify_need(4.2)
````

จากนั้นใช้ `.apply()` กับคอลัมน์ `need_score`

````python
df["need_level"] = (
    df["need_score"]
    .apply(classify_need)
)

df
````

`.apply()` นำ Function ไปใช้กับค่าทีละค่าใน Series

เหมาะเมื่อ

- มีกฎที่อธิบายด้วย Function ได้ชัดเจน
- Logic มีหลายระดับ
- Function จะถูกนำกลับมาใช้ซ้ำ

สำหรับการคำนวณง่าย ๆ ควรใช้การคำนวณกับทั้งคอลัมน์โดยตรงก่อน เพราะมักสั้นและทำงานได้เร็วกว่า

สร้างคอลัมน์ใหม่ 3 คอลัมน์ใน `survey_df`

### 1. `debt_to_income_ratio`

```python
debt_amount / monthly_income
```

### 2. `is_low_income`

กำหนดให้เป็น `True` เมื่อ `monthly_income` น้อยกว่า `15000`

### 3. `is_high_need`

หลังจากสร้างแล้ว ให้แสดงเฉพาะคอลัมน์

## แบบฝึกหัดที่ 9: ใช้ Function และ `.apply()`

สร้าง Function ชื่อ `classify_need_level()`

กำหนดเกณฑ์ดังนี้

| คะแนน | ระดับ |
|---:|---|
| ตั้งแต่ 4.5 ขึ้นไป | Very High |
| ตั้งแต่ 4.0 แต่น้อยกว่า 4.5 | High |
| ตั้งแต่ 3.0 แต่น้อยกว่า 4.0 | Medium |
| น้อยกว่า 3.0 | Low |

จากนั้นใช้ `.apply()` สร้างคอลัมน์ใหม่ชื่อ `need_level`

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 9
>
> เรียงเงื่อนไขจากคะแนนสูงไปต่ำ แล้วใช้ `.apply()` ส่งคะแนนแต่ละค่าเข้า Function

````python
def classify_need_level(score):
    if score >= 4.5:
        return "Very High"
    elif score >= 4.0:
        return "High"
    elif score >= 3.0:
        return "Medium"
    else:
        return "Low"
````

````python
survey_df["need_level"] = (
    survey_df["need_score"]
    .apply(classify_need_level)
)

survey_df.loc[
    :,
    [
        "respondent_id",
        "province",
        "need_score",
        "need_level",
    ],
]
````

## แบบฝึกหัดที่ 10: สร้างข้อมูลสำหรับติดตามงาน

เลือกเฉพาะผู้ตอบที่เข้าเงื่อนไขอย่างน้อยหนึ่งข้อ

- `need_level` เป็น `"Very High"`
- หรือ `is_low_income` เป็น `True`

เก็บผลลัพธ์ใน `follow_up_df`

จากนั้น

1. เลือกเฉพาะคอลัมน์  
   `respondent_id`, `province`, `monthly_income`, `need_score`, `need_level`
2. ใช้ Loop ผ่าน `follow_up_df` เพื่อแสดงข้อความในรูปแบบ

```text
Respondent 3 จาก Khon Kaen มี need level ระดับ Very High
```

แบบฝึกหัดนี้ใช้ Loop เพื่อสร้างข้อความสำหรับรายงาน ไม่ใช่เพื่อคำนวณคอลัมน์ใหม่

> ### เฉลยแบบฝึกหัดที่ 10
>
> ใช้ `|` เชื่อมเงื่อนไขที่ต้องการอย่างน้อยหนึ่งข้อ จากนั้นใช้ `.iterrows()` เพื่อสร้างข้อความจากแต่ละแถว
>
> ในกรณีนี้ Loop ใช้เพื่อแสดงข้อความ ไม่ได้ใช้คำนวณคอลัมน์

````python
follow_up_df = survey_df.loc[
    (
        survey_df["need_level"]
        == "Very High"
    )
    | (
        survey_df["is_low_income"]
    ),
    [
        "respondent_id",
        "province",
        "monthly_income",
        "need_score",
        "need_level",
    ],
]

follow_up_df
````

````python
for index, row in follow_up_df.iterrows():
    print(
        f"Respondent {row['respondent_id']} "
        f"จาก {row['province']} มี need level "
        f"ระดับ {row['need_level']}"
    )
````
