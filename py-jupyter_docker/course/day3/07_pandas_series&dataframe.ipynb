## 1. ทบทวน list

ตัวอย่างเช่น รายชื่อพื้นที่ที่เราต้องการวิเคราะห์

## 3. จาก list สู่ pandas Series 

ก่อนใช้ `Series` และ `DataFrame` เราต้อง import library `pandas` 

โดยทั่วไปนิยม import pandas ด้วยชื่อย่อว่า `pd`

````python
import pandas as pd
````

`Series` คือ data structure หนึ่งมิติของ pandas 

ให้มองว่า `Series` คล้ายกับ `list` ที่มี index และมีความสามารถเพิ่มเติมสำหรับงานข้อมูล

````python
need_scores_series = pd.Series([4.2, 3.8, 4.7, 2.9, 3.5]) 
need_scores_series
````

จากผลลัพธ์จะเห็นว่า `Series` มี 2 ส่วนหลัก 
- index อยู่ด้านซ้าย 
- value อยู่ด้านขวา 

ถ้าเราไม่ได้กำหนด index เอง pandas จะสร้าง index เริ่มจาก 0 ให้อัตโนมัติ

## 4. Series ต่างจาก list อย่างไร 
`list` และ `Series` ดูคล้ายกัน เพราะเป็นข้อมูลหนึ่งมิติ 

แต่ `Series` มีความสามารถที่เหมาะกับงานข้อมูลมากกว่า เช่น สามารถคำนวณกับค่าทั้งชุดได้ทันที

````python
need_scores_list = [4.2, 3.8, 4.7, 2.9, 3.5] 
need_scores_list
````

````python
need_scores_series = pd.Series(need_scores_list) 
need_scores_series
````

ตัวอย่างเช่น ถ้าต้องการบวกคะแนนทุกค่าเพิ่มอีก 0.5 

ถ้าเป็น `Series` สามารถเขียนได้โดยตรง

````python
need_scores_series + 0.5
````

แต่ถ้าเป็น `list` ธรรมดา เรามักต้องใช้ loop

````python
new_scores = [] 
for score in need_scores_list: 
    new_scores.append(score + 0.5) 

new_scores
````

ดังนั้น `Series` จึงเหมาะกว่าเมื่อเราต้องการทำงานกับข้อมูลทั้งชุด เช่น 
- คำนวณกับทุกค่า 
- เปรียบเทียบทุกค่า 
- สร้างเงื่อนไขกับทุกค่า 
- ใช้ต่อเป็น column ใน DataFrame    

## 5. การใช้ condition กับ Series 
เราได้เรียนการใช้ condition กับค่าหนึ่งค่า เช่น

````python
score = 4.2

if score >= 4.0: 
    print("High") 
else: 
    print("Normal")
````

แต่เมื่อใช้ `Series` เราสามารถเปรียบเทียบค่าทั้งชุดได้ในคำสั่งเดียว

````python
need_scores_series >= 4.0
````

ผลลัพธ์ที่ได้คือ `Series` ของค่า `True` และ `False` 

ค่านี้บอกว่าแต่ละแถวเข้าเงื่อนไขหรือไม่ 

แนวคิดนี้จะสำคัญมากเมื่อนำไปใช้กรองข้อมูลใน DataFrame

## 6. ทบทวน dictionary 
`dictionary` ใช้เก็บข้อมูลแบบ key-value 

ตัวอย่างเช่น ข้อมูลของผู้ตอบแบบสำรวจหนึ่งคน

````python
respondent = { 
    "respondent_id": 1, 
    "area": "Bangkok", 
    "age": 34, 
    "need_score": 4.2, 
    "status": "Completed" } 
respondent
````

เราสามารถเข้าถึงข้อมูลใน dictionary ด้วย key

````python
respondent["area"]
````

ถ้ามีข้อมูลหลายคน เราอาจใช้ dictionary ที่ value เป็น list 

ลักษณะนี้เริ่มคล้ายกับตารางข้อมูล

````python
survey_data = { 
    "respondent_id": [1, 2, 3, 4, 5], 
    "area": ["Bangkok", "Chiang Mai", "Bangkok", "Khon Kaen", "Songkhla"], 
    "age": [34, 52, 41, 29, 63], 
    "need_score": [4.2, 3.8, 4.7, 2.9, 3.5], 
    "status": ["Completed", "Completed", "Verified", "Draft", "Completed"] } 
survey_data
````

dictionary แบบนี้มีโครงสร้างที่ดีขึ้นกว่า list หลายตัว เพราะมีชื่อ field หรือชื่อ column ชัดเจน 

แต่ถ้าต้องการเลือกแถว กรองข้อมูล หรือคำนวณกับ column หลาย ๆ ตัว dictionary ยังไม่สะดวกเท่า DataFrame

## 7. จาก dictionary สู่ pandas DataFrame 
เราสามารถแปลง dictionary ให้เป็น DataFrame ได้ด้วย `pd.DataFrame()`

````python
df = pd.DataFrame(survey_data) 
df
````

ในตัวอย่างนี้ 
- key ของ dictionary กลายเป็นชื่อ column 
- list ที่อยู่ในแต่ละ key กลายเป็นข้อมูลใน column 
- แต่ละตำแหน่งของ list กลายเป็นข้อมูลแต่ละ row 

ตอนนี้ข้อมูลของเราอยู่ในรูปแบบตารางแล้ว

## 8. DataFrame คืออะไร 
`DataFrame` คือ data structure แบบ 2 มิติของ pandas 

ให้มองว่า DataFrame คล้ายกับตารางใน Excel 

DataFrame ประกอบด้วย 
- row หรือแถว

- index หรือป้ายกำกับของแถว 
- value หรือค่าข้อมูลในแต่ละช่อง

## 9. เปรียบเทียบ DataFrame กับ Excel 
| แนวคิด | Excel | pandas DataFrame | 
|---|---|---| 
| ตารางข้อมูล | Worksheet หรือ Table | DataFrame | 
| แถว | Row | Row | 
| คอลัมน์ | Column | Column | 
| หัวตาราง | Header | Column name | 
| เลขแถว | Row number | Index | 
| ช่องข้อมูล | Cell | Value | 
| เลือกคอลัมน์ | คลิกหัวคอลัมน์ | `df["column_name"]` | 
| เลือกแถว | คลิกเลขแถว | `.loc` หรือ `.iloc` | 
| กรองข้อมูล | Filter | Boolean selection | 

DataFrame ไม่ได้มาแทน Excel ทั้งหมด แต่ช่วยให้งานข้อมูลที่ต้องทำซ้ำ ตรวจสอบย้อนหลัง หรือจัดการข้อมูลจำนวนมาก ทำได้เป็นระบบมากขึ้น

## 14. การเลือกข้อมูลด้วย iloc `.iloc` 

ใช้เลือกข้อมูลจากตำแหน่งของ row และ column 

คำว่า `iloc` ให้จำว่าเกี่ยวกับ integer location หรือตำแหน่งที่เป็นตัวเลข เหมาะเมื่อเราต้องการเลือกข้อมูลตามลำดับตำแหน่ง

````python
df
````

เลือก row แรกของ DataFrame

````python
df.iloc[0]
````

เลือก row ที่ 1 ถึง row ที่ 3 ตามตำแหน่ง 

หมายเหตุ: อย่าลืมว่าการ slice ใน Python จะไม่รวมตำแหน่งท้ายสุด

````python
df.iloc[0:3]
````

เลือกค่าที่อยู่ row แรก และ column ที่สอง 

ในที่นี้ row แรกคือตำแหน่ง 0 และ column ที่สองคือตำแหน่ง 1

````python
df.iloc[0, 1]
````

เลือกหลาย row และหลาย column ด้วย `.iloc` 

ตัวอย่างนี้เลือก 3 row แรก และ 2 column แรก

````python
df.iloc[0:3, 0:2]
````

## 15. การเลือกข้อมูลด้วย loc 
`.loc` ใช้เลือกข้อมูลจาก label หรือชื่อ 

โดยทั่วไปมักใช้กับ 
- ชื่อ index 
- ชื่อ column 

ใน DataFrame ที่ยังไม่ได้ตั้ง index เอง index จะเป็นตัวเลข 0, 1, 2, ...

````python
df
````

เลือก row ที่มี index เป็น 0

````python
df.loc[0]
````

````python
df.loc[0, "area"]
````

เลือกทุก row แต่เลือกเฉพาะ column `area` และ `need_score`

````python
df.loc[:, ["area", "need_score"]]
````

## 16. ความแตกต่างระหว่าง loc และ iloc 
| คำสั่ง | ใช้เลือกจาก | ตัวอย่าง | 
|---|---|---| 
| `.iloc` | ตำแหน่งของ row/column | row ที่ 0, column ที่ 1 | 
| `.loc` | label หรือชื่อ | index 0, column `"area"` | 

จำง่าย ๆ คือ 
- `.iloc` ใช้ตำแหน่งที่เป็นตัวเลข (indicator location) 
- `.loc` ใช้ชื่อหรือ label

## 17. การเลือกข้อมูลแบบมีเงื่อนไข

## 21. การประยุกต์ใช้ if และ function กับ DataFrame 
จากการเรียนเรื่อง function เราสามารถสร้าง function เพื่อจัดกลุ่มคะแนนได้ 

ตัวอย่างเช่น 
- คะแนนตั้งแต่ 4.0 ขึ้นไป = High 
- คะแนนตั้งแต่ 3.0 ถึงน้อยกว่า 4.0 = Medium 
- คะแนนน้อยกว่า 3.0 = Low

````python
def classify_need(score): 
    if score >= 4.0: 
        return "High" 
    elif score >= 3.0: 
        return "Medium" 
    else: 
        return "Low"
````

ทดลองใช้ function กับค่าหนึ่งค่าก่อน

````python
classify_need(4.2)
````

เมื่อต้องการนำ function ไปใช้กับทั้ง column สามารถใช้ `.apply()` 

ในที่นี้เราจะนำ function `classify_need()` ไปใช้กับ column `need_score`

````python
df["need_level"] = df["need_score"].apply(classify_need) 
df
````

จุดสำคัญคือ เรานำความรู้เรื่อง `if` และ function จากวันที่ 1 มาใช้กับข้อมูลใน DataFrame 

สิ่งนี้ทำให้เราสามารถสร้างกฎหรือ logic สำหรับจัดกลุ่มข้อมูลได้อย่างเป็นระบบ

## 22. การประยุกต์ใช้ loop กับ DataFrame 
loop ยังสามารถใช้ร่วมกับ DataFrame ได้ 

ตัวอย่างแรก คือ loop ผ่านรายชื่อ column ที่ต้องการแสดง

````python
columns_to_show = ["area", "age", "need_score"] 

for column in columns_to_show: 
    print("Column:", column) 
    print(df[column]) 
    print()
````

ตัวอย่างต่อไป คือ loop ผ่านแต่ละ row ด้วย `.iterrows()`

วิธีนี้ช่วยให้เห็นว่าแต่ละ row ใน DataFrame สามารถถูกอ่านออกมาทีละแถวได้

````python
for index, row in df.iterrows(): 
    print("Respondent:", row["respondent_id"]) 
    print("Area:", row["area"]) 
    print("Need score:", row["need_score"]) 
    print("Need level:", row["need_level"]) 
    print("---")
````

อย่างไรก็ตาม ใน pandas เราไม่จำเป็นต้องใช้ loop เสมอไป 

หลายกรณี pandas สามารถทำงานกับทั้ง column ได้โดยตรง ซึ่งมักจะสั้นกว่า อ่านง่ายกว่า และเหมาะกับงานข้อมูลมากกว่า 

ตัวอย่างเช่น การสร้าง column `is_high_need` ก่อนหน้านี้ ไม่จำเป็นต้องเขียน loop เอง
