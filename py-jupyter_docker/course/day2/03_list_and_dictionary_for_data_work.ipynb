## 2. การสร้าง List

**List** คือโครงสร้างข้อมูลที่ใช้เก็บค่าหลายค่าเรียงตามลำดับ

ลักษณะสำคัญของ List คือ

- ใช้วงเล็บเหลี่ยม `[ ]`
- สมาชิกแต่ละตัวคั่นด้วย comma `,`
- ข้อมูลแต่ละค่ามีตำแหน่ง
- สามารถเก็บข้อมูลซ้ำได้
- สามารถเพิ่ม แก้ไข และลบข้อมูลได้

รูปแบบพื้นฐานคือ

```python

```

### List ของข้อความ

````python
products = [
    "Keyboard",
    "Mouse",
    "Monitor",
    "Printer",
]

products
````

### List ของตัวเลข

````python
monthly_sales = [
    12500,
    18300,
    15700,
    22100,
    19800,
    17600,
]

monthly_sales
````

### List ของ Boolean

````python
is_data_complete = [
    True,
    False,
    True,
    True,
]

is_data_complete
````

List สามารถเก็บข้อมูลต่างชนิดกันไว้ด้วยกันได้ เช่น

```python

```

อย่างไรก็ตาม ในงานข้อมูลควรพยายามให้ List หนึ่งชุดเก็บข้อมูลที่มีความหมายหรือชนิดเดียวกัน เช่น

- List ของชื่อพื้นที่
- List ของจำนวนผู้ตอบ
- List ของคะแนน
- List ของชื่อไฟล์

แนวทางนี้ช่วยให้โค้ดอ่านและนำไปใช้ต่อได้ง่ายกว่า

## 3. การเข้าถึงข้อมูลใน List

สมาชิกแต่ละค่าใน List มีตำแหน่งที่เรียกว่า **index**

Python เริ่มนับ index จาก `0`

กำหนด List ต่อไปนี้

```python
products = ["Keyboard", "Mouse", "Monitor", "Printer"]
```

ตำแหน่งของสมาชิกคือ

| ค่า | Index |
|---|---:|
| Keyboard | 0 |
| Mouse | 1 |
| Monitor | 2 |
| Printer | 3 |

````python
products = [
    "Keyboard",
    "Mouse",
    "Monitor",
    "Printer",
]
````

### เข้าถึงสมาชิกตัวแรก

````python
products[0]
````

### เข้าถึงสมาชิกตัวที่สอง

````python
products[1]
````

### เข้าถึงสมาชิกตัวที่สี่

````python
products[3]
````

ข้อควรจำคือ

> สมาชิกตัวแรกของ List อยู่ที่ index `0` ไม่ใช่ index `1`

ดังนั้น หากต้องการสมาชิกตัวที่ `n` โดยทั่วไปจะใช้ index `n - 1`

### การเข้าถึงข้อมูลจากท้าย List

Python สามารถใช้ index ติดลบเพื่อเริ่มนับจากท้าย List ได้

| Index | ความหมาย |
|---:|---|
| `-1` | สมาชิกตัวสุดท้าย |
| `-2` | สมาชิกตัวรองสุดท้าย |
| `-3` | สมาชิกตัวที่สามจากท้าย |

````python
products[-1]
````

````python
products[-2]
````

Index ติดลบมีประโยชน์เมื่อเราต้องการค่าท้าย List แต่ไม่ทราบจำนวนสมาชิกทั้งหมด

## 4. การเลือกข้อมูลบางช่วงด้วย slicing

การเลือกสมาชิกหลายค่าที่อยู่ต่อเนื่องกันเรียกว่า **slicing**

รูปแบบพื้นฐานคือ

```python
list_name[start:end]
```

Python จะเลือกข้อมูลตั้งแต่ index `start` จนถึงก่อน index `end`

กล่าวอีกแบบคือ

> รวมตำแหน่งเริ่มต้น แต่ไม่รวมตำแหน่งสิ้นสุด

````python
products = [
    "Keyboard",
    "Mouse",
    "Monitor",
    "Printer",
    "Webcam",
]

products[0:3]
````

ผลลัพธ์ประกอบด้วยสมาชิกที่ index `0`, `1` และ `2`

ไม่รวมสมาชิกที่ index `3`

### ละตำแหน่งเริ่มต้น

หากไม่ระบุ `start` Python จะเริ่มจากสมาชิกตัวแรก

````python
products[:3]
````

### ละตำแหน่งสิ้นสุด

หากไม่ระบุ `end` Python จะเลือกตั้งแต่ตำแหน่งเริ่มต้นไปจนถึงสมาชิกตัวสุดท้าย

````python
products[2:]
````

### เลือกสมาชิกทั้งหมด

การใช้ `[:]` จะเลือกสมาชิกทั้งหมดใน List

````python
products[:]
````

## 5. การเพิ่ม แก้ไข และลบข้อมูลใน List

List เป็นโครงสร้างข้อมูลที่สามารถเปลี่ยนแปลงได้ หรือเรียกว่า **mutable**

เราสามารถ

- เพิ่มสมาชิก
- แก้ไขสมาชิกเดิม
- ลบสมาชิก
- ตรวจสอบจำนวนสมาชิก

### 5.1 เพิ่มข้อมูลด้วย `.append()`

`.append()` ใช้เพิ่มสมาชิกใหม่ต่อท้าย List

````python
products = [
    "Keyboard",
    "Mouse",
    "Monitor",
]

products.append("Printer")

products
````

หลังใช้ `.append()` ค่า `"Printer"` จะอยู่ท้าย List

### 5.2 แก้ไขข้อมูลด้วย index

เราสามารถกำหนดค่าใหม่ให้ตำแหน่งที่ต้องการได้

````python
products[1] = "Wireless Mouse"

products
````

สมาชิกที่ index `1` ถูกเปลี่ยนจาก `"Mouse"` เป็น `"Wireless Mouse"`

### 5.3 ลบข้อมูลด้วย `.remove()`

`.remove()` ใช้ลบสมาชิกตามค่าที่ระบุ

````python
products.remove("Monitor")

products
````

หากค่าที่ระบุปรากฏหลายครั้ง `.remove()` จะลบเฉพาะค่าที่พบเป็นรายการแรก

````python
areas = [
    "Bangkok",
    "Chiang Mai",
    "Bangkok",
]

areas.remove("Bangkok")

areas
````

### 5.4 ตรวจสอบจำนวนสมาชิกด้วย `len()`

`len()` ใช้นับจำนวนสมาชิกใน List

````python
len(products)
````

## 7. การสร้าง Dictionary

**Dictionary** คือโครงสร้างข้อมูลที่เก็บข้อมูลเป็นคู่ของ

- **key** — ชื่อหรือรหัสที่ใช้อ้างอิง
- **value** — ค่าที่สัมพันธ์กับ key

รูปแบบพื้นฐานคือ

```python
dictionary_name = {
    key_1: value_1,
    key_2: value_2,
}
```

Dictionary ใช้วงเล็บปีกกา `{ }` และใช้ colon `:` คั่นระหว่าง key กับ value

````python
respondent_info = {
    "respondent_id": "R001",
    "area_name": "Bangkok",
    "age": 68,
    "has_disability": True,
}

respondent_info
````

Dictionary ด้านบนประกอบด้วยข้อมูลดังนี้

| Key | Value |
|---|---|
| `respondent_id` | `"R001"` |
| `area_name` | `"Bangkok"` |
| `age` | `68` |
| `has_disability` | `True` |

การมี key กำกับแต่ละค่าทำให้เราทราบความหมายของข้อมูลโดยไม่ต้องจดจำตำแหน่ง

### Key ต้องไม่ซ้ำกัน

ภายใน Dictionary เดียวกัน แต่ละ key ควรมีเพียงหนึ่งค่า

หากกำหนด key เดิมซ้ำ ค่าที่กำหนดภายหลังจะแทนที่ค่าก่อนหน้า

````python
area_info = {
    "area_code": "A001",
    "area_name": "Bangkok",
    "area_name": "Chiang Mai",
}

area_info
````

ผลลัพธ์ของ `area_name` จะเป็น `"Chiang Mai"` เพราะค่าหลังสุดแทนที่ค่าก่อนหน้า

ดังนั้น ควรตรวจสอบให้ key ใน Dictionary ไม่ซ้ำกันโดยไม่ได้ตั้งใจ

## 8. การเข้าถึงข้อมูลใน Dictionary

List ใช้ **index** เพื่อเข้าถึงข้อมูลตามตำแหน่ง

Dictionary ใช้ **key** เพื่อเข้าถึง value ที่สัมพันธ์กัน

````python
respondent_info = {
    "respondent_id": "R001",
    "area_name": "Bangkok",
    "age": 68,
    "has_disability": True,
}
````

### เข้าถึงค่าด้วย `[key]`

````python
respondent_info["area_name"]
````

````python
respondent_info["age"]
````

การใช้ key ทำให้โค้ดอ่านได้โดยตรงว่าต้องการข้อมูลใด เช่น

```python
respondent_info["age"]
```

หมายถึง ต้องการค่าอายุจากข้อมูลผู้ตอบ

### เข้าถึงค่าด้วย `.get()`

อีกวิธีหนึ่งคือใช้ `.get()`

````python
respondent_info.get("has_disability")
````

ความแตกต่างสำคัญคือ

- `dictionary[key]` จะเกิด `KeyError` หากไม่มี key
- `dictionary.get(key)` จะคืนค่า `None` หากไม่มี key

````python
respondent_info.get("income")
````

### กำหนดค่าเริ่มต้นให้ `.get()`

เราสามารถกำหนดค่าที่ต้องการให้คืนกลับเมื่อไม่พบ key

````python
respondent_info.get("income", "Unknown")
````

การใช้ `.get()` เหมาะเมื่อข้อมูลอาจไม่มี key ที่ต้องการ และเราไม่ต้องการให้โปรแกรมหยุดทำงานทันที

## 9. การเพิ่ม แก้ไข และลบข้อมูลใน Dictionary

Dictionary เป็นโครงสร้างข้อมูลที่แก้ไขได้เช่นเดียวกับ List

เราสามารถ

- เพิ่ม key-value ใหม่
- เปลี่ยน value ของ key เดิม
- ลบ key-value
- ตรวจสอบ key และ value ที่มีอยู่

### 9.1 เพิ่ม key-value ใหม่

````python
respondent_info["main_support_need"] = "Housing"

respondent_info
````

หาก key ยังไม่มีอยู่ Python จะเพิ่ม key-value ใหม่เข้าไปใน Dictionary

### 9.2 แก้ไข value ของ key เดิม

````python
respondent_info["area_name"] = "Chiang Mai"

respondent_info
````

หาก key มีอยู่แล้ว การกำหนดค่าใหม่จะเปลี่ยน value ของ key นั้น

### 9.3 ลบข้อมูลด้วย `del`

````python
del respondent_info["main_support_need"]

respondent_info
````

ควรใช้ `del` เมื่อมั่นใจว่า key มีอยู่ เพราะหากไม่มี key จะเกิด `KeyError`

### 9.4 ตรวจสอบ key ด้วย `.keys()`

## 11. การเลือกใช้ List และ Dictionary

List และ Dictionary ใช้เก็บหลายค่าเหมือนกัน แต่มีวิธีจัดโครงสร้างและอ้างอิงข้อมูลต่างกัน

| ประเด็น | List | Dictionary |
|---|---|---|
| รูปแบบ | `[value_1, value_2]` | `{key_1: value_1}` |
| ใช้อ้างอิงด้วย | index | key |
| ลำดับมีความสำคัญ | โดยทั่วไปใช่ | เน้นการค้นหาด้วย key |
| key ต้องไม่ซ้ำ | ไม่มี key | ใช่ |
| ค่าซ้ำได้ | ได้ | value ซ้ำได้ |
| เหมาะกับ | รายการข้อมูล | ข้อมูลที่มีชื่อหรือรหัสกำกับ |
| ตัวอย่าง | รายชื่อไฟล์ | รหัสพื้นที่ → ชื่อพื้นที่ |

### ใช้ List เมื่อ

- ต้องการเก็บข้อมูลหลายรายการ
- ต้องการรักษาลำดับ
- ต้องการอ้างอิงด้วยตำแหน่ง
- ต้องการนำสมาชิกไปประมวลผลทีละรายการ
- ข้อมูลแต่ละค่าอยู่ในกลุ่มความหมายเดียวกัน

ตัวอย่างเช่น

- รายชื่อไฟล์
- รายชื่อพื้นที่
- รายชื่อคอลัมน์
- รายชื่อ Sheet
- คะแนนหลายรายการ

### ใช้ Dictionary เมื่อ

- ต้องการเก็บข้อมูลเป็นคู่ key-value
- ต้องการค้นหาค่าด้วยชื่อหรือรหัส
- ต้องการทำ mapping
- ต้องการรวมค่ากำหนดที่มีชื่อแตกต่างกัน
- ต้องการเก็บหลายคุณลักษณะของรายการหนึ่งรายการ

ตัวอย่างเช่น

- รหัสพื้นที่ → ชื่อพื้นที่
- รหัสความต้องการ → ชื่อความต้องการ
- ชื่อคอลัมน์เดิม → ชื่อคอลัมน์ใหม่
- configuration ของรายงาน
- ข้อมูลผู้ตอบหนึ่งราย

### คำถามช่วยเลือกโครงสร้างข้อมูล

ลองถามตนเองว่า

> ข้อมูลนี้เป็น “รายการหลายค่า” หรือ “ความสัมพันธ์ระหว่างชื่อกับค่า”

- หากเป็นรายการหลายค่า ให้เริ่มพิจารณา **List**
- หากเป็นคู่ชื่อหรือรหัสกับค่า ให้เริ่มพิจารณา **Dictionary**

4. คำนวณจำนวนผู้ตอบรวมของทั้ง 3 พื้นที่

   `จำนวนผู้ตอบแบบสำรวจรวมเท่ากับ ... คน`

````python
area_names = [
    "Bangkok",
    "Chiang Mai",
    "Khon Kaen",
]

respondent_counts = {
    "Bangkok": 120,
    "Chiang Mai": 85,
    "Khon Kaen": 96,
}
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 9
>
> List ใช้เก็บลำดับชื่อพื้นที่ ส่วน Dictionary ใช้ค้นหาจำนวนผู้ตอบจากชื่อพื้นที่
>
> การดึงข้อมูลของพื้นที่สุดท้ายทำเป็นสองขั้นตอน
>

````python
bangkok_count = respondent_counts["Bangkok"]

last_area = area_names[-1]
last_area_count = respondent_counts[last_area]

total_respondents = (
    respondent_counts["Bangkok"]
    + respondent_counts["Chiang Mai"]
    + respondent_counts["Khon Kaen"]
)

summary = (
    f"จำนวนผู้ตอบแบบสำรวจรวมเท่ากับ "
    f"{total_respondents} คน"
)

print("จำนวนผู้ตอบใน Bangkok:", bangkok_count)
print("พื้นที่สุดท้าย:", last_area)
print(
    "จำนวนผู้ตอบในพื้นที่สุดท้าย:",
    last_area_count,
)
print(summary)
````
