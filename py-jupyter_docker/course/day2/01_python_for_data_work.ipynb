## 4. Workflow พื้นฐานของงานข้อมูลด้วย Python

งานข้อมูลด้วย Python มักมีลำดับการทำงานคล้ายกัน ไม่ว่าจะเป็นงานของ Data Analyst, Data Engineer หรือ Data Scientist

โดยทั่วไป workflow พื้นฐานประกอบด้วย

1. Import library
2. Load data
3. Inspect data
4. Clean data
5. Transform data
6. Analyze data
7. Present or export results

## Workflow: จาก Excel สู่ Python

| ขั้นตอน | ถ้าทำใน Excel | ถ้าทำใน Python |
|---|---|---|
| Load data | เปิดไฟล์ Excel | ใช้คำสั่งอ่านไฟล์ |
| Inspect data | ดูแถว/คอลัมน์ด้วยสายตา | ใช้คำสั่งตรวจสอบโครงสร้างข้อมูล |
| Clean data | ลบแถว แก้ชื่อคอลัมน์ กรองข้อมูล | เขียนคำสั่ง clean ข้อมูล |
| Transform data | ใส่สูตร สร้างคอลัมน์ใหม่ | สร้างคอลัมน์ใหม่ด้วย code |
| Analyze data | ใช้ Filter, Sort, PivotTable | ใช้ pandas เช่น filter, groupby, pivot_table |
| Export result | Save As หรือ Copy ไปไฟล์ใหม่ | ใช้คำสั่ง export เป็น Excel/CSV |

## แบบฝึกหัดที่ 4: ทดลองปรับ workflow

ให้รันโค้ดตัวอย่างด้านล่าง แล้วทดลองปรับจากการกรอง `Bangkok` เป็น `Chiang Mai`

จากนั้นสังเกตว่าผลลัพธ์เปลี่ยนไปอย่างไร

````python
import pandas as pd

sales_data = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Keyboard", "Mouse"],
    "region": ["Bangkok", "Chiang Mai", "Bangkok", "Khon Kaen", "Bangkok"],
    "sales": [1200, 450, 5200, 1150, 500]
})

selected_region = "Bangkok"

filtered_sales = sales_data[sales_data["region"] == selected_region]

sales_by_product = filtered_sales.groupby("product")["sales"].sum()

sales_by_product
````

### คำถามหลังทดลอง

1. เมื่อเปลี่ยน `selected_region` จาก `"Bangkok"` เป็น `"Chiang Mai"` ผลลัพธ์เปลี่ยนอย่างไร
2. ถ้าทำงานนี้ใน Excel คุณต้องคลิกหรือเปลี่ยนอะไรบ้าง
3. ถ้าต้องทำรายงานแบบเดียวกันทุกเดือน Python workflow นี้ช่วยลดงานซ้ำได้อย่างไร
