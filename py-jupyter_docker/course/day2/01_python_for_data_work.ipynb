## 1. ทำไมคนทำงานข้อมูลจึงใช้ Python

Python เป็นเครื่องมือที่ได้รับความนิยมในงานข้อมูล เพราะมีจุดเด่นดังนี้

- รูปแบบคำสั่งอ่านค่อนข้างง่าย
- มีเครื่องมือสำหรับจัดการและวิเคราะห์ข้อมูลจำนวนมาก
- ใช้ทำงานซ้ำตามขั้นตอนเดิมได้
- ทำงานกับไฟล์หลายประเภทได้
- เชื่อมต่อกับฐานข้อมูลและ API ได้
- ใช้ได้ตั้งแต่งานวิเคราะห์ข้อมูลไปจนถึง Machine Learning
- สามารถพัฒนาเป็นกระบวนการอัตโนมัติได้

สำหรับผู้ที่คุ้นเคยกับ Excel อยู่แล้ว Python ไม่จำเป็นต้องเข้ามาแทน Excel ทั้งหมด แต่เหมาะกับงานที่ Excel เริ่มจัดการได้ยากหรือมีขั้นตอนซ้ำจำนวนมาก

### ตัวอย่างสถานการณ์ที่ Python ช่วยได้ดี

สมมติว่าเราได้รับไฟล์ข้อมูลรายเดือนจากหลายพื้นที่ และต้องทำขั้นตอนต่อไปนี้

1. เปิดไฟล์จากทุกพื้นที่
2. รวมข้อมูลให้อยู่ในตารางเดียวกัน
3. ตรวจสอบข้อมูลที่ไม่ครบถ้วน
4. ลบข้อมูลซ้ำ
5. สร้างคอลัมน์คำนวณ
6. สรุปข้อมูลตามจังหวัดหรือประเภทผู้รับบริการ
7. ส่งออกเป็นรายงานใหม่

หากทำด้วย Excel เราอาจต้องเปิดหลายไฟล์ คัดลอกข้อมูล วางสูตร สร้าง PivotTable และบันทึกรายงานใหม่ทุกเดือน

หากใช้ Python เราสามารถเขียนขั้นตอนไว้ครั้งเดียว แล้วนำกลับมาใช้กับข้อมูลของเดือนถัดไปได้

## 2. Excel และ Python เหมาะกับงานแบบใด

Excel และ Python ต่างเป็นเครื่องมือที่มีประโยชน์ แต่เหมาะกับลักษณะงานต่างกัน

Excel เหมาะกับงานที่ต้องเปิดดู แก้ไข หรือตรวจสอบข้อมูลด้วยสายตาอย่างรวดเร็ว

Python เหมาะกับงานที่มีหลายขั้นตอน ต้องทำซ้ำ ใช้ข้อมูลจำนวนมาก หรือต้องการบันทึกขั้นตอนอย่างชัดเจน

### เปรียบเทียบ Excel กับ Python ในงานข้อมูล

| ประเด็น | Excel | Python |
|---|---|---|
| การเปิดดูข้อมูลอย่างรวดเร็ว | สะดวกมาก | ดูได้ผ่านตารางหรือคำสั่ง |
| การแก้ไขข้อมูลด้วยมือ | ทำได้สะดวก | ไม่ใช่จุดเด่นหลัก |
| การทำงานซ้ำทุกเดือน | อาจต้องคัดลอกและทำขั้นตอนเดิม | รัน workflow เดิมกับข้อมูลใหม่ |
| การตรวจสอบย้อนหลัง | ตรวจจากสูตรและไฟล์ | ตรวจจากโค้ดตามลำดับ |
| การทำงานกับหลายไฟล์ | เริ่มซับซ้อนเมื่อไฟล์มีจำนวนมาก | เหมาะกับการประมวลผลหลายไฟล์ |
| การทำงานกับข้อมูลจำนวนมาก | มีข้อจำกัดตามขนาดและเครื่อง | รองรับและต่อยอดได้มากกว่า |
| การเชื่อมต่อฐานข้อมูลหรือ API | ทำได้ในบางกรณี | ทำได้ยืดหยุ่น |
| Machine Learning | ไม่ใช่จุดเด่นหลัก | มีเครื่องมือรองรับโดยตรง |

### ไม่จำเป็นต้องเลือกเพียงเครื่องมือเดียว

ในการทำงานจริง เราสามารถใช้ Excel และ Python ร่วมกันได้ เช่น

1. ใช้ Python รวมและทำความสะอาดข้อมูล
2. ส่งออกผลลัพธ์เป็นไฟล์ Excel
3. ใช้ Excel ตรวจสอบข้อมูลหรือจัดรูปแบบรายงาน
4. นำขั้นตอนที่ต้องทำซ้ำกลับไปทำใน Python

สิ่งสำคัญจึงไม่ใช่ว่าเครื่องมือใดดีกว่าเสมอไป แต่คือการเลือกเครื่องมือให้เหมาะกับงาน

## 3. Workflow พื้นฐานของงานข้อมูลด้วย Python

งานข้อมูลด้วย Python มักดำเนินไปตามลำดับต่อไปนี้

1. **Import libraries** — เตรียมเครื่องมือที่ต้องใช้
2. **Load data** — นำเข้าข้อมูล
3. **Inspect data** — ตรวจสอบโครงสร้างและคุณภาพเบื้องต้น
4. **Clean data** — แก้ไขข้อมูลที่ไม่พร้อมใช้งาน
5. **Transform data** — สร้างหรือเปลี่ยนรูปข้อมูล
6. **Analyze data** — คำนวณและสรุปผล
7. **Present or export results** — นำเสนอหรือส่งออกผลลัพธ์

ลำดับอาจปรับเปลี่ยนตามลักษณะงาน แต่ควรจัดขั้นตอนให้ชัดเจนและสามารถตรวจสอบได้

### เปรียบเทียบ Workflow ใน Excel กับ Python

| ขั้นตอน | ตัวอย่างใน Excel | ตัวอย่างใน Python |
|---|---|---|
| เตรียมเครื่องมือ | เปิดโปรแกรมหรือ Add-in | Import library |
| นำเข้าข้อมูล | เปิดไฟล์หรือ Import Data | ใช้คำสั่งอ่านไฟล์ |
| ตรวจสอบข้อมูล | เลื่อนดูตารางและใช้ Filter | ตรวจจำนวนแถว คอลัมน์ และชนิดข้อมูล |
| ทำความสะอาด | ลบแถว แก้ข้อความ หรือ Find & Replace | เขียนคำสั่งทำความสะอาด |
| แปลงข้อมูล | ใช้สูตรสร้างคอลัมน์ | สร้างคอลัมน์ด้วยโค้ด |
| วิเคราะห์ | ใช้สูตร Filter หรือ PivotTable | ใช้คำสั่งกรองและสรุปข้อมูล |
| ส่งออก | Save As หรือ Copy ไปไฟล์ใหม่ | ใช้คำสั่งส่งออก Excel หรือ CSV |

### ตัวอย่าง Workflow สำหรับข้อมูลแบบสำรวจ

สมมติว่าเราได้รับข้อมูลแบบสำรวจจากหลายพื้นที่

Workflow อาจประกอบด้วย

1. อ่านไฟล์ข้อมูลจากทุกพื้นที่
2. รวมข้อมูลเป็นตารางเดียว
3. ตรวจสอบจำนวนแถวและชื่อคอลัมน์
4. ตรวจสอบค่าว่างและข้อมูลซ้ำ
5. ทำให้ชื่อพื้นที่มีรูปแบบเดียวกัน
6. สรุปจำนวนผู้ตอบตามจังหวัด
7. ส่งออกตารางสรุปเป็นไฟล์รายงาน

เมื่อจัดขั้นตอนเป็น workflow เราจะเห็นได้ชัดว่าข้อมูลผ่านการเปลี่ยนแปลงใดบ้างก่อนกลายเป็นรายงาน

## 4. Reproducibility: การทำงานที่ตรวจสอบและทำซ้ำได้

**Reproducibility** ในบริบทของบทเรียนนี้ หมายถึง การจัดกระบวนการทำงานให้สามารถนำข้อมูลชุดเดิมหรือข้อมูลชุดใหม่ที่มีโครงสร้างเดียวกันมาประมวลผลซ้ำตามขั้นตอนเดิมได้

กระบวนการที่ดีควรตอบคำถามได้ว่า

- ข้อมูลมาจากที่ใด
- ตรวจสอบข้อมูลอย่างไร
- แก้ไขหรือลบข้อมูลใด
- ใช้สูตรหรือเงื่อนไขอะไร
- ได้ผลลัพธ์จากขั้นตอนใด
- หากได้รับข้อมูลเดือนใหม่ จะรันกระบวนการเดิมอย่างไร

### ปัญหาที่อาจเกิดขึ้นเมื่อทำงานด้วยมือ

สมมติว่าเราต้องสรุปรายงานด้วย Excel ทุกเดือน ปัญหาที่อาจเกิดขึ้น ได้แก่

- คัดลอกข้อมูลมาไม่ครบ
- วางข้อมูลผิดตำแหน่ง
- สูตรไม่ได้ลากลงมาครบทุกแถว
- ใช้ Filter ค้างไว้โดยไม่รู้ตัว
- PivotTable ไม่ได้ Refresh
- แก้ไขข้อมูลต้นฉบับโดยไม่บันทึกขั้นตอน
- มีไฟล์รายงานหลายเวอร์ชันจนไม่แน่ใจว่าไฟล์ใดล่าสุด
- ผู้ทำงานคนอื่นไม่ทราบว่ารายงานถูกคำนวณอย่างไร

Python ช่วยลดความเสี่ยงเหล่านี้ได้ เพราะขั้นตอนต่าง ๆ สามารถเขียนและเก็บไว้เป็นโค้ด

### ตัวอย่างแนวคิดการทำซ้ำ

สมมติว่าในแต่ละเดือนเราได้รับไฟล์ต่อไปนี้

```text
survey_2026_01.csv
survey_2026_02.csv
survey_2026_03.csv
```

หากไฟล์ทั้งสามมีโครงสร้างเดียวกัน เราสามารถใช้ workflow เดิมในการ

1. อ่านไฟล์
2. ตรวจสอบข้อมูล
3. กรองข้อมูล
4. สรุปผล
5. ส่งออกรายงาน

สิ่งที่เปลี่ยนอาจมีเพียงชื่อไฟล์หรือเดือนที่ต้องการวิเคราะห์ โดยไม่จำเป็นต้องสร้างขั้นตอนใหม่ทั้งหมด

## 5. Python ในบทบาทต่าง ๆ ของงานข้อมูล

Python ถูกนำไปใช้แตกต่างกันตามหน้าที่ของผู้ปฏิบัติงาน

บทบาทที่พบได้บ่อย ได้แก่

- Data Analyst
- Data Engineer
- Data Scientist

แต่ละบทบาทมีเป้าหมายต่างกัน แต่ใช้พื้นฐาน Python และการจัดการข้อมูลร่วมกัน

### Data Analyst

Data Analyst ใช้ Python เพื่อ

- อ่านข้อมูลจาก Excel, CSV หรือฐานข้อมูล
- ตรวจสอบคุณภาพข้อมูลเบื้องต้น
- ทำความสะอาดข้อมูล
- กรองข้อมูล
- สร้างคอลัมน์คำนวณ
- สรุปข้อมูลตามกลุ่ม
- สร้างกราฟ
- จัดทำตารางหรือรายงาน

Python ช่วยให้กระบวนการวิเคราะห์มีขั้นตอนชัดเจนและทำซ้ำได้

### Data Engineer

Data Engineer ใช้ Python เพื่อ

- รับข้อมูลจากหลายแหล่ง
- รวมข้อมูลจากหลายไฟล์
- เชื่อมต่อฐานข้อมูล
- ดึงข้อมูลจาก API
- ตรวจสอบและแปลงรูปแบบข้อมูล
- สร้าง Data Pipeline
- ทำงานแบบอัตโนมัติ
- จัดการ Error และบันทึก Log

Python ช่วยให้ข้อมูลเคลื่อนจากแหล่งต้นทางไปยังระบบปลายทางอย่างเป็นขั้นตอน

### Data Scientist

Data Scientist ใช้ Python เพื่อ

- สำรวจข้อมูลเชิงลึก
- วิเคราะห์ความสัมพันธ์ของตัวแปร
- เตรียมข้อมูลสำหรับสร้างแบบจำลอง
- สร้างตัวแปรหรือ Feature
- ฝึก Machine Learning Model
- ประเมินผล Model
- ทดลองสมมติฐาน
- อธิบายผลลัพธ์ด้วยตารางและกราฟ

Python ช่วยสนับสนุนกระบวนการทดลองและสร้างแบบจำลองจากข้อมูล

### ภาพรวมการใช้ Python ตามบทบาท

| บทบาท | เป้าหมายหลัก | ตัวอย่างงาน |
|---|---|---|
| Data Analyst | วิเคราะห์และสื่อสารผล | สรุปข้อมูล สร้างกราฟ จัดทำรายงาน |
| Data Engineer | จัดการการไหลของข้อมูล | รวมไฟล์ เชื่อมต่อระบบ สร้าง Pipeline |
| Data Scientist | ทดลองและสร้างแบบจำลอง | วิเคราะห์เชิงสถิติและ Machine Learning |

พื้นฐานที่ทั้งสามบทบาทใช้ร่วมกัน ได้แก่

- การเข้าใจชนิดและโครงสร้างข้อมูล
- การเขียนคำสั่งตามลำดับ
- การใช้เงื่อนไข
- การทำงานซ้ำ
- การสร้าง Function
- การจัดการข้อมูลตาราง

## 6. Python Libraries สำหรับงานข้อมูล

**Library** คือชุดเครื่องมือหรือคำสั่งที่ถูกพัฒนาไว้ให้เรานำมาใช้งาน โดยไม่ต้องเขียนทุกอย่างขึ้นใหม่ด้วยตนเอง

Python มี Library สำหรับงานข้อมูลหลายประเภท เช่น การจัดการตาราง การคำนวณ การสร้างกราฟ การเชื่อมต่อระบบ และ Machine Learning

### Library ที่พบได้บ่อยในงานข้อมูล

| Library | ตัวอย่างการใช้งาน |
|---|---|
| `pandas` | จัดการข้อมูลตารางจาก Excel และ CSV |
| `numpy` | คำนวณเชิงตัวเลขและจัดการ Array |
| `matplotlib` | สร้างกราฟพื้นฐาน |
| `seaborn` | สร้างกราฟเพื่อสำรวจข้อมูลเชิงสถิติ |
| `scikit-learn` | สร้างและประเมิน Machine Learning Model |
| `requests` | รับข้อมูลจาก Web API |
| `sqlalchemy` | เชื่อมต่อและทำงานกับฐานข้อมูล |

ในหลักสูตรนี้ เราจะเน้น Python พื้นฐานและ `pandas` สำหรับจัดการข้อมูลตาราง ส่วน Library อื่นจะเรียนรู้ตามความเหมาะสมของแต่ละหัวข้อ

### การนำ Library มาใช้

ก่อนใช้ Library เราต้องนำ Library นั้นเข้ามาในโปรแกรมด้วยคำสั่ง `import`

ตัวอย่างการนำ `pandas` มาใช้

````python
%pip install pandas
````

````python
import pandas as pd
````

คำสั่งด้านบนประกอบด้วย

- `import pandas` หมายถึง นำ Library ชื่อ `pandas` มาใช้งาน
- `as pd` หมายถึง กำหนดชื่อย่อของ `pandas` เป็น `pd`

การใช้ชื่อย่อช่วยให้เขียนคำสั่งได้สั้นลง เช่น

```python
pd.DataFrame(...)
```

### ขั้นตอนที่ 1: เตรียมและโหลดข้อมูลตัวอย่าง

ตัวอย่างนี้สร้างข้อมูลขึ้นภายใน Notebook เพื่อให้เห็นภาพรวมของ DataFrame

ยังไม่จำเป็นต้องเข้าใจรายละเอียดของทุกคำสั่ง เพราะการสร้างและจัดการ DataFrame จะเรียนอย่างละเอียดในบทถัดไป

````python
import pandas as pd

sales_data = pd.DataFrame(
    {
        "order_id": [1001, 1002, 1003, 1004, 1005],
        "product": [
            "Keyboard",
            "Mouse",
            "Monitor",
            "Keyboard",
            "Mouse",
        ],
        "region": [
            "Bangkok",
            "Chiang Mai",
            "Bangkok",
            "Khon Kaen",
            "Bangkok",
        ],
        "sales": [1200, 450, 5200, 1150, 500],
    }
)

sales_data
````

ผลลัพธ์แสดงข้อมูลในรูปตารางที่เรียกว่า **DataFrame**

DataFrame มีลักษณะคล้ายตารางใน Excel โดยประกอบด้วย

- แถว
- คอลัมน์
- ชื่อคอลัมน์
- ค่าของข้อมูล

### ขั้นตอนที่ 2: ตรวจสอบข้อมูล

ก่อนวิเคราะห์ข้อมูล เราควรตรวจสอบโครงสร้างและตัวอย่างข้อมูลก่อน

````python
sales_data.shape
````

`.shape` แสดงจำนวนแถวและจำนวนคอลัมน์ในรูปแบบ

```python
(number_of_rows, number_of_columns)
```

ผลลัพธ์ `(5, 4)` หมายความว่า ข้อมูลมี 5 แถวและ 4 คอลัมน์

````python
sales_data.head()
````

`.head()` ใช้ดูข้อมูลส่วนบนของตาราง

ในงานจริง ข้อมูลอาจมีหลายพันหรือหลายล้านแถว จึงควรดูตัวอย่างข้อมูลบางส่วนก่อนแทนการเปิดดูข้อมูลทั้งหมด

### ขั้นตอนที่ 3: กรองข้อมูล

สมมติว่าต้องการดูเฉพาะรายการขายใน Bangkok

การทำงานนี้คล้ายกับการใช้ Filter ใน Excel

````python
bangkok_sales = sales_data[
    sales_data["region"] == "Bangkok"
]

bangkok_sales
````

โค้ดด้านบนเลือกเฉพาะแถวที่คอลัมน์ `region` มีค่าเท่ากับ `"Bangkok"`

ในบทนี้ให้สังเกตแนวคิดก่อน ส่วนวิธีเขียนเงื่อนไขและกรอง DataFrame จะเรียนอย่างละเอียดภายหลัง

### ขั้นตอนที่ 4: สรุปข้อมูล

สมมติว่าต้องการสรุปยอดขายรวมตามสินค้า

การทำงานนี้มีแนวคิดคล้ายกับ PivotTable ใน Excel

````python
sales_by_product = (
    sales_data
    .groupby("product")["sales"]
    .sum()
)

sales_by_product
````

ผลลัพธ์แสดงยอดขายรวมของแต่ละสินค้า

สิ่งสำคัญในตอนนี้คือการเห็นว่า ขั้นตอนที่เคยทำผ่านเมนูหรือ PivotTable ใน Excel สามารถเขียนเป็นคำสั่งที่บันทึกและรันซ้ำได้

```python
import pandas as pd

survey_data = pd.DataFrame(
    {
        "respondent_id": [1, 2, 3, 4, 5],
        "area": [
            "Bangkok",
            "Chiang Mai",
            "Bangkok",
            "Khon Kaen",
            "Bangkok",
        ],
        "support_need": [
            "Income",
            "Housing",
            "Mental health",
            "Income",
            "Housing",
        ],
    }
)

bangkok_survey = survey_data[
    survey_data["area"] == "Bangkok"
]

need_summary = (
    bangkok_survey
    .groupby("support_need")["respondent_id"]
    .count()
)

need_summary
```

## แบบฝึกหัดที่ 4: ทดลองปรับ Workflow

ให้รันโค้ดด้านล่าง แล้วเปลี่ยนค่าของ `selected_area` จาก `"Bangkok"` เป็น `"Chiang Mai"`

จากนั้นรันโค้ดใหม่และสังเกตการเปลี่ยนแปลงของผลลัพธ์

````python
import pandas as pd

survey_data = pd.DataFrame(
    {
        "respondent_id": [1, 2, 3, 4, 5],
        "area": [
            "Bangkok",
            "Chiang Mai",
            "Bangkok",
            "Khon Kaen",
            "Bangkok",
        ],
        "support_need": [
            "Income",
            "Housing",
            "Mental health",
            "Income",
            "Housing",
        ],
    }
)

selected_area = "Bangkok"

filtered_survey = survey_data[
    survey_data["area"] == selected_area
]

need_summary = (
    filtered_survey
    .groupby("support_need")["respondent_id"]
    .count()
)

need_summary
````

````python
# เปลี่ยน selected_area แล้วรัน Workflow อีกครั้ง

selected_area = "Chiang Mai"

filtered_survey = survey_data[
    survey_data["area"] == selected_area
]

need_summary = (
    filtered_survey
    .groupby("support_need")["respondent_id"]
    .count()
)

need_summary
````
