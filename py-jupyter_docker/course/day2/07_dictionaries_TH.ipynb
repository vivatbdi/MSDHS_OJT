# 7) Dictionaries (พจนานุกรม) สำหรับงานวิเคราะห์ข้อมูลเบื้องต้น

จนถึงตอนนี้ เราเก็บข้อมูลแบบ “หนึ่งคอลัมน์” ด้วย list  
แต่ในงานจริง ข้อมูลหนึ่งแถว (record) มักมีหลายคอลัมน์ เช่น

- ชื่อ
- อายุ
- คะแนน
- สถานะผ่าน/ไม่ผ่าน

ใน Python เราใช้ **dictionary** เพื่อเก็บข้อมูลแบบ  
👉 **หนึ่งแถว = หนึ่ง record (หลายคอลัมน์)**

---

## สิ่งที่คุณจะได้จากบทนี้
- เข้าใจว่า dictionary คืออะไร และใช้แทนข้อมูล 1 แถวได้
- เปรียบเทียบ dictionary กับ “แถวข้อมูล” ใน Excel
- อ่านค่า/แก้ไขค่าใน dictionary
- ใช้ dictionary ร่วมกับ list เพื่อแทน “ทั้งตารางข้อมูล”
- เตรียมความพร้อมก่อนเข้าสู่ Pandas DataFrame

---

## ภาพรวมเนื้อหา (ลำดับการเรียน)
1. Dictionary เทียบกับ Excel (แถว + ชื่อคอลัมน์)
2. สร้าง dictionary และอ่านค่า
3. แก้ไข/เพิ่มข้อมูลใน dictionary
4. ใช้ dictionary กับ if-else
5. List of dictionaries = ตารางข้อมูล
6. แบบฝึกหัดจากสถานการณ์จริงของงาน data

## A. Dictionary เทียบกับ Excel

ใน Excel:
- 1 แถว = 1 record
- แต่ละคอลัมน์มีชื่อ เช่น `name`, `score`, `age`
- เราอ้างอิงค่าด้วย “ชื่อคอลัมน์ + แถว”

ใน Python:
- 1 dictionary = 1 record
- key = ชื่อคอลัมน์
- value = ค่าของคอลัมน์นั้น

ตัวอย่าง: ข้อมูลนักเรียน 1 คน

````python
student = {
    "name": "Alice",
    "age": 20,
    "score": 78
}

student
````

## B. อ่านค่าจาก dictionary (เหมือนดูค่าจากคอลัมน์)

การอ่านค่าจาก dictionary ใช้รูปแบบ:
```python
dictionary["key"]
```

✅ ลองรัน

````python
student['name']
````

````python
student['score']
````

## C. แก้ไขและเพิ่มข้อมูลใน dictionary

ถ้าข้อมูลเปลี่ยน หรือเราคำนวณค่าใหม่ได้  
เราสามารถแก้ไขหรือเพิ่ม key ใหม่ได้ทันที

ตัวอย่าง:
- แก้คะแนน
- เพิ่มสถานะผ่าน/ไม่ผ่าน

✅ ลองรัน

````python
student["score"] = 82
student
````

````python
student["passed"] = student["score"] >= 50
student
````

## D. ใช้ dictionary ร่วมกับ if-else

เราสามารถใช้ค่าจาก dictionary มาเป็นเงื่อนไขได้  
เหมือนดูค่าจากหลายคอลัมน์ใน Excel แล้วตัดสินใจ

✅ ลองรัน

````python
if student["score"] >= 50:
    result = "Pass"
else:
    result = "Fail"

result
````

## E. List of dictionaries = ตารางข้อมูล

ในงานจริง เราไม่ได้มีข้อมูลแค่ 1 แถว  
แต่มีหลาย record

แนวคิดสำคัญมาก:
👉 **ตารางข้อมูล = list ของ dictionaries**

ตัวอย่าง: นักเรียนหลายคน

````python
students = [
    {"name": "Alice", "score": 82},
    {"name": "Bob", "score": 45},
    {"name": "Charlie", "score": 91}
]

students
````

## F. วนผ่านตารางข้อมูล (list of dictionaries)

เราสามารถใช้ for-loop เพื่ออ่านข้อมูลทีละแถว  
เหมือนการดูข้อมูลทีละ row ใน Excel

✅ ลองรัน

````python
for s in students:
    print(s["name"], "score =", s["score"])
````

## G. วิเคราะห์ข้อมูลจากตาราง (if + loop + dictionary)

ตัวอย่างงานจริง:
- ตรวจว่าแต่ละคนผ่านหรือไม่
- เพิ่มข้อมูล `passed` เข้าไปในแต่ละ record

✅ ลองรัน

````python
for s in students:
    s["passed"] = s["score"] >= 50

students
````

## H. แบบฝึกหัด 1: ข้อมูลพนักงาน

คุณมีข้อมูลพนักงาน 1 คน:
- name
- salary

ให้:
1) สร้าง dictionary ชื่อ `employee`
2) เพิ่ม key `high_income` โดยให้เป็น True ถ้า salary >= 50000

````python
employees = [
    {"name": "Ann", "salary": 52000},
    {"name": "Ben", "salary": 45000},
    {"name": "Cat", "salary": 60000}
]

for e in employees:
    e["high_income"] = e["salary"] >= 50000
    print(e["name"], e["salary"], e["high_income"])
````
