## 2. List คืออะไร

**List** คือโครงสร้างข้อมูลที่ใช้เก็บค่าหลายค่าไว้ในลำดับเดียวกัน

ลักษณะสำคัญของ list คือ

- อยู่ในวงเล็บเหลี่ยม `[ ]`
- สมาชิกแต่ละตัวคั่นด้วย comma `,`
- ข้อมูลมีลำดับ
- สามารถมีข้อมูลซ้ำได้

````python
regions = ["Bangkok", "Chiang Mai", "Bangkok", "Khon Kaen"]

sales_amounts = [1200, 450, 5200, 1150]

is_online_order = [True, False, True, False]
````

## 3. การเข้าถึงข้อมูลใน List ด้วย Index

ข้อมูลใน list มีลำดับตำแหน่ง เรียกว่า **index**

ใน Python index เริ่มจาก `0`

ตัวอย่าง list:

```python
products = ["Keyboard", "Mouse", "Monitor", "Printer"]
```

ตำแหน่งของข้อมูลคือ

| ค่า | Index |
| --- | --- | 
| Keyboard | 0 |
| Mouse | 1 |
| Monitor | 2 |
| Printer | 3 |

````python
products = ["Keyboard", "Mouse", "Monitor", "Printer"]

products[0]
````

````python
products[1]
````

````python
products[3]
````

ข้อควรจำ:

> Python เริ่มนับ index จาก 0 ไม่ใช่ 1

เรื่องนี้เป็นจุดที่ผู้เริ่มต้นมักสับสน เพราะใน Excel เราคุ้นเคยกับแถวที่เริ่มจาก 1 เช่น row 1, row 2, row 3

แต่ใน Python ตำแหน่งแรกของ list คือ index 0

## การเข้าถึงข้อมูลจากท้าย List

Python สามารถใช้ index ติดลบเพื่อเข้าถึงข้อมูลจากท้าย list ได้

| Index | ความหมาย |
|---|---|
| `-1` | ตัวสุดท้าย |
| `-2` | ตัวรองสุดท้าย |
| `-3` | ตัวที่สามจากท้าย |

````python
products[-1]
````

````python
products[-2]
````

## การเลือกข้อมูลบางช่วงจาก List

เราสามารถเลือกข้อมูลบางช่วงจาก list ได้ เรียกว่า **slicing**

รูปแบบพื้นฐานคือ

```python
list_name[start:end]
```

โดย Python จะเลือกข้อมูลตั้งแต่ตำแหน่ง `start` ถึงก่อนตำแหน่ง `end`

````python
products = ["Keyboard", "Mouse", "Monitor", "Printer", "Webcam"]

products[0:3]
````

ผลลัพธ์คือข้อมูล index 0, 1 และ 2

แต่ไม่รวม index 3

นี่เป็นอีกจุดที่ควรระวัง เพราะ `end` ใน slicing จะไม่ถูกนับรวม

````python
products[:3]
````

````python
products[2:]
````

## 4. การเพิ่ม แก้ไข และลบข้อมูลใน List

List เป็นโครงสร้างข้อมูลที่แก้ไขได้

เราสามารถ

- เพิ่มข้อมูล
- แก้ไขข้อมูลเดิม
- ลบข้อมูล
- ตรวจสอบจำนวนสมาชิกใน list

### เพิ่มข้อมูลด้วย `.append()`

ใช้ `.append()` เพื่อเพิ่มข้อมูลใหม่ต่อท้าย list

````python
products = ["Keyboard", "Mouse", "Monitor"]

products.append("Printer")

products
````

### แก้ไขข้อมูลใน List

เราสามารถแก้ไขข้อมูลใน list โดยระบุ index ที่ต้องการแก้

````python
products[1] = "Wireless Mouse"

products
````

### ลบข้อมูลด้วย `.remove()`

ใช้ `.remove()` เพื่อลบค่าที่ระบุออกจาก list

````python
products.remove("Monitor")

products
````

### ตรวจสอบจำนวนสมาชิกด้วย `len()`

ใช้ `len()` เพื่อตรวจสอบจำนวนสมาชิกใน list

````python
len(products)
````

## 6. Dictionary คืออะไร

**Dictionary** คือโครงสร้างข้อมูลที่เก็บข้อมูลเป็นคู่ของ **key** และ **value**

รูปแบบพื้นฐานคือ

```python
{
    key: value
}
```

ตัวอย่างเช่น ต้องการเก็บข้อมูลสินค้า 1 รายการ

````python
product_info = {
    "product_id": "P001",
    "product_name": "Keyboard",
    "category": "Accessory",
    "price": 1200
}

product_info
````

จากตัวอย่างด้านบน

| Key | Value |
|---|---|
| product_id | P001 |
| product_name | Keyboard |
| category | Accessory |
| price | 1200 |

Dictionary เหมาะกับข้อมูลที่ต้องการระบุความหมายของแต่ละค่าอย่างชัดเจน

## 7. การเข้าถึงข้อมูลใน Dictionary ด้วย Key

ใน list เราใช้ index เพื่อเข้าถึงข้อมูล

แต่ใน dictionary เราใช้ key เพื่อเข้าถึง value

````python
product_info["product_name"]
````

````python
product_info["price"]
````

การใช้ key ทำให้โค้ดอ่านเข้าใจง่าย เพราะเราไม่ได้อ้างอิงด้วยตำแหน่ง แต่ใช้อ้างอิงด้วยชื่อความหมายของข้อมูล

## การใช้ `.get()`

นอกจากการใช้ `[key]` แล้ว เรายังสามารถใช้ `.get()` เพื่อดึงค่าจาก dictionary ได้

````python
product_info.get("category")
````

ข้อดีของ `.get()` คือ ถ้า key ไม่มีอยู่ใน dictionary จะไม่ error แต่จะได้ค่า `None`

````python
product_info.get("brand")
````

เราสามารถกำหนดค่า default ได้ด้วย

````python
product_info.get("brand", "Unknown")
````

## 8. การเพิ่ม แก้ไข และลบข้อมูลใน Dictionary

Dictionary สามารถแก้ไขได้เช่นเดียวกับ list

เราสามารถ

- เพิ่ม key-value ใหม่
- แก้ไข value ของ key เดิม
- ลบ key-value ที่ไม่ต้องการ

### เพิ่มข้อมูลใหม่

````python
product_info["brand"] = "LogiTech"

product_info
````

### แก้ไขข้อมูลเดิม

````python
product_info["price"] = 1350

product_info
````

### ลบข้อมูลด้วย `del`

````python
del product_info["brand"]

product_info
````

## ตรวจสอบ Key และ Value

เราสามารถดู key ทั้งหมด หรือ value ทั้งหมดใน dictionary ได้

````python
product_info.keys()
````

````python
product_info.values()
````

````python
product_info.items()
````

`items()` จะแสดงข้อมูลเป็นคู่ของ key และ value

สิ่งนี้มีประโยชน์มากเมื่อต้องการดูว่า dictionary เก็บข้อมูลอะไรไว้บ้าง

## 9. Dictionary ในบริบทงานข้อมูล

ในงานข้อมูล dictionary มักใช้ในสถานการณ์ต่อไปนี้

- mapping รหัสเป็นชื่อ
- mapping ค่าหมวดหมู่
- เปลี่ยนชื่อคอลัมน์
- เก็บ parameter สำหรับการวิเคราะห์
- เก็บข้อมูลสรุปแบบมีชื่อกำกับ

### ตัวอย่าง: Mapping รหัสสินค้าเป็นชื่อสินค้า

ใน Excel เราอาจมีตาราง lookup เช่น

| product_code | product_name |
|---|---|
| P001 | Keyboard |
| P002 | Mouse |
| P003 | Monitor |

ใน Python เราสามารถแทน lookup table แบบง่ายด้วย dictionary

````python
product_lookup = {
    "P001": "Keyboard",
    "P002": "Mouse",
    "P003": "Monitor"
}

product_lookup
````

````python
product_lookup["P002"]
````

แนวคิดนี้คล้ายกับการใช้ lookup table หรือ `VLOOKUP` ใน Excel

แต่ใน Python เราใช้ key เพื่อค้นหา value ที่ต้องการ

### ตัวอย่าง: Mapping รหัสภูมิภาคเป็นชื่อภูมิภาค

````python
region_mapping = {
    "BKK": "Bangkok",
    "CNX": "Chiang Mai",
    "KKN": "Khon Kaen",
    "HKT": "Phuket"
}

region_mapping["CNX"]
````

### ตัวอย่าง: เปลี่ยนชื่อคอลัมน์

ในงานข้อมูลจริง ไฟล์ Excel อาจมีชื่อคอลัมน์ที่อ่านยาก เช่น

| ชื่อคอลัมน์เดิม | ชื่อที่ต้องการ |
|---|---|
| Order ID | order_id |
| Product Name | product_name |
| Sales Amount | sales_amount |

เราสามารถเก็บ mapping นี้ไว้ใน dictionary ได้

````python
column_rename_mapping = {
    "Order ID": "order_id",
    "Product Name": "product_name",
    "Sales Amount": "sales_amount"
}

column_rename_mapping
````

Dictionary แบบนี้จะมีประโยชน์มากเมื่อใช้ pandas เพื่อเปลี่ยนชื่อคอลัมน์ให้เป็นมาตรฐานเดียวกัน

### ตัวอย่าง: เก็บ parameter สำหรับรายงาน

````python
report_config = {
    "report_month": "2026-05",
    "target_region": "Bangkok",
    "minimum_sales": 10000,
    "output_file": "monthly_sales_report.xlsx"
}

report_config
````

การเก็บค่า configuration ไว้ใน dictionary ช่วยให้แก้ไขเงื่อนไขรายงานได้ง่าย

เช่น หากต้องการเปลี่ยนเดือนรายงาน หรือเปลี่ยนภูมิภาคเป้าหมาย เราแก้ที่ dictionary เพียงจุดเดียว

````python
report_config["target_region"]
````

````python
report_config["output_file"]
````

## 10. เปรียบเทียบ List และ Dictionary

| ประเด็น | List | Dictionary |
|---|---|---|
| รูปแบบ | `[value1, value2, value3]` | `{key1: value1, key2: value2}` |
| ใช้อ้างอิงข้อมูลด้วย | index | key |
| เหมาะกับ | รายการข้อมูลที่มีลำดับ | ข้อมูลที่ต้องการชื่อกำกับ |
| ตัวอย่าง | รายชื่อไฟล์, รายชื่อคอลัมน์ | mapping, config, ข้อมูลสินค้า |
| คล้ายกับ Excel | range หรือ column of values | lookup table หรือ named fields |

## ควรใช้ List หรือ Dictionary

ใช้ **list** เมื่อ

- ต้องการเก็บรายการหลายค่า
- ลำดับของข้อมูลมีความหมาย
- ต้องการนำข้อมูลไปใช้ทีละรายการในภายหลัง
- เช่น รายชื่อไฟล์ รายชื่อคอลัมน์ รายการยอดขาย

ใช้ **dictionary** เมื่อ

- ต้องการเก็บข้อมูลเป็นคู่ key-value
- ต้องการค้นหาค่าจากชื่อหรือรหัส
- ต้องการ mapping ค่า
- ต้องการเก็บ configuration
- เช่น product code → product name, old column name → new column name

## 11. ข้อผิดพลาดที่พบบ่อย

### 1. ใช้ index ที่ไม่มีใน list

ถ้า list มีข้อมูล 3 ค่า index ที่ใช้ได้คือ 0, 1 และ 2

ถ้าเรียก index 3 จะเกิด error

````python
products = ["Keyboard", "Mouse", "Monitor"]

# ลองเอาเครื่องหมาย # ออกเพื่อดู error
products[3]
````

### 2. ลืมว่า index เริ่มจาก 0

ผู้เริ่มต้นมักคิดว่าข้อมูลตัวแรกอยู่ที่ตำแหน่ง 1  
แต่ใน Python ข้อมูลตัวแรกอยู่ที่ตำแหน่ง 0

````python
products[0]
````

### 3. ใช้ key ที่ไม่มีใน dictionary

ถ้าใช้ key ที่ไม่มีอยู่ จะเกิด error

````python
product_lookup = {
    "P001": "Keyboard",
    "P002": "Mouse"
}

# ลองเอาเครื่องหมาย # ออกเพื่อดู error
product_lookup["P999"]
````

ถ้าไม่แน่ใจว่า key มีอยู่หรือไม่ สามารถใช้ `.get()` ได้

````python
product_lookup.get("P999", "Unknown product")
````

### 4. สับสนระหว่าง list กับ dictionary

List ใช้วงเล็บเหลี่ยม `[ ]`

Dictionary ใช้วงเล็บปีกกา `{ }` และมีคู่ key-value

````python
product_list = ["Keyboard", "Mouse", "Monitor"]

product_dictionary = {
    "P001": "Keyboard",
    "P002": "Mouse",
    "P003": "Monitor"
}
````
