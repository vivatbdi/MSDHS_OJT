## A. Python และ Jupyter ทำงานเหมือน/ต่างจาก Excel อย่างไร?

หลายคนคุ้นเคยกับ **Excel** อยู่แล้ว ลองเทียบภาพรวมดังนี้

### Excel
- คุณทำงานใน **ตาราง (sheet)**
- ใส่สูตรใน **เซลล์ (cell)** เช่น `=AVERAGE(B2:B10)`
- ผลลัพธ์แสดงทันทีในตาราง

### Python + Jupyter
- คุณทำงานใน **สมุดโน้ต (notebook)**
- เขียนโค้ดใน **เซลล์ (cell)** เช่น `sum(...)`, `len(...)`, `print(...)`
- ผลลัพธ์แสดงใต้ cell ทันที

✅ จุดสำคัญ: ทั้งสองแบบ “ทดลอง → เห็นผล → แก้ไข” ได้เหมือนกัน  
ต่างกันคือ Python ทำให้เรา “ทำซ้ำ” และ “จัดการข้อมูล” ได้ยืดหยุ่นขึ้น

ถัดไปเราจะลองรัน cell จริง ๆ

## B. รู้จัก Cell และการรันโค้ด

ใน Jupyter จะมี cell หลัก ๆ 2 แบบ
- **Markdown cell**: สำหรับข้อความอธิบาย
- **Code cell**: สำหรับโค้ด Python

✅ วิธีรัน code cell  
- คลิกเลือก cell
- กด **Shift + Enter**

ลองรัน cell ด้านล่างนี้

```python
# Code cell ตัวอย่าง
print("ยินดีต้อนรับเข้าสู่บทเรียนการวิเคราะห์ข้อมูลเบื้องต้นโดยใช้ Python และ Jupyter")
```

## C. Python เป็นเครื่องคิดเลขสำหรับงานข้อมูล

ในงานวิเคราะห์ข้อมูล เราต้องคำนวณบ่อยมาก เช่น
- รวมค่า (sum)
- เฉลี่ย (average)
- เปรียบเทียบค่า

✅ ลองรันตัวอย่างต่อไปนี้

```python
100 + 250
```

```python
500 - 120
```

```python
45 * 3
```

```python
1000 / 4
```

### ตัวอย่างแบบ Excel → Python

สมมติใน Excel คุณอาจมีคะแนนรวม และจำนวนคน แล้วใช้สูตรหาเฉลี่ย  
ใน Python เราก็ทำได้คล้ายกัน โดยให้ชื่อกับค่าต่าง ๆ

✅ ลองรัน cell ด้านล่าง

```python
total_score = 305
number_of_students = 4

total_score / number_of_students
```

## D. ทำให้ผลลัพธ์อ่านง่ายด้วย `print()`

ใน Excel ผลลัพธ์แสดงใน cell  
ใน Jupyter เรามักใช้ `print()` เพื่ออธิบายผลลัพธ์ให้ชัดเจน

✅ ลองรันตัวอย่าง

```python
average = total_score / number_of_students
print("คะแนนเฉลี่ย =", average)
```

## E. Output คือสิ่งที่เราใช้ “อ่านข้อมูล”

เวลาเรารันโค้ดใน Jupyter  
- ผลลัพธ์จะแสดง **ใต้ cell** (เหมือนผลลัพธ์ที่แสดงใน cell ของ Excel)

ลองเทียบ 2 แบบนี้:

```python
average
```

```python
print(average)
```

## F. อ่าน error เบื้องต้น (เรื่องปกติของงานวิเคราะห์ข้อมูล)

บางครั้งเราจะพิมพ์ผิด หรือมีสัญลักษณ์ไม่ครบ แล้วเกิด error

✅ ลองรัน cell ด้านล่าง (มี error ตั้งใจใส่ไว้)

```python
print("การวิเคราะห์ข้อมูลที่ดีทำให้เข้าใจข้อมูล)
```

### วิธีอ่าน error แบบง่าย ๆ
1. ดู **บรรทัดสุดท้าย** ของ error เพื่อดู “ประเภทของปัญหา”
2. ดูว่า error ชี้ไปที่บรรทัดไหน
3. แก้ไข แล้วรันใหม่

✅ ลองแก้ไข cell ก่อนหน้า โดยเติมเครื่องหมาย `"` ที่หายไป แล้วรันอีกครั้ง

```python
print("การวิเคราะห์ข้อมูลที่ดีทำให้เข้าใจข้อมูล")
```

## G. การบันทึกและรีเซ็ตสภาพแวดล้อม

เมื่อทดลองหลายอย่างแล้ว คุณอาจอยากเริ่มใหม่ให้สะอาด  
เมนูที่ควรรู้:

- 💾 **Save**: File → Save (หรือ Ctrl / Cmd + S)
- 🔄 **Restart Kernel**: Kernel → Restart (ล้างค่าทั้งหมด)
- 🧹 **Restart & Clear Output**: Kernel → Restart & Clear Output

📌 ถ้า Restart แล้ว ต้องรัน cell ใหม่อีกครั้ง เพื่อให้ค่าต่าง ๆ กลับมา
