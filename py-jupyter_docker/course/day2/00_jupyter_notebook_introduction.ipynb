## 1. Jupyter Notebook คืออะไร

**Jupyter Notebook** คือเครื่องมือที่ช่วยให้เราเขียนและรันโค้ด Python แบบทีละส่วนได้

จุดเด่นของ Jupyter Notebook คือ เราสามารถรวมสิ่งต่อไปนี้ไว้ในที่เดียวกัน

- โค้ด Python
- ผลลัพธ์จากการรันโค้ด
- คำอธิบาย
- ตาราง
- กราฟ
- บันทึกการวิเคราะห์

ในงานวิเคราะห์ข้อมูล Notebook ช่วยให้เราทำงานอย่างเป็นขั้นตอน และสามารถย้อนกลับมาตรวจสอบได้ว่าเราใช้ข้อมูลอย่างไร คำนวณอย่างไร และได้ข้อสรุปอย่างไร

## เปรียบเทียบ Jupyter Notebook กับ Excel

| งานที่ทำ | Excel | Jupyter Notebook / Python |
|---|---|---|
| ใส่สูตรคำนวณ | พิมพ์สูตรใน Cell เช่น `=A1+B1` | เขียนคำสั่ง Python เช่น `100 + 200` |
| ดูผลลัพธ์ | แสดงผลใน Cell | แสดงผลใต้ Code Cell |
| จดคำอธิบาย | ใช้ Cell หรือ Comment | ใช้ Markdown Cell |
| ทำงานซ้ำ | Copy สูตร / Drag สูตร | เขียนโค้ดให้ทำซ้ำ |
| วิเคราะห์ข้อมูล | Filter, Sort, PivotTable | ใช้ Python และ pandas |
| ตรวจสอบขั้นตอนย้อนหลัง | ดูสูตรในแต่ละ Cell | อ่าน Notebook จากบนลงล่าง |

## 2. ส่วนประกอบพื้นฐานของ Notebook

Notebook ประกอบด้วยหน่วยย่อยที่เรียกว่า **Cell**

Cell คือพื้นที่แต่ละช่องใน Notebook ที่เราใช้เขียนเนื้อหา

Cell ที่ใช้บ่อยมี 2 ประเภทหลัก

1. **Code Cell**  
   ใช้เขียนและรันโค้ด Python

2. **Markdown Cell**  
   ใช้เขียนข้อความ คำอธิบาย หัวข้อ ตาราง หรือบันทึกประกอบการวิเคราะห์

## 3. Code Cell คืออะไร

**Code Cell** คือ Cell ที่ใช้เขียนคำสั่ง Python

เมื่อรัน Code Cell แล้ว ผลลัพธ์จะแสดงอยู่ด้านล่างของ Cell นั้น

````python
print("Hello, Jupyter Notebook")
````

จากตัวอย่างด้านบน

- `print()` คือคำสั่งที่ใช้แสดงผล
- ข้อความที่อยู่ในเครื่องหมาย `" "` คือข้อความที่ต้องการให้แสดงออกมา
- เมื่อรัน Cell แล้ว ผลลัพธ์จะแสดงด้านล่าง

````python
print("สวัสดีครับ/ค่ะ")
print("เรากำลังเริ่มเรียน Python สำหรับงานวิเคราะห์ข้อมูล")
````

## 4. Markdown Cell คืออะไร

**Markdown Cell** คือ Cell ที่ใช้เขียนข้อความอธิบาย ไม่ได้ใช้รันโค้ด Python

เราสามารถใช้ Markdown Cell เพื่อ

- เขียนหัวข้อ
- อธิบายขั้นตอนการวิเคราะห์
- บันทึกสมมติฐาน
- สรุปผลลัพธ์
- ทำตารางประกอบคำอธิบาย

ในการทำงานจริง Markdown Cell มีความสำคัญมาก เพราะช่วยให้ผู้อื่น รวมถึงตัวเราในอนาคต เข้าใจว่า Notebook นี้ทำอะไร

### ตัวอย่าง Markdown

เราสามารถเขียนรายการเป็น bullet ได้ เช่น

- ขั้นตอนที่ 1: โหลดข้อมูล
- ขั้นตอนที่ 2: ตรวจสอบข้อมูล
- ขั้นตอนที่ 3: วิเคราะห์ข้อมูล
- ขั้นตอนที่ 4: สรุปผล

หรือเขียนตัวหนา เช่น **ข้อความสำคัญ**

หรือเขียน code สั้น ๆ ในบรรทัดเดียว เช่น `print("Hello")`

## 5. วิธีการรัน Cell

เมื่อรัน Code Cell แล้ว Notebook จะแสดงผลลัพธ์อยู่ด้านล่าง Cell

ถ้าเป็น Markdown Cell เมื่อรันแล้ว ข้อความจะถูกจัดรูปแบบให้อ่านง่ายขึ้น

ใน Excel เราอาจเขียนว่า

```excel
=10+20
```

ใน Python เราเขียนว่า

```python
10 + 20
```

## 6. การใช้ Python แทนสูตรพื้นฐานใน Excel

ก่อนที่เราจะเรียน Python อย่างเป็นระบบในบทถัดไป เราจะทดลองใช้ Python ทำงานง่าย ๆ ที่คล้ายกับการใช้สูตรใน Excel

### ตัวอย่างที่ 1: การบวกเลข

ใน Excel:

```excel
=100+250
```

ใน Python:

````python
100 + 250
````

### ตัวอย่างที่ 2: การคำนวณยอดขายรวม

สมมติว่าเรามีข้อมูลดังนี้

| รายการ | จำนวน | ราคาต่อหน่วย |
|---|---:|---:|
| สินค้า A | 10 | 120 |
| สินค้า B | 5 | 300 |

ใน Excel เราอาจคำนวณยอดขายด้วยสูตร

```excel
=จำนวน*ราคาต่อหน่วย
```

ใน Python เราสามารถเขียนได้ดังนี้

````python
quantity_a = 10
price_a = 120

sales_a = quantity_a * price_a

sales_a
````

````python
quantity_b = 5
price_b = 300

sales_b = quantity_b * price_b

sales_b
````

````python
total_sales = sales_a + sales_b

total_sales
````

จากตัวอย่างนี้ เราเริ่มเห็นแนวคิดสำคัญของ Python คือ

- เราสามารถตั้งชื่อให้ข้อมูล เช่น `quantity_a`
- เราสามารถนำข้อมูลมาคำนวณได้
- เราสามารถเก็บผลลัพธ์ไว้ใช้ต่อได้ เช่น `sales_a`, `sales_b`, `total_sales`

แนวคิดนี้คล้ายกับการอ้างอิง Cell ใน Excel แต่ Python จะใช้ “ชื่อตัวแปร” แทนตำแหน่ง Cell

## 7. การทำงานกับข้อมูลหลายค่า

ใน Excel เรามักมีข้อมูลหลายแถว เช่น ยอดขายหลายรายการ

ใน Python เราสามารถเก็บข้อมูลหลายค่าไว้ในรูปแบบที่เรียกว่า **list**

ยังไม่ต้องกังวลเรื่องรายละเอียดของ list ในตอนนี้ เพราะเราจะเรียนอย่างเป็นระบบในบทถัดไป

ในบทนี้ให้เข้าใจก่อนว่า list คือการเก็บค่าหลายค่าไว้ด้วยกัน

````python
sales = [1200, 1500, 800, 2200, 1750]

sales
````

### การหาผลรวม

ใน Excel เราใช้

```excel
=SUM(A1:A5)
```

ใน Python เราใช้ sum()

````python
sum(sales)
````

### การหาจำนวนรายการ

ใน Excel เราอาจใช้

```excel
=COUNT(A1:A5)
```

ใน Python เราใช้ len()

````python
len(sales)
````

### การหาค่าเฉลี่ย

ใน Excel เราใช้

```excel
=AVERAGE(A1:A5)
```

ใน Python เราสามารถคำนวณจากผลรวม หารด้วยจำนวนรายการ

````python
average_sales = sum(sales) / len(sales)

average_sales
````

### การหาค่าสูงสุดและต่ำสุด

ใน Excel เราใช้

```excel
=MAX(A1:A5)
=MIN(A1:A5)
```

ใน Python เราใช้ max() และ min()

````python
max(sales)
````

````python
min(sales)
````
