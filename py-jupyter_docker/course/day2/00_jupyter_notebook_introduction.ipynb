## 1. Jupyter Notebook คืออะไร

**Jupyter Notebook** คือเครื่องมือที่ช่วยให้เราเขียนและรันโค้ด Python แบบทีละส่วนได้

ภายใน Notebook เราสามารถจัดเก็บสิ่งต่อไปนี้ไว้ด้วยกัน

- โค้ด Python
- ผลลัพธ์จากการรันโค้ด
- คำอธิบาย
- ตาราง
- กราฟ
- บันทึกและข้อสรุปจากการวิเคราะห์

ในงานวิเคราะห์ข้อมูล Notebook ช่วยให้เราทำงานอย่างเป็นลำดับ และย้อนกลับมาตรวจสอบได้ว่าเราใช้ข้อมูลอย่างไร คำนวณอย่างไร และได้ข้อสรุปอย่างไร

### เปรียบเทียบ Jupyter Notebook กับ Excel

| งานที่ทำ | Excel | Jupyter Notebook / Python |
|---|---|---|
| คำนวณข้อมูล | เขียนสูตรใน Cell เช่น `=A1+B1` | เขียนคำสั่ง เช่น `100 + 200` |

## 2. ส่วนประกอบของ Notebook

Notebook ประกอบด้วยช่องสำหรับใส่เนื้อหาที่เรียกว่า **Cell**

Cell ที่ใช้บ่อยมี 2 ประเภทหลัก

1. **Code Cell**  
   ใช้เขียนและรันคำสั่ง Python

2. **Markdown Cell**  
   ใช้เขียนหัวข้อ คำอธิบาย ตาราง หรือบันทึกผลการวิเคราะห์

## 3. วิธีการรัน Cell

การรัน Cell คือการสั่งให้ Notebook ประมวลผลเนื้อหาภายใน Cell นั้น

วิธีที่ใช้บ่อย ได้แก่

- คลิกปุ่ม **Run**
- กด **Shift + Enter** เพื่อรัน Cell และเลื่อนไปยัง Cell ถัดไป
- กด **Ctrl + Enter** เพื่อรัน Cell โดยยังเลือก Cell เดิมไว้

เมื่อรันแล้ว

- Code Cell จะแสดงผลลัพธ์ใต้ Cell
- Markdown Cell จะแสดงข้อความที่จัดรูปแบบแล้ว

## 4. การใช้งาน Code Cell

**Code Cell** ใช้สำหรับเขียนและรันคำสั่ง Python

ทดลองรัน Code Cell ต่อไปนี้

````python
print("Hello, Jupyter Notebook")
````

จากตัวอย่าง

- `print()` เป็นคำสั่งสำหรับแสดงผล
- ข้อความที่อยู่ภายในเครื่องหมาย `" "` คือข้อความที่ต้องการแสดง
- ผลลัพธ์จะแสดงอยู่ใต้ Code Cell หลังจากรัน

````python
print("สวัสดีครับ/ค่ะ")
print("เรากำลังเริ่มเรียน Python สำหรับงานวิเคราะห์ข้อมูล")
````

Code Cell หนึ่ง Cell สามารถมีคำสั่งได้มากกว่าหนึ่งบรรทัด

Python จะประมวลผลคำสั่งจากบรรทัดบนลงล่าง

## 5. การใช้งาน Markdown Cell

**Markdown Cell** ใช้เขียนข้อความประกอบการวิเคราะห์ โดยไม่ได้ใช้รันคำสั่ง Python

เราสามารถใช้ Markdown Cell เพื่อ

- เขียนชื่อเรื่องและหัวข้อ
- อธิบายขั้นตอนการทำงาน
- บันทึกสมมติฐาน
- สรุปผลการวิเคราะห์
- สร้างรายการและตาราง
- แสดงโค้ดตัวอย่างโดยไม่รันโค้ด

Markdown Cell ช่วยให้ผู้อื่น รวมถึงตัวเราในอนาคต เข้าใจว่า Notebook นี้ใช้ข้อมูลอะไร ทำอะไร และได้ผลลัพธ์อย่างไร

รายการแบบ bullet

## 6. การคำนวณพื้นฐานด้วย Python

เราสามารถใช้ Python เป็นเครื่องคิดเลขเพื่อคำนวณค่าพื้นฐานได้

ใน Excel เราอาจเขียน

```excel
=10+20
```

ใน Python เราเขียน


````python
10 + 20
````

### ตัวอย่างการคำนวณยอดขาย

สมมติว่ามีข้อมูลดังนี้

| รายการ | จำนวน | ราคาต่อหน่วย |
|---|---:|---:|
| สินค้า A | 10 | 120 |
| สินค้า B | 5 | 300 |

ยอดขายของแต่ละสินค้าคำนวณจาก

> จำนวน × ราคาต่อหน่วย

````python
quantity_a = 10
price_a = 120

sales_a = quantity_a * price_a

sales_a
````

````python
quantity_b = 5
price_b = 300

sales_b = quantity_b * price_b

sales_b
````

````python
total_sales = sales_a + sales_b

total_sales
````

จากตัวอย่าง เราเริ่มเห็นองค์ประกอบสำคัญของ Python ได้แก่

- การตั้งชื่อให้ข้อมูล เช่น `quantity_a`
- การนำข้อมูลไปคำนวณ
- การเก็บผลลัพธ์ไว้ใช้ต่อ เช่น `sales_a`
- การนำผลลัพธ์หลายค่ามาคำนวณร่วมกัน

ชื่อที่ใช้เก็บข้อมูลเรียกว่า **ตัวแปร** ซึ่งจะเรียนอย่างละเอียดในบทถัดไป

## 7. การคำนวณข้อมูลหลายค่า

ในงานข้อมูล เรามักต้องคำนวณค่าหลายรายการพร้อมกัน

Python สามารถเก็บค่าหลายค่าไว้ภายในเครื่องหมาย `[ ]` ได้ ดังตัวอย่างต่อไปนี้

````python
sales = [1200, 1500, 800, 2200, 1750]

sales
````

ข้อมูลที่อยู่ภายในเครื่องหมาย `[ ]` เรียกว่า **List**

ในบทนี้จะใช้ List เพื่อทดลองคำนวณข้อมูลหลายค่าเท่านั้น ส่วนรายละเอียดของ List จะเรียนในบทต่อไป

### การหาผลรวม

ใน Excel ใช้

```excel
=SUM(A1:A5)
```

ใน Python ใช้ `sum()`

````python
sum(sales)
````

### การนับจำนวนรายการ

ใน Excel อาจใช้

```excel
=COUNT(A1:A5)
```

ใน Python ใช้ `len()`

````python
len(sales)
````

### การหาค่าเฉลี่ย

ค่าเฉลี่ยคำนวณจาก

> ผลรวม ÷ จำนวนรายการ

````python
average_sales = sum(sales) / len(sales)

average_sales
````

### การหาค่าสูงสุดและค่าต่ำสุด

ใน Excel ใช้

```excel
=MAX(A1:A5)
=MIN(A1:A5)
```

ใน Python ใช้ `max()` และ `min()`

````python
max(sales)
````

````python
min(sales)
````

## 8. ลำดับการทำงานใน Notebook

Notebook ควรจัดเนื้อหาให้อ่านจากบนลงล่างตามลำดับต่อไปนี้

1. ระบุวัตถุประสงค์
2. เตรียมหรือกำหนดข้อมูล
3. เขียนโค้ดเพื่อคำนวณหรือวิเคราะห์
4. ตรวจสอบผลลัพธ์
5. เขียนข้อสังเกตหรือข้อสรุป

แนวทางนี้ทำให้ Notebook เป็นทั้งพื้นที่เขียนโค้ดและเอกสารบันทึกกระบวนการวิเคราะห์

### ลำดับการรัน Cell

ลำดับการรัน Cell มีผลต่อผลลัพธ์

ตัวอย่างเช่น ถ้า Cell หนึ่งต้องใช้ตัวแปร `total_sales` เราต้องรัน Cell ที่สร้าง `total_sales` ก่อน มิฉะนั้น Python จะยังไม่รู้จักตัวแปรดังกล่าว

หมายเลขที่แสดงด้านซ้ายของ Code Cell เช่น `[1]`, `[2]` หรือ `[3]` แสดงลำดับที่ Cell ถูกประมวลผล ไม่ใช่ตำแหน่งของ Cell ภายใน Notebook

### Kernel คืออะไร

**Kernel** คือส่วนที่ประมวลผลโค้ดและจดจำค่าต่าง ๆ ที่สร้างขึ้นระหว่างการทำงาน

ข้อควรทราบ

- เมื่อรัน Cell ตัวแปรจะถูกเก็บไว้ใน Kernel
- เมื่อแก้ไขข้อมูล ต้องรัน Cell ที่แก้ไขใหม่
- Cell อื่นที่อาศัยข้อมูลนั้นอาจต้องรันใหม่ด้วย
- เมื่อ Restart Kernel ค่าที่เคยเก็บไว้จะหายไป
- หลัง Restart Kernel ควรรัน Cell ใหม่ตั้งแต่ต้นตามลำดับ

การเลือกคำสั่ง **Restart Kernel and Run All Cells** ช่วยตรวจสอบว่า Notebook สามารถทำงานจากบนลงล่างได้โดยไม่อาศัยค่าที่ตกค้างจากการรันก่อนหน้า

## 9. การอ่าน Error เบื้องต้น

Error เป็นเรื่องปกติในการเขียน Python

เมื่อพบ Error ควรเริ่มจาก

1. อ่านข้อความบรรทัดสุดท้าย
2. ตรวจสอบชื่อคำสั่งหรือตัวแปร
3. ตรวจสอบว่าได้รัน Cell ที่จำเป็นแล้วหรือยัง
4. แก้ไขและทดลองรันใหม่

````python
# ทดลองรัน Cell นี้และสังเกต Error

total_revenue
````

Code ด้านบนจะเกิด `NameError` เพราะยังไม่มีการสร้างตัวแปรชื่อ `total_revenue`

Error ลักษณะนี้มักเกิดจาก

- พิมพ์ชื่อตัวแปรผิด
- ยังไม่ได้รัน Cell ที่สร้างตัวแปร
- Restart Kernel แล้วไม่ได้รัน Cell ก่อนหน้าใหม่

````python
# สร้างตัวแปรก่อนเรียกใช้งาน

total_revenue = 1000 + 2500 + 1800

total_revenue
````

หลังจากสร้างตัวแปร `total_revenue` แล้ว Python จึงสามารถแสดงค่าของตัวแปรได้

การพบ Error ไม่ได้หมายความว่าเราไม่สามารถเขียนโปรแกรมได้ แต่เป็นข้อมูลที่ช่วยให้เรารู้ว่าควรตรวจสอบหรือแก้ไขส่วนใด

## แบบฝึกหัดที่ 2: คำนวณคะแนนรวมจากแบบสำรวจ

มีข้อมูลคะแนนความต้องการความช่วยเหลือใน 3 ประเด็น โดยแต่ละประเด็นใช้คะแนนตั้งแต่ 1 ถึง 5

| ประเด็นที่สำรวจ | จำนวนผู้ตอบ | คะแนนเฉลี่ย |
|---|---:|---:|
| ความต้องการด้านรายได้ | 40 | 4.2 |
| ความต้องการด้านที่อยู่อาศัย | 35 | 3.8 |
| ความต้องการด้านสุขภาพจิต | 30 | 4.5 |

คะแนนรวมของแต่ละประเด็นคำนวณจาก

> จำนวนผู้ตอบ × คะแนนเฉลี่ย

ให้ใช้ Python คำนวณ

1. คะแนนรวมด้านรายได้
2. คะแนนรวมด้านที่อยู่อาศัย
3. คะแนนรวมด้านสุขภาพจิต
4. คะแนนรวมจากทั้ง 3 ประเด็น

````python
# ข้อมูลตั้งต้น

income_respondents = 40
income_average_score = 4.2

housing_respondents = 35
housing_average_score = 3.8

mental_health_respondents = 30
mental_health_average_score = 4.5
````

````python
# เขียนคำตอบของคุณใน Cell นี้
````

> ### เฉลยแบบฝึกหัดที่ 2
>
> คำนวณคะแนนรวมของแต่ละประเด็นก่อน แล้วจึงนำคะแนนทั้ง 3 ประเด็นมารวมกัน

````python
income_total_score = income_respondents * income_average_score
housing_total_score = housing_respondents * housing_average_score
mental_health_total_score = (
    mental_health_respondents * mental_health_average_score
)

overall_total_score = (
    income_total_score
    + housing_total_score
    + mental_health_total_score
)

print("คะแนนรวมด้านรายได้:", income_total_score)
print("คะแนนรวมด้านที่อยู่อาศัย:", housing_total_score)
print("คะแนนรวมด้านสุขภาพจิต:", mental_health_total_score)
print("คะแนนรวมทั้งหมด:", overall_total_score)
````
